```python
# 1. Bienvenida e Interacción Inicial
# Mensaje de bienvenida con decoración y emoji para hacer la consola amigable
print("============================================")  # línea decorativa superior
print("     🧮 Calculadora Financiera de Divisas")  # título con emoji
print("============================================")  # línea decorativa inferior
print("Hola —esta calculadora convierte COP a USD y EUR.")  # explicación breve del propósito

# 2. Entradas de Datos (Inputs & Casting)
# Usamos try-except para capturar errores de entrada (por ejemplo, letras donde esperan números)
try:
    # Pedimos el monto en Pesos Colombianos (COP)
    monto_cop_input = input("Ingrese el monto en Pesos Colombianos (COP): ")  # pedimos texto al usuario
    # Convertimos el texto ingresado a número decimal usando float()
    # Usamos float porque las finanzas necesitan decimales (centavos) y operaciones con punto flotante
    monto_cop = float(monto_cop_input)  # conversión segura a número decimal

    # Pedimos la TRM (tasa de cambio) del Dólar
    trm_dolar_input = input("Ingrese la TRM del Dólar (COP por USD): ")  # pedimos la TRM como texto
    trm_dolar = float(trm_dolar_input)  # convertimos a float por la misma razón anterior

    # Pedimos la tasa del Euro (COP por EUR)
    tasa_euro_input = input("Ingrese la tasa del Euro (COP por EUR): ")  # pedimos la tasa del Euro
    tasa_euro = float(tasa_euro_input)  # convertimos a float para cálculos precisos

# 3. Procesamiento (Cálculos)
    # Calculamos la conversión a Dólares aplicando la fórmula básica: monto / tasa
    monto_en_usd = monto_cop / trm_dolar  # división para convertir COP a USD

    # Calculamos la conversión a Euros aplicando la misma lógica
    monto_en_eur = monto_cop / tasa_euro  # división para convertir COP a EUR

    # Preparar cadenas que muestran la operación matemática realizada (ej: 10000 COP / 4000 COP/USD = 2.50 USD)
    # Creamos una cadena para la operación en USD usando f-string y formato :,.2f
    operation_usd = f"{monto_cop:,.2f} COP / {trm_dolar:,.2f} COP/USD = {monto_en_usd:,.2f} USD"  # operación matemática formateada
    # Creamos una cadena para la operación en EUR usando f-string y formato :,.2f
    operation_eur = f"{monto_cop:,.2f} COP / {tasa_euro:,.2f} COP/EUR = {monto_en_eur:,.2f} EUR"  # operación matemática formateada

    # 4. Salida de Información (Outputs)
    # Encabezado de resultados con decoración y emoji
    print("--------------------------------------------")  # separador
    print("Resultados de la conversión ✨")  # encabezado amigable
    print("--------------------------------------------")  # separador

    # Mostramos el monto original formateado con separadores de miles y dos decimales
    # Usamos f-string con :,.2f para formato thousands separator y dos decimales
    print(f"Monto original: COP {monto_cop:,.2f}")  # muestra monto en COP

    # Mostramos la operación y el resultado en dólares para mayor claridad
    print(f"Operación: {operation_usd} 💵")  # muestra la operación en USD con emoji

    # Mostramos la operación y el resultado en euros para mayor claridad
    print(f"Operación: {operation_eur} 💶")  # muestra la operación en EUR con emoji

    # Nota pedagógica simple para el usuario principiante
    print("--------------------------------------------")  # separador final
    print("Consejo: divide el monto COP entre la tasa para convertir.")  # explicación sencilla

# Manejo específico de errores: si el usuario ingresó texto no convertible a float
except ValueError:
    # Mensaje amistoso con emojis indicando el error y cómo corregirlo
    print("⚠️ Error: Por favor ingresa sólo números (ej. 150000, 4200.50). ❌")
```

# 📦 Tarea Clase 6: Módulo de Proveedores (POO Avanzada)

### Pseudocódigo / Lógica de Diseño:

1. **DEFINIR** la clase `Producto` (Simulación de la clase anterior):

   1.1. Atributos privados: `__nombre`, `__stock`.

   1.2. Métodos: `get_nombre()`, `get_stock()`, `aumentar_stock(cantidad)`.

2. **DEFINIR** la clase `Proveedor`:

   2.1. **Constructor (`__init__`)**:

        - Recibe: `nombre_empresa`, `nit`, `ciudad`.
        - Asigna a atributos PRIVADOS: `__nombre_empresa`, `__nit`, `__ciudad`.
   
   2.2. **Método `get_nit()`**:

        - Retorna el valor actual de `__nit`.
   
   2.3. **Método `set_nit(nuevo_nit)`**:

        - Validar que `nuevo_nit` no esté vacío.
        - Validar que `nuevo_nit` contenga solo números (formato válido).
        - SI es válido: Actualizar `__nit` y mostrar éxito.
        - SINO: Mostrar mensaje de error.

   2.4. **Método `suministrar_producto(producto, cantidad)`**:

        - Validar que la `cantidad` sea mayor a 0.
        - Llamar al método `aumentar_stock(cantidad)` del objeto `producto`.
        - Imprimir comprobante de entrega.

3. **PRUEBA DEL SISTEMA**:

   3.1. Crear un objeto `Producto` (Ej: Café).

   3.2. Crear un objeto `Proveedor` (Ej: Cooperativa Cafetera).

   3.3. Probar el método `set_nit` con un error intencional y luego con un acierto.

   3.4. Ejecutar `suministrar_producto` y verificar que el stock del producto aumentó.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# TEMA: POO Avanzada (Encapsulamiento e Interacción entre Objetos)
# ACTIVIDAD: Módulo de Proveedores
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. CLASE PRODUCTO (Simulación de la clase base para poder interactuar)
# ------------------------------------------------------------------------------
class Producto:
    def __init__(self, nombre, stock_inicial):
        
        self.__nombre = nombre
        self.__stock = stock_inicial
        
    def get_nombre(self):
        return self.__nombre
        
    def get_stock(self):
        return self.__stock
        
    # Método seguro para que el proveedor pueda aumentar el inventario
    def aumentar_stock(self, cantidad):
        
        if cantidad > 0:
            self.__stock += cantidad
            return True
        return False

# ------------------------------------------------------------------------------
# 2. CLASE PROVEEDOR (Desarrollo principal de la tarea)
# ------------------------------------------------------------------------------
class Proveedor:
    
    # CONSTRUCTOR: Inicializamos los datos del proveedor protegiéndolos con '__'
    def __init__(self, nombre_empresa, nit, ciudad):
        
        self.__nombre_empresa = nombre_empresa
        self.__nit = str(nit) # Nos aseguramos de que el NIT se maneje como texto
        self.__ciudad = ciudad

        print(f"🏢 Proveedor '{self.__nombre_empresa}' registrado exitosamente.")

    # ==========================================================================
    # MÉTODOS GET Y SET (ENCAPSULAMIENTO AVANZADO)
    # ==========================================================================
    
    # GETTER: Permite LEER el NIT de forma segura desde afuera de la clase
    def get_nit(self):
        return self.__nit
        
    # SETTER: Permite MODIFICAR el NIT, pero pasando por un filtro de seguridad (Sanity Check)
    
    def set_nit(self, nuevo_nit):
        
        # Convertimos a string y quitamos espacios en blanco
        nuevo_nit_limpio = str(nuevo_nit).strip()
        
        # VALIDACIÓN 1: Que no esté vacío
        
        if len(nuevo_nit_limpio) == 0:

            print("❌ Error: El NIT no puede estar vacío.")
            
        # VALIDACIÓN 2: Que contenga únicamente caracteres numéricos
        
        elif not nuevo_nit_limpio.isdigit():
            
            print("❌ Error: El NIT debe contener únicamente números (sin puntos, comas o letras).")
            
        # Si pasa los filtros, actualizamos el atributo privado
        else:
            self.__nit = nuevo_nit_limpio
            print(f"✅ NIT actualizado correctamente a: {self.__nit}")

    # ==========================================================================
    # INTERACCIÓN ENTRE OBJETOS
    # ==========================================================================
    
    # Este método recibe un OBJETO de tipo Producto y una cantidad numérica

    def suministrar_producto(self, producto, cantidad):
        
        # Validamos que no nos entreguen cantidades negativas o cero
        
        if cantidad <= 0:
            print("❌ Error: La cantidad a suministrar debe ser mayor a 0.")
            return # Usamos return para salir del método si la validación falla

        # Llamamos al método seguro del objeto 'producto' para aumentar su stock

        if producto.aumentar_stock(cantidad):

            print("\n" + "-"*50)
            print("📦 COMPROBANTE DE RECEPCIÓN DE MERCANCÍA")
            print("-"*50)
            
            print(f"Proveedor: {self.__nombre_empresa} (NIT: {self.__nit})")
            
            print(f"Producto Entregado: {producto.get_nombre()}")
            
            print(f"Cantidad Recibida: +{cantidad} unidades")
            
            print(f"Nuevo Stock Total: {producto.get_stock()} unidades")
            
            print("-"*50 + "\n")

        else:
            
            print("❌ Error interno al actualizar el inventario del producto.")
```

```python
# ==============================================================================
# 3. PRUEBA DEL SISTEMA (INSTANCIACIÓN Y EJECUCIÓN)
# ==============================================================================
if __name__ == "__main__":
    
    # 1. Instanciamos un producto de nuestra cafetería
    cafe_tostao = Producto(nombre="Café de Origen Tostao", stock_inicial=20)
    
    # 2. Instanciamos a nuestro proveedor
    proveedor_cafe = Proveedor(nombre_empresa="Cooperativa Cafetera Nacional", nit="900123456", ciudad="Bogotá")
    
    # 3. PRUEBA DEL SETTER (Seguridad de Datos)
    print("\n--- Pruebas de Seguridad (Setter) ---")
    
    # Intento fallido: Contiene letras
    proveedor_cafe.set_nit("900123ABC") 
    
    # Intento fallido: Está vacío
    proveedor_cafe.set_nit("   ")       
    
    # Intento exitoso: Formato correcto
    proveedor_cafe.set_nit("800987654") 
    
    # 4. PRUEBA DE INTERACCIÓN ENTRE OBJETOS
    
    print("\n--- Prueba de Suministro de Inventario ---")
    
    print(f"Stock inicial de {cafe_tostao.get_nombre()}: {cafe_tostao.get_stock()} unidades.")
    
    # El proveedor entrega 50 unidades del producto 'cafe_tostao'
    proveedor_cafe.suministrar_producto(producto=cafe_tostao, cantidad=50)
```

# CONECTAR DATOS SQL EN POWER BI

## MÉTODO 1: El Estándar de la Industria (Mediante Controlador ODBC)

Este es el método más profesional, ya que permite a Power BI ver todas las tablas y relaciones automáticamente, como si fuera un servidor SQL corporativo.

### Paso 1: Descargar e instalar el "Traductor" (Driver ODBC)

- Cierren Power BI si lo tienen abierto.

- Descarguen el controlador oficial y gratuito de SQLite para Windows desde: http://www.ch-werner.de/sqliteodbc/ (Aconsejo descargar el archivo sqliteodbc_w64.exe si su Windows es de 64 bits).

- Instálenlo dando clic en "Siguiente" (Next) con las opciones por defecto.

### Paso 2: Crear el Origen de Datos en Windows

- En el buscador de Windows (tecla Windows), escriban y abran la aplicación: "Orígenes de datos ODBC (64 bits)".

- Vayan a la pestaña "DSN de usuario" o "DSN de sistema" y hagan clic en el botón "Agregar...".

- Busquen en la lista "SQLite3 ODBC Driver", selecciónenlo y hagan clic en "Finalizar".

- Se abrirá una pequeña ventana. Llenen solo dos campos:

    Data Source Name: Pónganle un nombre sin espacios (Ej: Cafeteria_BD).

    Database Name: Hagan clic en el botón "Browse..." y busquen en su computador el archivo .db que generaron en VS Code.

- Hagan clic en "OK" y cierren la herramienta de Windows.

### Paso 3: Conectar en Power BI

- Abran Power BI Desktop.

- En la cinta de opciones de Inicio, hagan clic en "Obtener datos" -> "Más...".

- En el buscador escriban "ODBC" y selecciónenlo.

- En el menú desplegable "Nombre del origen de datos (DSN)", seleccionen el nombre que crearon en el paso anterior (Cafeteria_BD).

- Hagan clic en "Aceptar". (Si les pide credenciales, seleccionen "Windows" o "Predeterminado" sin usuario/contraseña, ya que SQLite es un archivo local).

### ¡Listo! Verán una ventana del "Navegador" con todas sus tablas. Seleccionen las que necesiten y hagan clic en "Cargar".

# MÉTODO 2: El Método del CTO (Usando un Script de Python)

## Paso 1: Validar Python en Power BI

- Abran Power BI Desktop.

- Vayan a Archivo -> Opciones y configuración -> Opciones.

- En el panel izquierdo, busquen "Creación de scripts de Python" y asegúrense de que Power BI haya detectado la ruta donde tienen instalado Python en su computador.

## Paso 2: Conectar mediante Código

- En Power BI, hagan clic en "Obtener datos" -> "Más...".

- Escriban "Python" en el buscador y seleccionen "Script de Python".

- Aparecerá un cuadro de texto en blanco. Peguen el siguiente código (asegurándose de cambiar la ruta por la ubicación real de su archivo .db):

```python
# Primero debemos tener instaladas las siguientes librerías:

%pip install pandas matplotlib
```

```python
import sqlite3
import pandas as pd

# 1. Definir la ruta EXACTA donde está el archivo .db en su computador.
# OJO: Usar doble barra invertida (\\) o barra normal (/) para evitar errores en Windows.
ruta_bd = r"C:\Users\diego\OneDrive\Documentos\Clase_Python_U_Sabana\2026-1\parcial_2_cafeteria.db"

# 2. Crear la conexión
conexion = sqlite3.connect(ruta_bd)

# 3. Usar Pandas para leer las tablas y convertirlas en DataFrames.
# (Deben hacer esto por cada tabla que quieran importar)
df_clientes = pd.read_sql_query("SELECT * FROM clientes", conexion)
df_productos = pd.read_sql_query("SELECT * FROM productos", conexion)
df_proveedor = pd.read_sql_query("SELECT * FROM proveedores", conexion)
df_ventas = pd.read_sql_query("SELECT * FROM ventas", conexion)

# 4. Cerrar la conexión
conexion.close()

# Power BI detectará automáticamente todas las variables que empiecen con 'df_' y las mostrará como tablas.
```

## PASO 3: Cargar.

- Hagan clic en "Aceptar".

- Power BI ejecutará el script y mostrará las tablas df_clientes, df_productos y df_ventas en la ventana del Navegador.

- Selecciónenlas y den clic en "Cargar".

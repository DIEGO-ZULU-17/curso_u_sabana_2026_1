## Descripción de la Tarea:

En este ejercicio aplicarán los conceptos de Casting, Input/Output y Operaciones Aritméticas para resolver un problema real de finanzas internacionales: la conversión de divisas.

Estructura del Archivo:

Requerimientos del Script:

    Solicitar al usuario un valor en Pesos Colombianos (COP).

    Solicitar la TRM del día (Tasa de Cambio).

    Calcular e imprimir el equivalente en Dólares (USD).

    Calcular e imprimir el equivalente en Euros (EUR). Nota: Consideren investigar la paridad EUR/USD actual o solicitar una segunda tasa de conversión.

## Criterios de Evaluación:

- Claridad en los mensajes de salida (print con formato).

```python
"""
Este script permite a un usuario convertir un monto específico de Pesos Colombianos (COP) 
a Dólares Estadounidenses (USD) y Euros (EUR), 
solicitando las tasas de cambio (TRM) del día en tiempo real.
"""

# --- 1. BIENVENIDA E INTERACCIÓN INICIAL ---
print("=== 🏦 CALCULADORA FINANCIERA DE DIVISAS EICEA 🏦 ===")
print("Herramienta para la conversión rápida de flujo de caja.\n")

# --- 2. ENTRADAS DE DATOS (INPUTS & CASTING) ---
# Solicitamos los datos al usuario.
# Usamos float() porque las divisas y tasas manejan decimales.
try:
    monto_cop = float(input(">> Ingrese el monto en Pesos Colombianos (COP) a convertir: "))
    trm_dolar = float(input(">> Ingrese la TRM del Dólar hoy (COP/USD): "))
    tasa_euro = float(input(">> Ingrese la tasa del Euro hoy (COP/EUR): "))

    # --- 3. PROCESAMIENTO (CÁLCULOS ARITMÉTICOS) ---
    # Aplicamos la lógica matemática: Divisa Extranjera = Moneda Local / Tasa de Cambio
    total_usd = monto_cop / trm_dolar
    total_eur = monto_cop / tasa_euro

    # --- 4. SALIDA DE INFORMACIÓN (OUTPUTS) ---
    # Usamos f-strings con formato {:,.2f} para poner separadores de miles y 2 decimales.
    # Esto garantiza la "Claridad en los mensajes".

    print("\n" + "="*40)
    print("       📊 REPORTE DE CONVERSIÓN       ")
    print("="*40)
    print(f"Monto Original:    $ {monto_cop:,.2f} COP")
    print("-" * 40)
    print(f"Estiamdo en Dólares: US$ {total_usd:,.2f}")
    print(f"Estimado en Euros:   €   {total_eur:,.2f}")
    print("="*40)
    print("Cálculo finalizado exitosamente.")

except ValueError:
    # (Opcional)} Manejo básico por si el usuario escribe texto en vez de números
    print("\n❌ ERROR: Por favor ingrese solo valores numéricos (sin letras ni símbolos).")
```

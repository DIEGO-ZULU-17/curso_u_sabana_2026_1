```python
"""El Bucle for (Iterador)

A diferencia del while, el for sabe exactamente cuántas veces va a ejecutarse."""

# Ejemplo (Finanzas - Proyección de Flujo de Caja):

flujos = [100, 200, 300, 400]
total = 0

for flujo in flujos:
    total += flujo
    print(f"Acumulado actual: {total}")
```

```python
""" La Función range()

Generando secuencias numéricas automáticamente.

Ideal para repetir una acción "n" veces o recorrer índices.

    range(5) -> 0, 1, 2, 3, 4 (El fin NO se incluye).

    range(1, 6) -> 1, 2, 3, 4, 5.

    range(0, 10, 2) -> 0, 2, 4, 6, 8 (Saltos de 2 en 2)."""


# Ejemplo (Amortización):

for mes in range(1, 13, 3):
    print(f"Calculando intereses del mes {mes}...")
```

```python
""" Bucles Anidados (Nested Loops)

Un ciclo dentro de otro. ¡Cuidado con la eficiencia!"""

# Ejemplo (Logística - Rutas de Distribución):

ciudades = ["Bogotá", "Medellín"]
vehiculos = ["Camión A", "Camión B"]

for ciudad in ciudades:
    for vehiculo in vehiculos:
        print(f"Enviando {vehiculo} a {ciudad}")
```

```python
# El orden de los bucles anidados afecta el resultado:

ciudades = ["Bogotá", "Medellín"]
vehiculos = ["Camión A", "Camión B"]

for vehiculo in vehiculos:
    for ciudad in ciudades:
        print(f"Enviando {vehiculo} a {ciudad}")
```

```python
"""Listas (La Navaja Suiza)

Colecciones ordenadas y mutables. Se usan para TODO.

Operaciones Clave:

    Crear: precios = [10.5, 20.0, 15.0]

    Agregar: precios.append(30.0) (Al final).

    Insertar: precios.insert(0, 5.0) (Al inicio).

    Eliminar: precios.remove(20.0) (Por valor).

    Ordenar: precios.sort() (De menor a mayor).

Tip: len(precios) nos dice cuántos datos tenemos."""

precios = [10.5, 20.0, 15.0]
print(precios) # [10.5, 20.0, 15.0]

precios.append(30.0) # (Al final).
print(precios) # [10.5, 20.0, 15.0, 30.0]

precios.insert(0, 5.0) # (Al inicio).
print(precios) # [5.0, 10.5, 20.0, 15.0, 30.0]

precios.remove(20.0) # (Por valor).
print(precios) # [5.0, 10.5, 15.0, 30.0]

precios.sort() # (De menor a mayor).
print(precios) # [5.0, 10.5, 15.0, 30.0]

len(precios) # Nos dice cuántos datos tenemos.
```

```python
""" Las tuplas (Tuples) son como listas, pero INMUTABLES. No se pueden modificar después de creadas.
Características:

    Se definen con paréntesis: trm_historica = (4100, 4050, 4200).

    No tienen métodos de modificación (append, remove, etc.).

    Son útiles para datos que no deben cambiar, como coordenadas o constantes.
"""

# Ejemplo Tupla:

trm_historica = (4100, 4050, 4200)

print(trm_historica) # (4100, 4050, 4200)

# trm_historica[0] = 4300  # <- ¡ERROR! No se puede cambiar.

```

```python
""" Diccionarios (Key-Value)

La estructura más importante para Data Science y JSON.

Permite acceder a los datos por un "nombre" (clave) y no por una posición."""

# Ejemplo (Perfil de Cliente):

cliente = {
    "nombre": "Juan Perez",
    "saldo": 5000000,
    "es_vip": True
}

print(cliente) # Imprime {'nombre': 'Juan Perez', 'saldo': 4900000, 'es_vip': True}

print(cliente["saldo"]) # Imprime 5000000

cliente["saldo"] -= 100000 # Modificar valor

print(cliente["saldo"]) # Imprime 4900000
```

## Taller de Repaso (Simulacro Parcial)

### Ejercicio: El Analista de Ventas

Tienen una lista de ventas diarias: ventas = [150, 300, 50, 120, 500, 25]

    Use un bucle for para calcular el Total Vendido.

    Use un condicional dentro del bucle para contar cuántas ventas fueron superiores a 100.

    Si encuentra una venta menor a 30, use continue para ignorarla en un reporte impreso.

    Al final, imprima el promedio de ventas.

```python
ventas = [150, 300, 50, 120, 500, 25]  # Lista de ventas diarias

total_vendido = 0  # Acumulador para el total vendido
contador_superiores_100 = 0  # Contador de ventas mayores a 100

for venta in ventas:  # Recorrer cada venta en la lista

    total_vendido += venta  # Sumar la venta al total vendido. total_vendido = total_vendido + venta

    if venta > 100:  # Si la venta es mayor a 100...
       contador_superiores_100 += 1  # ...aumentar el contador correspondiente

    if venta < 30:  # Si la venta es menor a 30...
        continue  # ...ignorarla en el reporte impreso y pasar a la siguiente

    print(f"Venta reportada: {venta}")  # Imprimir la venta en el reporte (si no fue ignorada)

promedio = total_vendido / len(ventas)  # Calcular el promedio de todas las ventas. len devuelve la cantidad de ventas en la lista.

print(f"\n Total vendido: {total_vendido}")  # Imprimir el total vendido
print(f"\n Cantidad de ventas > 100: {contador_superiores_100}")  # Imprimir cuántas ventas > 100
print(f"\n Promedio de ventas: {promedio:.2f}")  # Imprimir el promedio de ventas con dos decimales
```

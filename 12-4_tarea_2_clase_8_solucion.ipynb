```python
# SOLUCIÓN COMPLETA DE LA TAREA 2 - CLASE 8: LIMPIEZA Y NORMALIZACIÓN DE DATOS CON PANDAS

# Importamos la librería 'pandas' y le asignamos el alias estándar 'pd' para no escribir su nombre completo cada vez.
import pandas as pd

# Importamos el módulo 'datetime' para poder obtener la fecha actual del sistema automáticamente.
from datetime import datetime

# Importamos el módulo 'os' para interactuar con el sistema operativo y crear el archivo de prueba.
import os

# ==============================================================================
# 0. SETUP INICIAL: CREACIÓN DEL ARCHIVO SUCIO (Para que el código sea ejecutable)
# ==============================================================================

# Definimos el contenido del archivo CSV sucio exactamente como se entregó en el enunciado.
contenido_csv = """id_venta,cliente_info,producto_info,cantidad,precio_total,fecha
101,Diego Zuluaga - Profesor,Café Tostao - 5000,2,10000,2026-04-01
102,ana gomez - estudiante,Chocolatina Jet - 1200,NaN,6000,2026-04-02
103,Carlos Ruiz - PROFESOR,Empanada - $3000,3,9000,
104,MARIA LOPEZ - Estudiante,Jugo Hit - 2500,1,NaN,2026-04-03
105,Luis Perez - Externo,Café Tostao - 5000,4,20000,2026-04-04"""

# Abrimos (o creamos) el archivo 'ventas_sucias.csv' en modo escritura ('w') con codificación UTF-8.
with open('ventas_sucias.csv', 'w', encoding='utf-8') as archivo:
    # Escribimos el contenido de texto dentro del archivo físico.
    archivo.write(contenido_csv)

# Otra opción es cargar el archivo csv y guardarlo en un DataFrame para luego exportarlo, 
# pero el método directo con 'open' es más simple para este caso de setup.

# ==============================================================================
# 1. CARGA DE DATOS
# ==============================================================================

# Imprimimos un mensaje en consola para indicar el inicio del proceso.
print("🔄 Iniciando proceso de limpieza y normalización de datos 🔄")

# Utilizamos la función 'read_csv' de Pandas para cargar el archivo en la memoria RAM.
# El resultado se guarda en una variable llamada 'df' (Data Frame), que es una tabla bidimensional.
df = pd.read_csv('ventas_sucias.csv')

# Imprimimos el DataFrame original para ver su estado antes de la limpieza.
print("\n📊 DATAFRAME ORIGINAL (SUCIO):")
# Mostramos la tabla en la consola.
print(df)
```

```python
# ==============================================================================
# ANÁLISIS DE DATOS SUCIOS: Identificar anomalías y problemas de calidad
# ==============================================================================

# Recargamos el DataFrame original sucio para análisis
df_sucio = pd.read_csv('ventas_sucias.csv')

print("🔍 IDENTIFICACIÓN DE DATOS SUCIOS EN EL DATAFRAME ORIGINAL\n")

# 1. Mostrar el DataFrame sucio completo
print("📊 DATAFRAME ORIGINAL:")
print(df_sucio)
```

```python
# 2. Verificar tipos de datos
print("\n📋 TIPOS DE DATOS:")
print(df_sucio.dtypes)
```

```python
# 3. Identificar valores nulos (NaN)
print("\n❌ VALORES NULOS (NaN) POR COLUMNA:")
print(df_sucio.isnull().sum())
```

```python
# 4. Mostrar filas con valores nulos
print("\n⚠️ FILAS CON VALORES NULOS:")

# .any se usa para verificar si hay al menos un valor nulo en cada fila (axis=1).
print(df_sucio[df_sucio.isnull().any(axis=1)])
```

```python
# 5. Verificar inconsistencias en texto (mayúsculas/minúsculas)
print("\n🔤 INCONSISTENCIAS EN NOMBRES DE CLIENTES:")
print(df_sucio['cliente_info'])
```

```python
# 6. Identificar problemas de formato en precios (con símbolos $)
print("\n💵 PRECIOS CON SÍMBOLOS $ EN PRODUCTO_INFO:")
print(df_sucio['producto_info'])
```

```python
# 7. Verificar violaciones de 1NF (valores no atómicos)
print("\n🚨 VIOLACIONES DE 1NF (Valores no atómicos):")

print("   - cliente_info contiene: Nombre + Tipo Cliente (debe separarse)")
print("   - producto_info contiene: Nombre Producto + Precio (debe separarse)")
```

```python
# 8. Análisis de completitud de datos
print("\n📈 COMPLETITUD DE DATOS:")

# .notnull().sum() permite contar cuántos valores no nulos hay en cada columna, 
# luego se divide por el total de filas y se multiplica por 100 para obtener el porcentaje. 
# Finalmente, se redondea a 2 decimales.

completitud = (df_sucio.notnull().sum() / len(df_sucio) * 100).round(2)

print(completitud)
```

```python
# 9. Resumen de problemas identificados
print("\n📝 RESUMEN DE PROBLEMAS:")
print("   ✗ Valores NaN en 'cantidad'")
print("   ✗ Valores NaN en 'precio_total'")
print("   ✗ Valores NaN en 'fecha'")
print("   ✗ Nombres de clientes con inconsistencias de formato")
print("   ✗ Símbolos '$' en precios unitarios")
print("   ✗ cliente_info no es atómico (1NF)")
print("   ✗ producto_info no es atómico (1NF)")
```

```python
# ==============================================================================
# 2. NORMALIZACIÓN (1NF - Atomicidad)
# ==============================================================================
# La Primera Forma Normal (1NF) dicta que cada celda debe contener un solo valor.

# Tomamos la columna 'cliente_info' y usamos '.str.split()' para dividir el texto donde haya un guion (' - ').
# El parámetro 'expand=True' convierte el resultado en dos columnas separadas.
# Asignamos estas dos nuevas columnas a 'Nombre_Cliente' y 'Tipo_Cliente' en el DataFrame.
df[['Nombre_Cliente', 'Tipo_Cliente']] = df['cliente_info'].str.split(' - ', expand=True)

# Repetimos el mismo proceso para 'producto_info', separando el nombre del producto y su precio unitario.
# Asignamos el resultado a las nuevas columnas 'Nombre_Producto' y 'Precio_Unitario'.
df[['Nombre_Producto', 'Precio_Unitario']] = df['producto_info'].str.split(' - ', expand=True)

# Usamos el método '.drop()' para eliminar las columnas originales que ya no necesitamos.
# 'columns=[...]' indica qué columnas borrar. Reasignamos el resultado a 'df' para guardar los cambios.
df = df.drop(columns=['cliente_info', 'producto_info'])

# Imprimimos el DataFrame después de la normalización para verificar los cambios.
print("\n📊 DATAFRAME DESPUÉS DE NORMALIZAR (1NF):")
print(df)
```

```python
# ==============================================================================
# 3. ESTANDARIZACIÓN DE TEXTO Y NÚMEROS
# ==============================================================================

# Seleccionamos la columna 'Nombre_Cliente' y usamos '.str.title()' para poner la primera letra de cada palabra en mayúscula.
df['Nombre_Cliente'] = df['Nombre_Cliente'].str.title()

# Hacemos lo mismo con 'Tipo_Cliente' para estandarizar textos como "PROFESOR" o "estudiante".
df['Tipo_Cliente'] = df['Tipo_Cliente'].str.title()

# Seleccionamos la columna 'Precio_Unitario' y usamos '.str.replace()' para buscar el símbolo '$' y reemplazarlo por nada ('').
# 'regex=False' le dice a Pandas que busque el texto exacto y no una expresión regular.
df['Precio_Unitario'] = df['Precio_Unitario'].str.replace('$', '', regex=False)

# Convertimos la columna 'Precio_Unitario' (que actualmente es texto) a un formato numérico (float/int).
# 'pd.to_numeric()' es la función ideal de Pandas para hacer este "Casting" de forma segura.
df['Precio_Unitario'] = pd.to_numeric(df['Precio_Unitario'])

print("\n📊 DATAFRAME DESPUÉS DE ESTANDARIZAR TEXTO Y NÚMEROS:")
print(df)
```

```python
# ==============================================================================
# 4. LIMPIEZA DE NULOS (NaN)
# ==============================================================================

# Para rellenar la 'cantidad' vacía (NaN), usamos matemáticas: precio_total / Precio_Unitario.
# '.fillna()' busca los valores nulos y los reemplaza por el resultado de la operación matemática.
df['cantidad'] = df['cantidad'].fillna(df['precio_total'] / df['Precio_Unitario'])

# Hacemos un Sanity Check inverso: Si el 'precio_total' es el que está vacío (NaN), lo calculamos multiplicando.
# Multiplicamos la 'cantidad' por el 'Precio_Unitario' y rellenamos los vacíos.
df['precio_total'] = df['precio_total'].fillna(df['cantidad'] * df['Precio_Unitario'])

# Obtenemos la fecha actual del sistema usando 'datetime.now()'.
# '.strftime('%Y-%m-%d')' formatea la fecha para que quede como "Año-Mes-Día" (Ej: 2026-04-14).
fecha_actual = datetime.now().strftime('%Y-%m-%d')

# Seleccionamos la columna 'fecha' y usamos '.fillna()' para reemplazar los valores nulos por la fecha actual.
df['fecha'] = df['fecha'].fillna(fecha_actual)

print("\n📊 DATAFRAME DESPUÉS DE LIMPIAR NULOS:")
print(df)
```

```python
# ==============================================================================
# 5. REORDENAMIENTO Y EXPORTACIÓN
# ==============================================================================

# Opcional pero recomendado: Reordenamos las columnas para que el DataFrame sea más lógico al leerlo.
# Pasamos una lista con el orden exacto que deseamos.
columnas_ordenadas =['id_venta', 'fecha', 'Nombre_Cliente', 'Tipo_Cliente', 'Nombre_Producto', 'Precio_Unitario', 'cantidad', 'precio_total']

# Aplicamos el nuevo orden al DataFrame.
df = df[columnas_ordenadas]

# Imprimimos el DataFrame final para auditar que la limpieza fue exitosa.
print("\n✨ DATAFRAME LIMPIO Y NORMALIZADO (1NF):")
# Mostramos la tabla final en la consola.
print(df)

# Exportamos el DataFrame limpio a un nuevo archivo CSV usando '.to_csv()'.
# 'index=False' evita que Pandas guarde la columna de índices (0, 1, 2...) en el archivo final.
df.to_csv('ventas_limpias.csv', index=False)

# Imprimimos un mensaje de éxito indicando que el archivo fue creado.
print("\n💾 Archivo 'ventas_limpias.csv' exportado exitosamente y listo para Power BI.")
```

```python
import sqlite3

# ==============================================================================
# 6. EXPORTACIÓN A BASE DE DATOS SQLite
# ==============================================================================


# Creamos una conexión a la base de datos SQLite (se crea automáticamente si no existe)
conexion = sqlite3.connect('ventas_limpias.db')

# Exportamos el DataFrame limpio a una tabla SQL llamada 'ventas'
# if_exists='replace' sobrescribe la tabla si ya existe
# index=False evita guardar el índice de Pandas
df.to_sql('ventas', conexion, if_exists='replace', index=False)

# Confirmamos los cambios
conexion.commit()

# Cerramos la conexión
conexion.close()

print("✅ Base de datos 'ventas_limpias.db' creada exitosamente")
print("📂 Ahora puedes abrir el archivo con SQLITE EXPLORER")
```

```python
# ==============================================================================
# 7. ANÁLISIS EXPLORATORIO (Opcional para profundizar en el dataset limpio)
# ==============================================================================

df_limpio = pd.read_csv('ventas_limpias.csv')

# Mostramos información general del DataFrame limpio: tipos de datos, no nulos, memoria utilizada.
print("📋 INFORMACIÓN GENERAL DEL DATASET:")
df_limpio.info()
```

```python
# Mostramos estadísticas descriptivas: media, mediana, desviación estándar, mínimos y máximos.
print("\n📊 ESTADÍSTICAS DESCRIPTIVAS:")
print(df_limpio.describe())
```

```python
# Contamos cuántas ventas realizó cada cliente para identificar clientes frecuentes.
print("\n👥 CANTIDAD DE VENTAS POR CLIENTE:")
print(df_limpio['Nombre_Cliente'].value_counts())
```

```python
# Agrupamos por tipo de cliente y calculamos estadísticas de ventas por segmento.
print("\n🎓 RESUMEN DE VENTAS POR TIPO DE CLIENTE:")
print(df_limpio.groupby('Tipo_Cliente')[['cantidad', 'precio_total']].agg(['sum', 'mean', 'count']))
```

```python
# Identificamos el producto más vendido contando cuántas veces aparece en el dataset.
print("\n🏆 PRODUCTO MÁS VENDIDO:")
print(df_limpio['Nombre_Producto'].value_counts())
```

```python
# Calculamos el ingreso total por producto multiplicando cantidad por precio unitario.
df_limpio['ingreso_por_venta'] = df_limpio['cantidad'] * df_limpio['Precio_Unitario']

print("\n💰 INGRESOS TOTALES POR PRODUCTO:")

print(df_limpio.groupby('Nombre_Producto')['ingreso_por_venta'].sum().sort_values(ascending=False))
```

```python
# Identificamos la venta con mayor monto individual para análisis de casos extremos.
print("\n⭐ VENTA DE MAYOR VALOR:")

venta_maxima = df_limpio.loc[df_limpio['precio_total'].idxmax()]

print(venta_maxima)
```

```python
# Calculamos el ticket promedio (monto promedio por transacción) para medir rentabilidad.
print(f"\n🎟️ TICKET PROMEDIO: ${df_limpio['precio_total'].mean():.2f}")

# Verificamos si existen valores nulos residuales después de la limpieza.
print("\n🔍 VALIDACIÓN: VALORES NULOS RESTANTES:")
print(df_limpio.isnull().sum())
```

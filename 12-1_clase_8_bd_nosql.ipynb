# 📄 Módulo 1: Bases de Datos No-Relacionales (NoSQL con CSV)

# **Objetivo:** Implementar un CRUD completo leyendo y sobrescribiendo archivos planos.

```python
# csv nos sirve para almacenar datos de manera estructurada, similar a una tabla, lo que facilita su lectura y manipulación. 
# Es especialmente útil para exportar e importar datos entre diferentes sistemas y aplicaciones.

# os nos permite interactuar con el sistema operativo, como crear, leer, escribir y eliminar archivos. 
# En este caso, lo usaremos para manejar el archivo CSV donde se almacenarán los datos de los clientes.
import csv
import os

# 1. CLASE CLIENTE (POO)
class Cliente:
    def __init__(self, id_cliente, nombre, tipo):
        self.id_cliente = id_cliente
        self.nombre = nombre
        self.tipo = tipo

# 2. CONFIGURACIÓN DEL ARCHIVO
archivo_csv = "clientes_nosql.csv"
print(f"Archivo CSV configurado: {archivo_csv}")
```

```python
# 3. CREATE (Crear e Insertar)

def crear_cliente_csv(cliente):
    # os.path.isfile permite verificar si el archivo ya existe para decidir si se deben escribir los encabezados o no.
    # os significa sistema operativo, y path es la ruta del archivo. 
    # isfile verifica si el archivo existe en esa ruta.
    
    archivo_existe = os.path.isfile(archivo_csv)
    
    # 'a' (append) agrega al final del archivo
    with open(archivo_csv, mode='a', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        if not archivo_existe:
            writer.writerow(["ID", "Nombre", "Tipo"]) # Encabezados
            
        writer.writerow([cliente.id_cliente, cliente.nombre, cliente.tipo])
    print(f"✅ Cliente {cliente.nombre} guardado en CSV.")
```

```python
crear_cliente_csv(Cliente(1, "Juan Pérez", "Regular"))
crear_cliente_csv(Cliente(2, "María Gómez", "Premium"))
```

```python
# 4. READ (Leer)

def leer_clientes_csv():

    print("\n👥 BASE DE DATOS CLIENTES (NoSQL):")

    if not os.path.isfile(archivo_csv):

        print("El archivo no existe.")

        return

    # with open permite abrir el archivo y asegurarse de que se cierre automáticamente después de su uso.
    # mode='r' es de read e indica que se abrirá en modo lectura.
    # encoding='utf-8' asegura que se manejen correctamente los caracteres especiales.
    # csv.reader se utiliza para leer el contenido del archivo CSV, y luego se itera sobre cada fila para imprimirla.
    #      
    with open(archivo_csv, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        for fila in reader:
            print(f" | ".join(fila))
    print("-" * 40)
```

```python
leer_clientes_csv()
```

```python
# 5. UPDATE (Actualizar - Leer todo, modificar, sobrescribir)

def actualizar_tipo_cliente(id_buscar, nuevo_tipo):

    # filas_temporales se usa para almacenar temporalmente las filas del archivo CSV mientras se busca y actualiza el cliente.
    # actualizado es una bandera que indica si se encontró y actualizó el cliente.

    filas_temporales =[]
    actualizado = False
    
    # Leer a memoria
    with open(archivo_csv, mode='r', encoding='utf-8') as file:

        # csv.reader(file) se usa para leer el contenido del archivo CSV, 
        # y luego se itera sobre cada fila para buscar el cliente con el ID especificado.
        reader = csv.reader(file)

        # Usamos for para iterar sobre cada fila del archivo CSV. 
        # Si la primera columna (ID) coincide con el id_buscar, se actualiza el tipo en esa fila.
        
        # .append se usa para agregar cada fila (modificada o no) a la lista filas_temporales, 
        # que se usará para sobrescribir el archivo después de la actualización.

        for fila in reader:
            if fila[0] == str(id_buscar):
                fila[2] = nuevo_tipo # Modificamos el tipo
                actualizado = True
            filas_temporales.append(fila)
            
    # Sobrescribir ('w' de write, borra lo anterior)

    if actualizado:

        # newline se usa para evitar líneas en blanco adicionales al escribir en el archivo CSV, 

        with open(archivo_csv, mode='w', newline='', encoding='utf-8') as file:

            # .writerows se usa para escribir todas las filas almacenadas 
            # en filas_temporales de una sola vez, sobrescribiendo el contenido anterior del archivo CSV.

            writer = csv.writer(file)
            writer.writerows(filas_temporales)
            
        print(f"🔄 Cliente ID {id_buscar} actualizado.")
```

```python
actualizar_tipo_cliente(1, "VIP")

leer_clientes_csv()
```

```python
# 6. DELETE (Eliminar - Leer todo, omitir el borrado, sobrescribir)

def eliminar_cliente_csv(id_eliminar):

    filas_temporales =[]
    
    with open(archivo_csv, mode='r', encoding='utf-8') as file:

        reader = csv.reader(file)

        for fila in reader:

            if fila[0] != str(id_eliminar): # Guardamos todos EXCEPTO el eliminado
                filas_temporales.append(fila) # append para agregar a la lista filas_temporales las filas que no coinciden con el ID a eliminar.
                
    with open(archivo_csv, mode='w', newline='', encoding='utf-8') as file:

        writer = csv.writer(file)
        writer.writerows(filas_temporales)
        
    print(f"❌ Cliente ID {id_eliminar} eliminado.")
```

```python
eliminar_cliente_csv(2)

leer_clientes_csv()
```

```python
# --- PRUEBAS EN VIVO ---

# Limpiamos el archivo para la prueba

# os.path.isfile(archivo_csv) verifica si el archivo CSV ya existe. 
# Si existe, os.remove(archivo_csv) lo elimina para asegurarnos de que comenzamos con un archivo limpio para las pruebas.

# os.remove se usa para eliminar el archivo CSV si ya existe, lo que nos permite comenzar con un archivo limpio para las pruebas.

if os.path.isfile(archivo_csv): os.remove(archivo_csv)

c1 = Cliente(101, "Diego Zuluaga", "Profesor")
c2 = Cliente(102, "Ana Gómez", "Estudiante")

crear_cliente_csv(c1)
crear_cliente_csv(c2)
leer_clientes_csv()

actualizar_tipo_cliente(102, "Egresada")
leer_clientes_csv()

eliminar_cliente_csv(101)
leer_clientes_csv()
```

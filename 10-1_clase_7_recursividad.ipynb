# 🪆 Módulo 1: Recursividad en la Cafetería U. Sabana
**Objetivo:** Comprender cómo una función puede llamarse a sí misma para resolver problemas anidados.

### El Problema: "Las Anchetas de Regalo"
Para el Día de la Madre, la cafetería vende anchetas. Una ancheta puede contener productos individuales (Café, Chocolatina) o **incluso otras anchetas más pequeñas adentro**. 

¿Cómo calculamos el precio total de una lista que tiene listas adentro? Usando **Recursividad**.

**Reglas de la Recursividad:**
1. **Caso Base:** La condición que detiene el ciclo (Ej: Encontramos un producto final, no una caja).
2. **Paso Recursivo:** La función se llama a sí misma para abrir la siguiente caja.

```python
# ==============================================================================
# FUNCIONES RECURSIVAS
# ==============================================================================

# 1. Definimos una clase Producto simplificada para el ejercicio
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

# 2. LA FUNCIÓN RECURSIVA
def calcular_total_ancheta(item):
    """
    Calcula el precio total. Si encuentra una lista, se llama a sí misma.
    """
    # CASO BASE: Si el item es un Producto, simplemente devolvemos su precio.
    # isinstance() verifica si una variable es de un tipo de dato específico.
    if isinstance(item, Producto):
        return item.precio
    
    # PASO RECURSIVO: Si el item es una lista (una ancheta), la abrimos.
    elif isinstance(item, list):
        
        total_acumulado = 0

        for sub_item in item:
            # ¡LA MAGIA!: La función se llama a sí misma pasándole el contenido interno
            total_acumulado += calcular_total_ancheta(sub_item)
        return total_acumulado
    
    # Seguridad: Si no es ni producto ni lista, vale 0
    return 0
```

```python
# 3. PRUEBA DEL SISTEMA
# Creamos el inventario base
cafe = Producto("Café Tostao", 5000)
chocolatina = Producto("Chocolatina Jet", 1200)
galletas = Producto("Galletas Festival", 800)
jugo = Producto("Jugo Hit", 2500)

# Armamos una ancheta anidada (Listas dentro de listas)
ancheta_pequena =[chocolatina, galletas] # Valor: 2000
ancheta_premium =[cafe, jugo, ancheta_pequena] # Valor: 5000 + 2500 + 2000 = 9500

# Ejecutamos la recursividad
total_pagar = calcular_total_ancheta(ancheta_premium)

print("="*40)
print(f"🎁 Total a cobrar por la Ancheta Premium: ${total_pagar:,.2f} COP")
print("="*40)
```

## Pseudocódigo / Lógica de Diseño:

1. DEFINIR la clase `CarritoDeCompras`.

2. DEFINIR el constructor (estado inicial):

   2.1. Crear lista vacía `items` para guardar los productos.

   2.2. Crear variable numérica `total` inicializada en 0.

3. DEFINIR método `agregar_producto(nombre, precio)`:

   3.1. Agregar el producto y su precio a la lista `items`.

   3.2. Sumar el `precio` a la variable `total`.

   3.3. IMPRIMIR mensaje de confirmación de agregado.

4. DEFINIR método `ver_carrito()`:

   4.1. SI la lista `items` está vacía, IMPRIMIR advertencia.

   4.2. SINO, RECORRER la lista `items` e IMPRIMIR cada producto con su precio.

   4.3. IMPRIMIR el `total` acumulado a pagar.

5. PRUEBA DEL CÓDIGO:

   5.1. INSTANCIAR el objeto `mi_compra`.

   5.2. EJECUTAR `agregar_producto` 3 veces con datos diferentes.

   5.3. EJECUTAR `ver_carrito` para comprobar el resultado.
   

```python
# ==============================================================================
# TEMA: Programación Orientada a Objetos (POO)
# ACTIVIDAD: Tarea Clase 5: Modelando un Carrito de Compras (POO)
# ==============================================================================

# 1. DEFINICIÓN DE LA CLASE
# Creamos el "plano" o "molde" que definirá el comportamiento de cualquier carrito.

class CarritoDeCompras:
    
    # 2. EL CONSTRUCTOR (__init__)
    # Este método especial se ejecuta automáticamente al instanciar (crear) un objeto.
    # 'self' representa al objeto específico que se está creando en la memoria.

    def __init__(self):

        # Atributo 'items': Inicializamos una LISTA vacía propia del objeto.

        self.items =[]

        # Atributo 'total': Inicializamos el acumulador de dinero en 0.0 (float).

        self.total = 0.0
        
        print("🛒 Nuevo carrito de compras inicializado y listo para usar.")

    # 3. MÉTODO: AGREGAR PRODUCTO
    # Las funciones dentro de una clase se llaman 'métodos'. Siempre reciben 'self' como primer parámetro.

    def agregar_producto(self, nombre_producto, precio_producto):

        # Creamos un DICCIONARIO temporal con los datos del producto para mantener el orden.

        producto = {
            "nombre": nombre_producto,
            "precio": precio_producto
        }
        
        # Agregamos el diccionario a la lista 'items' del objeto actual (self).
        self.items.append(producto)
        
        # Sumamos el precio del nuevo producto al atributo 'total' del objeto actual.
        self.total += precio_producto
        
        # Imprimimos confirmación usando f-strings y formato de moneda.
        print(f"✅ Agregado: {nombre_producto} por ${precio_producto:,.2f} COP")

    # 4. MÉTODO: VER CARRITO
    # Este método lee el estado interno del objeto y lo muestra al usuario.

    def ver_carrito(self):

        print("\n" + "="*40)
        print("🧾 RESUMEN DE SU CARRITO DE COMPRAS")
        print("="*40)
        
        # Validación (Sanity Check): Verificamos si el carrito está vacío usando len()
        if len(self.items) == 0:
            
            print("Tu carrito está vacío. ¡Agrega algunos productos!")
        
        else:

            # Recorremos la lista de items usando un ciclo 'for'
            
            for item in self.items:
                
                # Accedemos a los valores del diccionario por su clave
                
                nombre = item["nombre"]
                precio = item["precio"]
                
                print(f"🛍️ {nombre.ljust(20)} : ${precio:,.2f} COP")  # nombre.ljust(20) justifica el nombre a la izquierda en un ancho de 20 caracteres para alinear la salida
            
            # Imprimimos el total acumulado leyendo el atributo self.total

            print("-" * 40)
            print(f"💰 TOTAL A PAGAR       : ${self.total:,.2f} COP")
        print("="*40 + "\n")


# ==============================================================================
# 5. PRUEBA DE LA CLASE (INSTANCIACIÓN Y EJECUCIÓN)
# ==============================================================================

# Instanciamos la clase: Creamos un objeto real llamado 'mi_compra'

mi_compra = CarritoDeCompras()

# Usamos la notación de punto (.) para acceder al método 'agregar_producto'
# Pasamos los argumentos requeridos: nombre y precio.

mi_compra.agregar_producto("Computador Portátil", 3500000)

mi_compra.agregar_producto("Mouse Inalámbrico", 85000)

mi_compra.agregar_producto("Monitor 24 Pulgadas", 650000)

# Usamos el método 'ver_carrito' para auditar que la lógica funcionó correctamente.
mi_compra.ver_carrito()
```

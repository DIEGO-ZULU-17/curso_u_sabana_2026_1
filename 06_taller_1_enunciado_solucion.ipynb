# 🚀 RETO 1: LOGÍSTICA - "Ruta-Óptima"

Perfil: Administración de Negocios Internacionales / Logística.

Problema: Una empresa de envíos en Bogotá necesita clasificar paquetes antes de cargarlos a los camiones.

Funcionalidad:

Pida el peso (kg) y destino de 5 paquetes.

Clasifique automáticamente: Si pesa < 1kg es "Documento", < 10kg es "Paquetería", > 10kg es "Carga".

Calcule el costo de envío (Tarifa base x Kg).

Imprima un manifiesto de carga ordenado por peso.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# RETO 1: LOGÍSTICA - "RutaÓptima"
# ==============================================================================

# 1. INICIALIZACIÓN DE VARIABLES GLOBALES
# Creamos una lista vacía que funcionará como nuestra "Base de Datos" temporal.
manifiesto_carga = []

# Definimos una tarifa base por kilogramo en COP (Ejemplo: $4,500 por Kg)
TARIFA_BASE_KG = 4500



print("🚛 BIENVENIDO AL SISTEMA DE LOGÍSTICA 'RUTA-ÓPTIMA' 🚛")

# 2. MENÚ INTERACTIVO (CICLO WHILE)
# El sistema se mantendrá en ejecución hasta que el usuario elija la opción "3" (Salir).
while True:
    print("\n" + "="*40)
    print("📋 MENÚ PRINCIPAL:")
    print("1. Registrar 5 paquetes")
    print("2. Generar y ver Manifiesto de Carga")
    print("3. Salir del sistema")
    print("="*40)
    
    # Capturamos la decisión del usuario
    opcion = input(">> Seleccione una opción (1, 2 o 3): ")
    
    # --- OPCIÓN 1: INGRESO DE DATOS Y CLASIFICACIÓN ---
    if opcion == "1":
        print("\n📦 MÓDULO DE REGISTRO DE PAQUETES")
        
        # Utilizamos un ciclo 'for' para pedir exactamente 5 paquetes
        for i in range(1, 6):
            print(f"\n--- Ingresando Paquete #{i} ---")
            
            # INPUT Y CASTING: Solicitamos destino (texto) y peso (decimal)
            destino = input("Destino del paquete (Ej: Medellín): ").strip()
            peso = float(input("Peso del paquete en KG: "))
            
            # LÓGICA DE NEGOCIO (IF / ELIF / ELSE): Clasificación del paquete
            if peso < 1:
                categoria = "Documento"
            elif peso < 10:
                categoria = "Paquetería"
            else:
                categoria = "Carga"
            
            # CÁLCULO ARITMÉTICO: Costo de envío
            costo_envio = peso * TARIFA_BASE_KG
            
            # MANEJO DE LISTAS: Guardamos los datos del paquete como una sub-lista.
            # Nota: Ponemos el 'peso' de primero para que .sort() ordene basándose en este valor numérico.
            paquete_actual = [peso, destino, categoria, costo_envio]
            
            # Añadimos el paquete a nuestra lista principal (Base de Datos)
            manifiesto_carga.append(paquete_actual)
            
        print("\n✅ ¡Los 5 paquetes han sido registrados exitosamente!")

    # --- OPCIÓN 2: PROCESAMIENTO Y REPORTE ---
    elif opcion == "2":
        # Validación de seguridad: Comprobamos si hay paquetes registrados
        if len(manifiesto_carga) == 0:
            print("\n⚠️ Advertencia: Aún no hay paquetes registrados. Vaya a la opción 1.")
            continue # Vuelve al inicio del menú
        
        # ORDENAMIENTO DE DATOS:
        # El método .sort() ordena la lista de menor a mayor.
        # Como nuestras sub-listas empiezan por el 'peso', se ordenarán por peso automáticamente.
        manifiesto_carga.sort()
        
        print("\n" + "*"*50)
        print("📄 MANIFIESTO DE CARGA (ORDENADO POR PESO)")
        print("*"*50)
        
        # RECORRIDO DE DATOS (CICLO FOR): Imprimimos el reporte formateado
        total_recaudado = 0
        
        # Desempaquetamos cada sub-lista en sus respectivas variables
        for paquete in manifiesto_carga:
            peso_p = paquete[0]
            destino_p = paquete[1]
            categoria_p = paquete[2]
            costo_p = paquete[3]
            
            # Sumamos el costo al total general de la ruta
            total_recaudado += costo_p
            
            # Usamos f-strings para mostrar la información tabulada y organizada
            print(f"📦 Destino: {destino_p.upper()} | Peso: {peso_p}kg | Tipo: {categoria_p} | Flete: ${costo_p:,.2f}")
            
        print("-" * 50)
        print(f"💰 PROYECCIÓN TOTAL DEL FLETE: ${total_recaudado:,.2f} COP")
        print("*"*50)

    # --- OPCIÓN 3: SALIR DEL SISTEMA ---
    elif opcion == "3":
        print("\nCerrando sistema Ruta-Óptima... ¡Buen viaje! 🚛💨")
        break # Rompe el ciclo 'while' y finaliza el programa

    # --- VALIDACIÓN DE ENTRADA INCORRECTA ---
    else:
        print("\n❌ Error: Opción no válida. Por favor digite 1, 2 o 3.")
```

Puntos clave (Autogestión y Sanity Check):

    El orden en las sublistas: Se debe resaltar que en la línea paquete_actual = [peso, destino, categoria, costo_envio], el peso se colocó intencionalmente en la primera posición (índice 0). Esto se hace porque la función manifiesto_carga.sort() ordena las listas basándose siempre en el primer elemento.

    Ciclo while True: Es la estructura estándar para construir aplicaciones de consola (Menús). Solo se detiene gracias a la instrucción break en la Opción 3.

    Manejo de Errores Lógicos: La condición if len(manifiesto_carga) == 0: previene que el programa intente imprimir un manifiesto vacío, lo que demuestra un pensamiento computacional maduro.

# 🥗 RETO 2: GASTRONOMÍA - "Chef-Costos"

Perfil: Gastronomía.

Problema: Un restaurante necesita controlar el costo de sus platos para no perder dinero por la inflación de alimentos en Colombia.

Funcionalidad:

Permita ingresar 5 ingredientes para una receta (Nombre y Precio de Mercado).

Valide si el precio subió más del 20% respecto a un precio base (decisión lógica).

Calcule el costo total del plato y sugiera un Precio de Venta (Costo + 35% ganancia).

Muestre la lista de ingredientes ordenada del más caro al más barato.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# RETO 2: GASTRONOMÍA - "Chef-Costos" (Control de Inflación)
# ==============================================================================

# 1. INICIALIZACIÓN DE VARIABLES
# Creamos una lista vacía que funcionará como la base de datos de nuestra receta.
ingredientes_receta = []

# Definimos el margen de ganancia esperado para el restaurante (35%).
MARGEN_GANANCIA = 0.35

print("👨‍🍳 BIENVENIDO AL SISTEMA DE COSTEO 'CHEF-COSTOS' 👩‍🍳")

# 2. MENÚ INTERACTIVO (CICLO WHILE)
# El menú seguirá activo hasta que el usuario decida salir (Opción 3).
while True:
    print("\n" + "="*45)
    print("🥘 MENÚ PRINCIPAL:")
    print("1. Ingresar 5 ingredientes de la receta")
    print("2. Analizar Costos y Sugerir Precio de Venta")
    print("3. Salir del sistema")
    print("="*45)
    
    # Capturamos la opción ingresada por el usuario
    opcion = input(">> Seleccione una opción (1, 2 o 3): ")
    
    # --- OPCIÓN 1: INGRESO DE DATOS ---
    if opcion == "1":
        print("\n🛒 MÓDULO DE INGRESO DE INGREDIENTES")
        
        # Usamos un ciclo 'for' para pedir exactamente los 5 ingredientes solicitados.
        for i in range(1, 6):
            print(f"\n--- Ingrediente #{i} ---")
            
            # INPUT Y CASTING: Nombre (String) y Precios (Float)
            nombre = input("Nombre del ingrediente: ").strip().capitalize()
            precio_base = float(input("Precio histórico/base (COP): "))
            precio_mercado = float(input("Precio actual en el mercado (COP): "))
            
            # LÓGICA ARITMÉTICA: Calculamos el porcentaje de variación del precio.
            # Fórmula: ((Precio Nuevo - Precio Viejo) / Precio Viejo) * 100
            variacion_porcentual = ((precio_mercado - precio_base) / precio_base) * 100
            
            # MANEJO DE LISTAS: Guardamos los datos en una sub-lista.
            # OJO: Ponemos 'precio_mercado' de primero para que .sort() ordene basándose en este valor.
            ingrediente_actual = [precio_mercado, nombre, variacion_porcentual]
            
            # Añadimos el ingrediente a la lista principal
            ingredientes_receta.append(ingrediente_actual)
            
        print("\n✅ ¡Los 5 ingredientes han sido registrados exitosamente!")

    # --- OPCIÓN 2: PROCESAMIENTO, ALERTAS Y REPORTE ---
    elif opcion == "2":
        # Validación de seguridad (Sanity Check): Evitamos que el código falle si la lista está vacía.
        if len(ingredientes_receta) == 0:
            print("\n⚠️ Advertencia: No hay ingredientes registrados. Ejecute la opción 1 primero.")
            continue # Salta el resto del código y vuelve a mostrar el menú
        
        # ORDENAMIENTO DE DATOS:
        # El método .sort() ordena de menor a mayor.
        # Al agregar 'reverse=True', le decimos que ordene de MAYOR a MENOR (Del más caro al más barato).
        ingredientes_receta.sort(reverse=True)
        
        print("\n" + "*"*55)
        print("📊 REPORTE DE COSTOS E INFLACIÓN (De mayor a menor)")
        print("*"*55)
        
        costo_total_plato = 0
        
        # RECORRIDO DE DATOS (CICLO FOR): Analizamos ingrediente por ingrediente.
        for ingrediente in ingredientes_receta:
            precio_m = ingrediente[0]
            nombre_i = ingrediente[1]
            inflacion = ingrediente[2]
            
            # Sumamos el precio actual al costo total del plato
            costo_total_plato += precio_m
            
            # LÓGICA DE NEGOCIO (IF/ELSE): Validamos si la inflación supera el 20%
            if inflacion > 20:
                alerta = f"⚠️ ALERTA: Subió {inflacion:.1f}%"
            else:
                alerta = f"✅ Estable ({inflacion:.1f}%)"
            
            # Imprimimos la fila del ingrediente con formato
            print(f"🥦 {nombre_i}: ${precio_m:,.2f} COP | {alerta}")
            
        # CÁLCULO DEL PRECIO SUGERIDO: Costo + 35%
        ganancia_esperada = costo_total_plato * MARGEN_GANANCIA
        precio_venta_sugerido = costo_total_plato + ganancia_esperada
        
        print("-" * 55)
        print(f"📉 COSTO TOTAL DEL PLATO:   ${costo_total_plato:,.2f} COP")
        print(f"📈 PRECIO VENTA SUGERIDO:   ${precio_venta_sugerido:,.2f} COP (Margen 35%)")
        print("*"*55)

    # --- OPCIÓN 3: SALIR ---
    elif opcion == "3":
        print("\nCerrando sistema Chef-Costos... ¡Éxitos en el servicio! 🍽️")
        break # Rompe el ciclo 'while True' y termina el script

    # --- VALIDACIÓN DE ERRORES ---
    else:
        print("\n❌ Error: Opción no válida. Por favor digite 1, 2 o 3.")
```

Puntos clave:

    El uso de reverse=True: En la línea ingredientes_receta.sort(reverse=True), es fundamental que por defecto .sort() ordena de forma ascendente. El requerimiento exige mostrar la lista "del más caro al más barato", por lo que esta pequeña modificación en el argumento soluciona el problema elegantemente.

    Cálculo de Inflación: La variación porcentual ((precio_mercado - precio_base) / precio_base) * 100 es una fórmula económica real. Este tipo de lógicas demuestra cómo Python no solo automatiza, sino que permite aplicar conceptos teóricos de Administración y Economía en tiempo real.

    El Sanity Check: Si un mesero o administrador presiona la opción 2 antes de la 1, el if len(ingredientes_receta) == 0: salva al programa de un colapso. Esto enseña a anticiparse a los errores del usuario.

#  RETO 3: ECONOMÍA - "Dolar-Track"

Perfil: Economía y Finanzas.

Problema: Un inversionista necesita analizar el comportamiento de la TRM (Tasa Representativa del Mercado) de la última semana para decidir si comprar dólares.

Funcionalidad:

Ingrese la TRM de los últimos 7 días.

Calcule el promedio de la semana.

Use un for para identificar cuántos días la TRM estuvo por encima del promedio (Alerta de Venta) y cuántos por debajo (Alerta de Compra).

Determine la volatilidad (Diferencia entre el valor máximo y mínimo).

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# RETO 3: ECONOMÍA - "Dolar-Track" (Análisis de TRM y Volatilidad)
# ==============================================================================

# 1. INICIALIZACIÓN DE VARIABLES GLOBALES
# Creamos una lista vacía para almacenar la TRM de los últimos 7 días.
historial_trm = []

print("📈 BIENVENIDO AL SISTEMA DE ANÁLISIS CAMBIARIO 'DOLAR-TRACK' 📉")

# 2. MENÚ INTERACTIVO (CICLO WHILE)
# El menú seguirá activo hasta que el usuario decida salir (Opción 3).
while True:
    print("\n" + "="*50)
    print("📊 MENÚ PRINCIPAL:")
    print("1. Ingresar TRM de los últimos 7 días")
    print("2. Generar Análisis de Mercado y Volatilidad")
    print("3. Salir del sistema")
    print("="*50)
    
    # Capturamos la decisión del usuario
    opcion = input(">> Seleccione una opción (1, 2 o 3): ")
    
    # --- OPCIÓN 1: INGRESO DE DATOS ---
    if opcion == "1":
        print("\n💵 MÓDULO DE REGISTRO DE TRM")
        
        # Limpiamos la lista por si el usuario decide volver a ingresar los datos
        historial_trm.clear()
        
        # Usamos un ciclo 'for' para pedir exactamente los 7 días de la semana
        for dia in range(1, 8):
            # INPUT Y CASTING: Solicitamos la tasa y la convertimos a float (decimal)
            trm_diaria = float(input(f"Ingrese la TRM del Día {dia} (COP): "))
            
            # MANEJO DE LISTAS: Agregamos el valor a nuestra lista histórica
            historial_trm.append(trm_diaria)
            
        print("\n✅ ¡Los 7 registros de TRM han sido guardados exitosamente!")

    # --- OPCIÓN 2: PROCESAMIENTO, LÓGICA Y REPORTE ---
    elif opcion == "2":
        # Validación de seguridad (Sanity Check): Comprobamos que existan datos
        if len(historial_trm) == 0:
            print("\n⚠️ Advertencia: No hay datos registrados. Ejecute la opción 1 primero.")
            continue # Salta el resto del código y vuelve a mostrar el menú
        
        # CÁLCULO 1: Promedio de la semana
        # sum() suma todos los elementos de la lista y len() cuenta cuántos hay
        suma_trm = sum(historial_trm)
        dias_totales = len(historial_trm)
        promedio_semana = suma_trm / dias_totales
        
        # Contadores para las alertas de mercado
        alertas_venta = 0
        alertas_compra = 0
        
        # RECORRIDO DE DATOS (CICLO FOR) Y LÓGICA DE NEGOCIO (IF/ELIF):
        for trm in historial_trm:
            # Si la TRM es mayor al promedio, es un buen momento para vender dólares
            if trm > promedio_semana:
                alertas_venta += 1
            # Si la TRM es menor al promedio, el dólar está barato, es buen momento de comprar
            elif trm < promedio_semana:
                alertas_compra += 1
                
        # ORDENAMIENTO DE DATOS Y VOLATILIDAD:
        # Creamos una copia de la lista para no alterar el orden histórico original de los días.
        trm_ordenada = historial_trm.copy()
        
        # El método .sort() ordena de menor a mayor.
        trm_ordenada.sort()
        
        # Al estar ordenada, el valor mínimo queda en la primera posición [0] 
        # y el máximo en la última [-1].
        trm_minima = trm_ordenada[0]
        trm_maxima = trm_ordenada[-1]
        
        # CÁLCULO 2: Volatilidad (Diferencia entre el pico más alto y el más bajo)
        volatilidad = trm_maxima - trm_minima
        
        # IMPRESIÓN DEL REPORTE (OUTPUT):
        print("\n" + "*"*50)
        print("📈 REPORTE DE MERCADO CAMBIARIO (SEMANAL)")
        print("*"*50)
        print(f"💵 Promedio de la semana:  ${promedio_semana:,.2f} COP")
        print(f"🔴 Días Alerta de Venta (TRM > Promedio):  {alertas_venta} días")
        print(f"🟢 Días Alerta de Compra (TRM < Promedio): {alertas_compra} días")
        print("-" * 50)
        print(f"📈 TRM Más Alta: ${trm_maxima:,.2f} COP")
        print(f"📉 TRM Más Baja: ${trm_minima:,.2f} COP")
        print(f"⚡ VOLATILIDAD (Spread): ${volatilidad:,.2f} COP")
        print("*"*50)

    # --- OPCIÓN 3: SALIR ---
    elif opcion == "3":
        print("\nCerrando sistema Dolar-Track... ¡Exitosas inversiones! 💼")
        break # Rompe el ciclo 'while True' y termina el script

    # --- VALIDACIÓN DE ERRORES ---
    else:
        print("\n❌ Error: Opción no válida. Por favor digite 1, 2 o 3.")
```

Puntos clave:

    El uso de sum() y len(): En lugar de hacer un for adicional solo para sumar los valores (lo cual es válido pero más largo), en Python se puede aprovechar la función integrada sum(lista). Esto fomenta escribir código más "Pythónico" y eficiente.

    El método .copy(): Al llegar a la parte del ordenamiento, es vital hacer trm_ordenada = historial_trm.copy(). Si solo hicieran historial_trm.sort(), destruirían el orden cronológico original de los días (Día 1, Día 2...). Esto enseña a proteger la integridad de las series de tiempo.

    Indexación Extrema ([0] y [-1]): Una vez la lista está ordenada con .sort(), el valor mínimo por lógica siempre será el índice 0 y el valor máximo siempre será el índice -1 (que en Python significa "el último elemento"). Esta es la forma más elegante de encontrar máximos y mínimos si se requería usar .sort() obligatoriamente por rúbrica.

# 🛍️ RETO 4: MERCADEO - "Fideliza-Puntos"

Perfil: Mercadeo y Administración.

Problema: Una tienda de ropa quiere lanzar un programa de lealtad pero no sabe quiénes son sus mejores clientes.

Funcionalidad:

Registre el nombre y el valor total de compra de 5 clientes.

Asigne una categoría: "Bronce" (< $100k), "Plata" (< $500k), "Oro" (> $500k).

Si es "Oro", calcule un bono de regalo del 10% de su compra.

Imprima el reporte de clientes ordenado alfabéticamente para enviar correos.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# RETO 4: MERCADEO - "Fideliza-Puntos" (Programa de Lealtad)
# ==============================================================================

# 1. INICIALIZACIÓN DE VARIABLES GLOBALES
# Creamos una lista vacía que funcionará como el CRM (Customer Relationship Management) del programa.
base_clientes = []

print("🛍️ BIENVENIDO AL SISTEMA DE LEALTAD 'FIDELIZA-PUNTOS' 🛍️")

# 2. MENÚ INTERACTIVO (CICLO WHILE)
# El sistema se mantendrá en ejecución constante hasta que el usuario decida salir (Opción 3).
while True:
    print("\n" + "="*45)
    print("🎯 MENÚ PRINCIPAL:")
    print("1. Registrar 5 clientes y sus compras")
    print("2. Generar Reporte de Lealtad (Alfabético)")
    print("3. Salir del sistema")
    print("="*45)
    
    # Capturamos la decisión del usuario
    opcion = input(">> Seleccione una opción (1, 2 o 3): ")
    
    # --- OPCIÓN 1: INGRESO DE DATOS Y SEGMENTACIÓN ---
    if opcion == "1":
        print("\n👥 MÓDULO DE REGISTRO DE CLIENTES")
        
        # Limpiamos la base de datos por si el usuario repite el proceso de registro
        base_clientes.clear()
        
        # Utilizamos un ciclo 'for' para pedir exactamente los 5 clientes solicitados
        for i in range(1, 6):
            print(f"\n--- Cliente #{i} ---")
            
            # INPUT Y CASTING: Nombre (String) y Valor de Compra (Float)
            # Usamos .title() para asegurar que la primera letra del nombre esté en mayúscula (útil para ordenar)
            nombre_cliente = input("Nombre completo del cliente: ").strip().title()
            valor_compra = float(input("Valor total de sus compras (COP): "))
            
            # LÓGICA DE NEGOCIO (IF / ELIF / ELSE): Asignación de categoría y bono
            bono_regalo = 0  # Inicializamos el bono en cero por defecto
            
            if valor_compra < 100000:
                categoria = "Bronce"
                
            elif valor_compra < 500000:
                categoria = "Plata"
                
            else: # Compras mayores o iguales a 500.000
                categoria = "Oro"
                # CÁLCULO ARITMÉTICO: Bono del 10% solo para clientes Oro
                bono_regalo = valor_compra * 0.10
            
            # MANEJO DE LISTAS: Estructuramos los datos del cliente.
            # OJO: Ponemos 'nombre_cliente' en la primera posición para que .sort() lo use como criterio principal.
            cliente_actual = [nombre_cliente, valor_compra, categoria, bono_regalo]
            
            # Agregamos el cliente a nuestro CRM (lista principal)
            base_clientes.append(cliente_actual)
            
        print("\n✅ ¡Los 5 clientes han sido segmentados y guardados exitosamente!")

    # --- OPCIÓN 2: PROCESAMIENTO Y REPORTE ---
    elif opcion == "2":
        # Validación de seguridad (Sanity Check): Evitamos imprimir una lista vacía
        if len(base_clientes) == 0:
            print("\n⚠️ Advertencia: El CRM está vacío. Ejecute la opción 1 primero.")
            continue # Devuelve el flujo al inicio del 'while'
        
        # ORDENAMIENTO DE DATOS:
        # El método .sort() ordena las listas basándose en su primer elemento.
        # Como nuestro primer elemento es el nombre, la lista quedará ordenada ALFABÉTICAMENTE.
        base_clientes.sort()
        
        print("\n" + "*"*60)
        print("💌 REPORTE DE LEALTAD PARA EMAIL MARKETING (A - Z)")
        print("*"*60)
        
        # RECORRIDO DE DATOS (CICLO FOR): Analizamos cliente por cliente
        for cliente in base_clientes:
            # Desempaquetamos los datos de la sub-lista para mayor claridad
            nombre_c = cliente[0]
            compra_c = cliente[1]
            categoria_c = cliente[2]
            bono_c = cliente[3]
            
            # Formateamos el mensaje dependiendo de si el cliente tiene bono o no
            if categoria_c == "Oro":
                mensaje_bono = f"🎁 ¡Bono de Regalo ganado!: ${bono_c:,.2f}"
            else:
                mensaje_bono = "Sigue comprando para llegar a Oro."
                
            # IMPRESIÓN (OUTPUT): Usamos f-strings para tabular y formatear montos monetarios
            print(f"👤 {nombre_c} | Categoría: {categoria_c.upper()} | Compras: ${compra_c:,.2f}")
            print(f"   ↳ {mensaje_bono}\n")
            
        print("*"*60)

    # --- OPCIÓN 3: SALIR DEL SISTEMA ---
    elif opcion == "3":
        print("\nCerrando sistema Fideliza-Puntos... ¡Aumentar las ventas es el objetivo! 🚀")
        break # Rompe el ciclo 'while' y finaliza la ejecución

    # --- VALIDACIÓN DE ENTRADA INCORRECTA ---
    else:
        print("\n❌ Error: Opción no válida. Por favor digite 1, 2 o 3.")
```

Puntos clave:

    El poder de .title(): Al pedir el nombre con input("...").strip().title(), nos aseguramos de que "ana", "ANA" y "Ana" se guarden todas como "Ana". En Python, las mayúsculas importan al momento de ordenar (una 'A' mayúscula va antes que una 'a' minúscula). Estandarizar la entrada asegura que el método .sort() funcione perfectamente como un orden alfabético real.

    Posición de las variables en la sub-lista: En la instrucción [nombre_cliente, valor_compra, categoria, bono_regalo], la decisión de poner el nombre de primero no fue casualidad. La rúbrica exige que el reporte salga ordenado alfabéticamente. Puesto que .sort() lee siempre el índice 0, esto soluciona el problema de ordenamiento sin necesidad de programar algoritmos complejos.

    Manejo de estados (El Bono en Cero): Iniciar la variable bono_regalo = 0 antes del bloque if/elif/else es una excelente práctica. Evita el famoso error de "variable no definida" si el código nunca entra a la condición "Oro". Esto se llama programación defensiva.

# 🏢 RETO 5: ADMINISTRACIÓN - "Nomina-Express"

Perfil: Administración de Empresas / Servicio.

Problema: Una PyME necesita liquidar la nómina semanal de sus empleados operativos teniendo en cuenta horas extra.

Funcionalidad:

Ingrese nombre y horas trabajadas de 5 empleados.

Si horas > 48, las adicionales se pagan como extras (+25% del valor hora).

Si el sueldo total es < 2 SMMLV, agregue Auxilio de Transporte (Simulado).

Muestre el desprendible de pago total de la empresa.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# RETO 5: ADMINISTRACIÓN - "Nomina-Express" (Liquidación Semanal y Horas Extra)
# ==============================================================================

# 1. INICIALIZACIÓN DE VARIABLES Y CONSTANTES GLOBALES
# Lista vacía que funcionará como nuestra base de datos de Recursos Humanos.
base_nomina = []

# Definimos las constantes económicas (Simuladas para el ejercicio).
# En administración es vital no dejar números "mágicos" sueltos en el código.
VALOR_HORA_BASE = 9000  # Tarifa estándar por hora en COP
RECARGO_EXTRA = 1.25    # 25% de recargo por hora extra laboral
SMMLV_MENSUAL = 1400000 # Salario Mínimo simulado
AUXILIO_TRANS_MENSUAL = 162000 

# Para liquidación semanal, dividimos los topes mensuales en 4 semanas.
TOPE_2_SMMLV_SEMANAL = (SMMLV_MENSUAL * 2) / 4
AUXILIO_TRANS_SEMANAL = AUXILIO_TRANS_MENSUAL / 4

print("🏢 BIENVENIDO AL SISTEMA DE RECURSOS HUMANOS 'NOMINA-EXPRESS' 🏢")

# 2. MENÚ INTERACTIVO (CICLO WHILE)
# Mantenemos el programa corriendo hasta que el administrador decida salir.
while True:
    print("\n" + "="*50)
    print("💼 MENÚ PRINCIPAL:")
    print("1. Registrar horas trabajadas de 5 empleados")
    print("2. Generar Desprendibles de Pago (Alfabético)")
    print("3. Salir del sistema")
    print("="*50)
    
    # Capturamos la opción del usuario
    opcion = input(">> Seleccione una opción (1, 2 o 3): ")
    
    # --- OPCIÓN 1: REGISTRO DE EMPLEADOS ---
    if opcion == "1":
        print("\n📝 MÓDULO DE INGRESO DE TIEMPOS")
        
        # Limpiamos la nómina previa por si se realiza un nuevo cálculo
        base_nomina.clear()
        
        # Ciclo 'for' para pedir exactamente la información de 5 empleados operativos.
        for i in range(1, 6):
            print(f"\n--- Empleado #{i} ---")
            
            # INPUT Y CASTING: Nombre (String) y Horas (Float, pueden haber medias horas)
            # Usamos .title() para estandarizar nombres y que el ordenamiento sea perfecto.
            nombre_emp = input("Nombre completo del empleado: ").strip().title()
            horas_trabajadas = float(input("Horas laboradas en la semana: "))
            
            # LÓGICA DE NEGOCIO (IF / ELSE): Cálculo de Horas Extras (> 48h)
            # Nota: Aunque la ley en Colombia está reduciendo la jornada, usamos 48h según el requerimiento.
            if horas_trabajadas > 48:
                horas_normales = 48
                horas_extras = horas_trabajadas - 48
            else:
                horas_normales = horas_trabajadas
                horas_extras = 0
                
            # CÁLCULOS ARITMÉTICOS: Salario Base y Pago de Extras
            pago_normal = horas_normales * VALOR_HORA_BASE
            pago_extras = horas_extras * (VALOR_HORA_BASE * RECARGO_EXTRA)
            subtotal_salario = pago_normal + pago_extras
            
            # LÓGICA DE NEGOCIO (IF / ELSE): Auxilio de Transporte
            # Si el salario proyectado es menor a 2 SMMLV, tiene derecho a auxilio.
            if subtotal_salario < TOPE_2_SMMLV_SEMANAL:
                pago_auxilio = AUXILIO_TRANS_SEMANAL
            else:
                pago_auxilio = 0
                
            # Cálculo del devengado total semanal
            total_devengado = subtotal_salario + pago_auxilio
            
            # MANEJO DE LISTAS: Agrupamos los datos.
            # Ubicamos 'nombre_emp' de primero para que .sort() ordene la nómina alfabéticamente.
            empleado_actual = [nombre_emp, horas_normales, horas_extras, pago_auxilio, total_devengado]
            
            # Guardamos el registro en la lista de la empresa
            base_nomina.append(empleado_actual)
            
        print("\n✅ ¡Los 5 empleados han sido liquidados y registrados exitosamente!")

    # --- OPCIÓN 2: PROCESAMIENTO Y REPORTE ---
    elif opcion == "2":
        # Validación de seguridad (Sanity Check):
        if len(base_nomina) == 0:
            print("\n⚠️ Advertencia: No hay registros de nómina. Ejecute la opción 1 primero.")
            continue
        
        # ORDENAMIENTO DE DATOS:
        # .sort() organizará la lista de listas evaluando el primer elemento (el nombre).
        base_nomina.sort()
        
        print("\n" + "*"*65)
        print("📄 DESPRENDIBLES DE PAGO SEMANAL (ORDEN ALFABÉTICO)")
        print("*"*65)
        
        # Acumulador para saber cuánto dinero debe tener la empresa en bancos.
        costo_total_nomina = 0
        
        # RECORRIDO DE DATOS (CICLO FOR):
        for empleado in base_nomina:
            # Desempaquetado de la sub-lista
            nombre_e = empleado[0]
            h_normales = empleado[1]
            h_extras = empleado[2]
            aux_trans = empleado[3]
            total_pago = empleado[4]
            
            # Sumamos al acumulador de la empresa
            costo_total_nomina += total_pago
            
            # Imprimimos el desprendible individual usando f-strings para tabular
            print(f"👤 Empleado: {nombre_e}")
            print(f"   ↳ H. Normales: {h_normales}h | H. Extras: {h_extras}h")
            print(f"   ↳ Aux. Trans:  ${aux_trans:,.2f}")
            print(f"   ↳ TOTAL PAGAR: ${total_pago:,.2f} COP\n")
            
        print("-" * 65)
        print(f"💰 COSTO TOTAL NÓMINA EMPRESA: ${costo_total_nomina:,.2f} COP")
        print("*"*65)

    # --- OPCIÓN 3: SALIR ---
    elif opcion == "3":
        print("\nCerrando sistema Nomina-Express... ¡Nómina liquidada con éxito! 🏦")
        break

    # --- VALIDACIÓN DE ERRORES ---
    else:
        print("\n❌ Error: Opción no válida. Por favor digite 1, 2 o 3.")
```

Puntos clave:

    Evitar los "Números Mágicos": En Administración, codificar valores directos como 9000 o 162000 dentro de los if es una mala práctica. Al usar constantes al inicio del archivo (ej. VALOR_HORA_BASE = 9000), si la ley cambia o el salario mínimo sube al año siguiente, solo deben cambiar la variable en la línea 10 y no revisar las 100 líneas de código hacia abajo. Esto es escalabilidad.

    Cálculo de Proporcionalidad (Semanal): Dado que el problema exige un pago semanal, se deben dividir los topes mensuales legales (SMMLV y Auxilio de Transporte) entre 4 para no otorgar subsidios desproporcionados. Un buen administrador debe auditar ("Sanity Check") que los datos arrojados por el sistema tengan sentido en la vida real.

    El tope de las 48 Horas: La Ley 2101 de 2021 en Colombia está reduciendo gradualmente la jornada laboral a 42 horas, pero, para mantener la fidelidad a las reglas del cliente (el enunciado del taller), se parametrizó la validación en 48 horas exactas. Esto demuestra atención al detalle del cliente frente a las reglas del mercado.

# 🚀 RETO 1: LOGÍSTICA - "Ruta-Óptima"

Perfil: Administración de Negocios Internacionales / Logística.

Problema: Una empresa de envíos en Bogotá necesita clasificar paquetes antes de cargarlos a los camiones.

### Funcionalidad:

Pida el peso (kg) y destino de 5 paquetes.

Clasifique automáticamente: Si pesa < 1kg es "Documento", < 10kg es "Paquetería", > 10kg es "Carga".

Calcule el costo de envío (Tarifa base x Kg).

Imprima un manifiesto de carga ordenado por peso.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# TALLER CORTE 2 - RETO 1: LOGÍSTICA "RUTA-ÓPTIMA"
# ARQUITECTURA: POO (3 Padres, 3 Hijos, Encapsulamiento) + SQLite (CRUD, JOIN, PK/FK)
# ==============================================================================

import sqlite3

# ==============================================================================
# CONFIGURACIÓN DEL SISTEMA
# ==============================================================================
# Nombre de la base de datos (cambiar aquí para modificar en todo el sistema)
NOMBRE_BD = 'reto_1_logistica.db'

# ==============================================================================
# 1. PROGRAMACIÓN ORIENTADA A OBJETOS (POO) - JERARQUÍA DE CLASES
# ==============================================================================

# --- JERARQUÍA 1: CLIENTES ---
class EntidadComercial:
    """Clase Padre 1: Representa cualquier entidad que interactúa con la empresa."""
    def __init__(self, nombre):
        # Encapsulamiento: Atributo protegido
        # Si no se usara _ sería un atributo público, es decir, accesible directamente desde fuera de la clase sin restricciones. 
        # Por ejemplo, cliente._nombre sería accesible y saldría un error si se intentara acceder a cliente.__nombre porque es privado.
        self._nombre = nombre

class Cliente(EntidadComercial):
    """Clase Hijo 1: Representa a un cliente específico."""
    def __init__(self, id_cliente, nombre, email):
        super().__init__(nombre)
        # Encapsulamiento: Atributos privados
        self.__id_cliente = id_cliente
        self.__email = email

    # Getters y Setters
    def get_id(self): return self.__id_cliente
    def get_nombre(self): return self._nombre
    def get_email(self): return self.__email


# --- JERARQUÍA 2: VEHÍCULOS ---
class Transporte:
    """Clase Padre 2: Representa un medio de transporte genérico."""
    def __init__(self, placa):
        self._placa = placa

class Camion(Transporte):
    """Clase Hijo 2: Representa un camión de la flota logística."""
    def __init__(self, placa, capacidad_kg):
        super().__init__(placa)
        self.__capacidad_kg = capacidad_kg

    def get_placa(self): return self._placa
    def get_capacidad(self): return self.__capacidad_kg


# --- JERARQUÍA 3: OPERACIONES LOGÍSTICAS ---
class OperacionBase:
    """Clase Padre 3: Representa una operación logística base."""
    def __init__(self, destino):
        self._destino = destino

class Envio(OperacionBase):
    """Clase Hijo 3: Representa el envío de un paquete específico."""
    def __init__(self, destino, peso, id_cliente, placa_camion):
        super().__init__(destino)
        self.__peso = peso
        self.__id_cliente = id_cliente
        self.__placa_camion = placa_camion
        self.__tarifa_base = 5000  # Tarifa base por Kg en COP
        
        # Atributos calculados automáticamente
        self.__categoria = self._clasificar_paquete()
        self.__costo_total = self._calcular_costo()

    # Métodos privados para lógica interna (Encapsulamiento de lógica de negocio)
    def _clasificar_paquete(self):
        """Clasifica el paquete según las reglas de negocio del Reto 1."""
        if self.__peso < 1:
            return "Documento"
        elif self.__peso < 10:
            return "Paquetería"
        else:
            return "Carga"

    def _calcular_costo(self):
        """Calcula el costo total multiplicando el peso por la tarifa base."""
        return self.__peso * self.__tarifa_base

    # Getters
    def get_destino(self): return self._destino
    def get_peso(self): return self.__peso
    def get_categoria(self): return self.__categoria
    def get_costo(self): return self.__costo_total
    def get_id_cliente(self): return self.__id_cliente
    def get_placa_camion(self): return self.__placa_camion


# ==============================================================================
# 2. PERSISTENCIA DE DATOS (SQLite) - INICIALIZACIÓN Y POBLADO
# ==============================================================================

def inicializar_base_datos():
    """Crea las tablas y asegura el mínimo de 5 registros por tabla."""
    try:
        # Conectamos a la base de datos (se crea si no existe)
        conn = sqlite3.connect(NOMBRE_BD)
        cursor = conn.cursor()
        
        # Activamos las llaves foráneas (Obligatorio en SQLite)
        cursor.execute("PRAGMA foreign_keys = ON;")

        # Creación de Tabla 1: Clientes
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS clientes (
                id_cliente INTEGER PRIMARY KEY AUTOINCREMENT,
                nombre TEXT NOT NULL,
                email TEXT UNIQUE
            )
        ''')

        # Creación de Tabla 2: Camiones
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS camiones (
                placa TEXT PRIMARY KEY,
                capacidad_kg REAL NOT NULL
            )
        ''')

        # Creación de Tabla 3: Envios (Con 2 Llaves Foráneas)
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS envios (
                id_envio INTEGER PRIMARY KEY AUTOINCREMENT,
                destino TEXT NOT NULL,
                peso REAL NOT NULL,
                categoria TEXT NOT NULL,
                costo REAL NOT NULL,
                id_cliente INTEGER,
                placa_camion TEXT,
                FOREIGN KEY (id_cliente) REFERENCES clientes(id_cliente),
                FOREIGN KEY (placa_camion) REFERENCES camiones(placa)
            )
        ''')

        # Inserción de 5 registros mínimos (Regla de Oro) usando INSERT OR IGNORE
        clientes_data =[
            ('Empresa Andina', 'contacto@andina.com'),
            ('Textiles Bogotá', 'logistica@textiles.com'),
            ('Tech Colombia', 'envios@techcol.com'),
            ('Distribuidora Sur', 'gerencia@distsur.com'),
            ('Importaciones XYZ', 'info@xyz.com')
        ]
        cursor.executemany("INSERT OR IGNORE INTO clientes (nombre, email) VALUES (?, ?)", clientes_data)

        camiones_data =[
            ('KTO-001', 5000), ('KTO-002', 8000), ('KTO-003', 12000),
            ('KTO-004', 3500), ('KTO-005', 15000)
        ]
        cursor.executemany("INSERT OR IGNORE INTO camiones (placa, capacidad_kg) VALUES (?, ?)", camiones_data)

        envios_data =[
            ('Medellín', 0.5, 'Documento', 2500, 1, 'KTO-001'),
            ('Cali', 5.0, 'Paquetería', 25000, 2, 'KTO-002'),
            ('Barranquilla', 15.0, 'Carga', 75000, 3, 'KTO-003'),
            ('Bucaramanga', 0.8, 'Documento', 4000, 4, 'KTO-004'),
            ('Cartagena', 8.5, 'Paquetería', 42500, 5, 'KTO-005')
        ]
        cursor.executemany('''
            INSERT OR IGNORE INTO envios (destino, peso, categoria, costo, id_cliente, placa_camion) 
            VALUES (?, ?, ?, ?, ?, ?)
        ''', envios_data)

        conn.commit()
    except sqlite3.Error as e:
        print(f"⚠️ Error al inicializar la base de datos: {e}")
    finally:
        conn.close()


def mostrar_datos_iniciales():
    """Imprime todos los registros creados en las tablas de prueba al iniciar el programa."""
    try:
        conn = sqlite3.connect(NOMBRE_BD)
        cursor = conn.cursor()
        
        print("\n" + "="*75)
        print("📂 DATOS INICIALES DISPONIBLES EN LA BASE DE DATOS")
        print("="*75)
        
        # Mostrar Clientes
        cursor.execute("SELECT * FROM clientes")
        print("\n🏢 CLIENTES (ID | Nombre | Email):")
        for c in cursor.fetchall(): 
            print(f"   {c[0]} | {c[1]} | {c[2]}")
        
        # Mostrar Camiones
        cursor.execute("SELECT * FROM camiones")
        print("\n🚚 CAMIONES (Placa | Capacidad en Kg):")
        for cam in cursor.fetchall(): 
            print(f"   {cam[0]} | {cam[1]:,.0f} kg")
        
        # Mostrar Envíos
        cursor.execute("SELECT * FROM envios")
        print("\n📦 ENVÍOS (ID | Destino | Peso | Categoría | Costo | ID Cliente | Placa):")
        for e in cursor.fetchall(): 
            print(f"   {e[0]} | {e[1]} | {e[2]} kg | {e[3]} | ${e[4]:,.0f} | Cliente: {e[5]} | {e[6]}")
        
        print("="*75)
        
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar datos iniciales: {e}")
    finally:
        if 'conn' in locals(): conn.close()


# ==============================================================================
# 3. OPERACIONES CRUD Y MENÚ INTERACTIVO
# ==============================================================================

def registrar_envio():
    """[CREATE] Registra un nuevo envío en la base de datos."""
    try:
        print("\n--- REGISTRAR NUEVO ENVÍO ---")
        id_cliente = int(input("ID del Cliente (Ej: 1 al 5): "))
        placa_camion = input("Placa del Camión asignado (Ej: KTO-001): ").strip().upper()
        destino = input("Ciudad de destino: ").strip().title()
        peso = float(input("Peso del paquete en Kg: "))

        # Validamos lógicamente el peso
        if peso <= 0:
            print("❌ Error: El peso debe ser mayor a 0.")
            return

        # Instanciamos el objeto Envio (POO) para que calcule categoría y costo
        nuevo_envio = Envio(destino, peso, id_cliente, placa_camion)

        # Guardamos en SQLite
        conn = sqlite3.connect(NOMBRE_BD)
        cursor = conn.cursor()
        cursor.execute("PRAGMA foreign_keys = ON;")
        
        cursor.execute('''
            INSERT INTO envios (destino, peso, categoria, costo, id_cliente, placa_camion)
            VALUES (?, ?, ?, ?, ?, ?)
        ''', (nuevo_envio.get_destino(), nuevo_envio.get_peso(), nuevo_envio.get_categoria(), 
              nuevo_envio.get_costo(), nuevo_envio.get_id_cliente(), nuevo_envio.get_placa_camion()))
        
        conn.commit()
        print(f"✅ Envío registrado con éxito. Categoría: {nuevo_envio.get_categoria()} | Costo: ${nuevo_envio.get_costo():,.2f}")
    
    except ValueError:
        print("⚠️ Error: Ingresó un texto donde se esperaba un número.")
    except sqlite3.IntegrityError:
        print("⚠️ Error: El ID de cliente o la placa del camión no existen en la base de datos.")
    except sqlite3.Error as e:
        print(f"⚠️ Error de base de datos: {e}")
    finally:
        if 'conn' in locals(): conn.close()


def ver_manifiesto():
    """[READ] Muestra el manifiesto de carga usando INNER JOIN, ordenado por peso."""
    try:
        conn = sqlite3.connect(NOMBRE_BD)
        cursor = conn.cursor()
        
        # Consulta JOIN uniendo las 3 tablas y ordenando por peso (Requisito del Reto 1)
        consulta = '''
            SELECT 
                envios.id_envio, 
                clientes.nombre, 
                envios.destino, 
                envios.peso, 
                envios.categoria, 
                envios.costo, 
                camiones.placa
            FROM envios
            INNER JOIN clientes ON envios.id_cliente = clientes.id_cliente
            INNER JOIN camiones ON envios.placa_camion = camiones.placa
            ORDER BY envios.peso ASC
        '''
        cursor.execute(consulta)
        registros = cursor.fetchall()

        print("\n" + "="*85)
        print("📄 MANIFIESTO DE CARGA (ORDENADO POR PESO ASCENDENTE)")
        print("="*85)
        
        if not registros:
            print("No hay envíos registrados.")
        else:
            for r in registros:
                print(f"ID: {r[0]} | Cliente: {r[1][:15].ljust(15)} | Destino: {r[2].ljust(12)} | "
                      f"Peso: {r[3]:>5}kg | Tipo: {r[4].ljust(10)} | Camión: {r[6]} | Costo: ${r[5]:,.2f}")
        print("="*85)
        
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar manifiesto: {e}")
    finally:
        if 'conn' in locals(): conn.close()


def actualizar_destino():
    """[UPDATE] Actualiza la ciudad de destino de un envío existente."""
    try:
        id_envio = int(input("\nIngrese el ID del envío a actualizar: "))
        nuevo_destino = input("Ingrese la nueva ciudad de destino: ").strip().title()

        conn = sqlite3.connect(NOMBRE_BD)
        cursor = conn.cursor()
        
        # Ejecutamos el UPDATE
        cursor.execute("UPDATE envios SET destino = ? WHERE id_envio = ?", (nuevo_destino, id_envio))
        
        # Verificamos si se afectó alguna fila
        if cursor.rowcount > 0:
            conn.commit()
            print(f"✅ Destino del envío #{id_envio} actualizado a {nuevo_destino}.")
        else:
            print(f"⚠️ No se encontró ningún envío con el ID {id_envio}.")
            
    except ValueError:
        print("⚠️ Error: El ID debe ser un número entero.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al actualizar: {e}")
    finally:
        if 'conn' in locals(): conn.close()


def eliminar_envio():
    """[DELETE] Elimina un envío del sistema."""
    try:
        id_envio = int(input("\nIngrese el ID del envío a cancelar/eliminar: "))

        conn = sqlite3.connect(NOMBRE_BD)
        cursor = conn.cursor()
        
        cursor.execute("DELETE FROM envios WHERE id_envio = ?", (id_envio,))
        
        if cursor.rowcount > 0:
            conn.commit()
            print(f"✅ Envío #{id_envio} eliminado del sistema exitosamente.")
        else:
            print(f"⚠️ No se encontró ningún envío con el ID {id_envio}.")
            
    except ValueError:
        print("⚠️ Error: El ID debe ser un número entero.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al eliminar: {e}")
    finally:
        if 'conn' in locals(): conn.close()


def menu_principal():
    """Controlador principal del programa (Ciclo While)."""
    # 1. Preparamos la base de datos antes de arrancar
    inicializar_base_datos()
    
    # 2. Imprimimos los datos iniciales creados en las tablas
    mostrar_datos_iniciales()
    
    # 3. Bucle infinito para mantener el programa vivo
    while True:
        print("\n" + "*"*40)
        print("🚛 SISTEMA LOGÍSTICO 'RUTA-ÓPTIMA' 🚛")
        print("*"*40)
        print("1. [CREATE] Registrar nuevo envío")
        print("2. [READ] Ver Manifiesto de Carga (JOIN)")
        print("3.[UPDATE] Actualizar destino de un envío")
        print("4. [DELETE] Cancelar/Eliminar un envío")
        print("5. Salir del sistema")
        
        opcion = input("\n>> Seleccione una opción (1-5): ").strip()
        
        if opcion == "1":
            registrar_envio()
        elif opcion == "2":
            ver_manifiesto()
        elif opcion == "3":
            actualizar_destino()
        elif opcion == "4":
            eliminar_envio()
        elif opcion == "5":
            print("\nCerrando el sistema Ruta-Óptima... ¡Buen viaje! 🚛💨")
            break
        else:
            print("\n❌ Opción no válida. Por favor, digite un número del 1 al 5.")

# ==============================================================================
# 4. PUNTO DE ENTRADA DEL SCRIPT
# ==============================================================================
if __name__ == "__main__":
    menu_principal()
```

# 🥗 RETO 2: GASTRONOMÍA - "Chef-Costos"

Perfil: Gastronomía.

Problema: Un restaurante necesita controlar el costo de sus platos para no perder dinero por la inflación de alimentos en Colombia.

### Funcionalidad:

Permita ingresar 5 ingredientes para una receta (Nombre y Precio de Mercado).

Valide si el precio subió más del 20% respecto a un precio base (decisión lógica).

Calcule el costo total del plato y sugiera un Precio de Venta (Costo + 35% ganancia).

Muestre la lista de ingredientes ordenada del más caro al más barato.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# TALLER CORTE 2 - RETO 2: GASTRONOMÍA "CHEF-COSTOS"
# ARQUITECTURA: POO (3 Padres, 3 Hijos, Encapsulamiento) + SQLite (CRUD, JOIN, PK/FK)
# ==============================================================================

import sqlite3

# Definimos el nombre de la base de datos como una constante global
DB_NAME = "reto_2_chef_costos.db"

# ==============================================================================
# 1. PROGRAMACIÓN ORIENTADA A OBJETOS (POO) - JERARQUÍA DE CLASES
# ==============================================================================

# --- JERARQUÍA 1: PROVEEDORES ---
class Entidad:
    """Clase Padre 1: Representa una entidad genérica del negocio."""
    def __init__(self, nombre):
        # Atributo protegido (accesible por las clases hijas)
        self._nombre = nombre

class Proveedor(Entidad):
    """Clase Hijo 1: Representa a un proveedor de insumos."""
    def __init__(self, id_proveedor, nombre, telefono):
        super().__init__(nombre)
        # Encapsulamiento: Atributos privados
        self.__id_proveedor = id_proveedor
        self.__telefono = telefono

    # Getters para acceder a los datos privados de forma segura
    def get_id(self): return self.__id_proveedor
    def get_nombre(self): return self._nombre
    def get_telefono(self): return self.__telefono


# --- JERARQUÍA 2: PLATOS (RECETAS) ---
class Preparacion:
    """Clase Padre 2: Representa una preparación culinaria base."""
    def __init__(self, nombre_plato):
        self._nombre_plato = nombre_plato

class Plato(Preparacion):
    """Clase Hijo 2: Representa un plato final en el menú del restaurante."""
    def __init__(self, nombre_plato, costo_total):
        super().__init__(nombre_plato)
        self.__costo_total = costo_total
        # Lógica de negocio: El precio sugerido es el costo + 35% de ganancia
        self.__precio_sugerido = costo_total * 1.35

    def get_nombre(self): return self._nombre_plato
    def get_costo(self): return self.__costo_total
    def get_precio_sugerido(self): return self.__precio_sugerido


# --- JERARQUÍA 3: INGREDIENTES ---
class Insumo:
    """Clase Padre 3: Representa un insumo básico."""
    def __init__(self, nombre_ingrediente):
        self._nombre_ingrediente = nombre_ingrediente

class IngredienteMercado(Insumo):
    """Clase Hijo 3: Representa un ingrediente con análisis de inflación."""
    def __init__(self, nombre_ingrediente, precio_base, precio_mercado, id_proveedor):
        super().__init__(nombre_ingrediente)
        self.__precio_base = precio_base
        self.__precio_mercado = precio_mercado
        self.__id_proveedor = id_proveedor

    # Getters
    def get_nombre(self): return self._nombre_ingrediente
    def get_precio_base(self): return self.__precio_base
    def get_precio_mercado(self): return self.__precio_mercado
    def get_id_proveedor(self): return self.__id_proveedor

    # Lógica de negocio encapsulada: Cálculo de inflación
    def calcular_inflacion(self):
        """Calcula el porcentaje de incremento del precio."""
        if self.__precio_base <= 0:
            return 0.0
        return ((self.__precio_mercado - self.__precio_base) / self.__precio_base) * 100

    def evaluar_alerta(self):
        """Retorna una alerta si la inflación supera el 20%."""
        inflacion = self.calcular_inflacion()
        if inflacion > 20:
            return f"⚠️ ALERTA: Subió {inflacion:.1f}%"
        return f"✅ Estable ({inflacion:.1f}%)"


# ==============================================================================
# 2. PERSISTENCIA DE DATOS (SQLite) - INICIALIZACIÓN Y POBLADO
# ==============================================================================

def inicializar_base_datos():
    """Crea las tablas y asegura el mínimo de 5 registros por tabla."""
    try:
        # El bloque 'with' abre y cierra la conexión automáticamente de forma segura
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Activamos las llaves foráneas (Obligatorio en SQLite)
            cursor.execute("PRAGMA foreign_keys = ON;")

            # Creación de Tabla 1: Proveedores (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS proveedores (
                    id_proveedor INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre TEXT NOT NULL,
                    telefono TEXT NOT NULL
                )
            ''')

            # Creación de Tabla 2: Platos (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS platos (
                    id_plato INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre_plato TEXT NOT NULL,
                    costo_total REAL NOT NULL,
                    precio_sugerido REAL NOT NULL
                )
            ''')

            # Creación de Tabla 3: Ingredientes (Tiene 2 FK, relaciona Platos y Proveedores)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS ingredientes_mercado (
                    id_ingrediente INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre_ingrediente TEXT NOT NULL,
                    precio_base REAL NOT NULL,
                    precio_mercado REAL NOT NULL,
                    id_plato INTEGER,
                    id_proveedor INTEGER,
                    FOREIGN KEY (id_plato) REFERENCES platos(id_plato) ON DELETE CASCADE,
                    FOREIGN KEY (id_proveedor) REFERENCES proveedores(id_proveedor)
                )
            ''')

            # Inserción de 5 registros mínimos (Regla de Oro) usando INSERT OR IGNORE
            proveedores_data =[
                ('Carnes La Sabana', '3001112233'),
                ('Lácteos El Prado', '3102223344'),
                ('Verduras Frescas', '3203334455'),
                ('Salsas y Especias', '3004445566'),
                ('Panadería Central', '3115556677')
            ]
            # Insertamos solo si la tabla está vacía para no duplicar en cada ejecución
            cursor.execute("SELECT COUNT(*) FROM proveedores")
            if cursor.fetchone()[0] == 0:
                cursor.executemany("INSERT INTO proveedores (nombre, telefono) VALUES (?, ?)", proveedores_data)

            platos_data =[
                ('Hamburguesa Gourmet', 15000, 20250),
                ('Pizza Artesanal', 18000, 24300),
                ('Ensalada César', 10000, 13500),
                ('Pasta Carbonara', 12000, 16200),
                ('Sopa de Tomate', 8000, 10800)
            ]
            cursor.execute("SELECT COUNT(*) FROM platos")
            if cursor.fetchone()[0] == 0:
                cursor.executemany("INSERT INTO platos (nombre_plato, costo_total, precio_sugerido) VALUES (?, ?, ?)", platos_data)

            ingredientes_data =[
                ('Carne de Res', 5000, 6500, 1, 1),
                ('Queso Mozzarella', 2000, 2200, 2, 2),
                ('Lechuga Romana', 1000, 1500, 3, 3),
                ('Salsa de Tomate', 1500, 1600, 4, 4),
                ('Pan Brioche', 1200, 1200, 1, 5)
            ]
            cursor.execute("SELECT COUNT(*) FROM ingredientes_mercado")
            if cursor.fetchone()[0] == 0:
                cursor.executemany('''
                    INSERT INTO ingredientes_mercado (nombre_ingrediente, precio_base, precio_mercado, id_plato, id_proveedor) 
                    VALUES (?, ?, ?, ?, ?)
                ''', ingredientes_data)

            conn.commit()
    except sqlite3.Error as e:
        print(f"⚠️ Error al inicializar la base de datos: {e}")

def mostrar_datos_iniciales():
    """Imprime todos los registros creados en las tablas de prueba al iniciar el programa."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            print("\n" + "="*70)
            print("📂 DATOS INICIALES DISPONIBLES EN LA BASE DE DATOS")
            print("="*70)
            
            # Mostrar Proveedores
            cursor.execute("SELECT * FROM proveedores")
            print("🏢 PROVEEDORES (ID | Nombre | Teléfono):")
            for p in cursor.fetchall(): print(f"   {p[0]} | {p[1]} | {p[2]}")
            
            # Mostrar Platos
            cursor.execute("SELECT * FROM platos")
            print("\n🍲 PLATOS (ID | Nombre | Costo | Precio Sugerido):")
            for pl in cursor.fetchall(): print(f"   {pl[0]} | {pl[1]} | ${pl[2]:,.0f} | ${pl[3]:,.0f}")
            
            # Mostrar Ingredientes
            cursor.execute("SELECT id_ingrediente, nombre_ingrediente, precio_mercado, id_plato, id_proveedor FROM ingredientes_mercado")
            print("\n🥦 INGREDIENTES (ID | Nombre | Precio Mercado | ID Plato | ID Proveedor):")
            for i in cursor.fetchall(): print(f"   {i[0]} | {i[1]} | ${i[2]:,.0f} | Plato: {i[3]} | Prov: {i[4]}")
            print("="*70 + "\n")
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar datos iniciales: {e}")


# ==============================================================================
# 3. OPERACIONES CRUD Y MENÚ INTERACTIVO
# ==============================================================================

def registrar_receta():
    """[CREATE] Registra un nuevo plato y sus 5 ingredientes en la base de datos."""
    try:
        print("\n--- 📝 REGISTRAR NUEVA RECETA ---")
        nombre_plato = input("Nombre del nuevo plato: ").strip().title()
        
        lista_ingredientes =[]
        costo_total = 0.0

        # Pedimos exactamente 5 ingredientes usando un ciclo for
        for i in range(1, 6):
            print(f"\n🔸 Ingrediente #{i}")
            nombre_ing = input("   Nombre: ").strip().title()
            precio_base = float(input("   Precio Base (Histórico) en COP: "))
            precio_mercado = float(input("   Precio Actual de Mercado en COP: "))
            id_proveedor = int(input("   ID del Proveedor (Ej: 1 al 5): "))
            
            # Validaciones lógicas
            if precio_base < 0 or precio_mercado < 0:
                print("   ❌ Error: Los precios no pueden ser negativos. Intente de nuevo.")
                return

            # Instanciamos el objeto (POO)
            nuevo_ingrediente = IngredienteMercado(nombre_ing, precio_base, precio_mercado, id_proveedor)
            lista_ingredientes.append(nuevo_ingrediente)
            
            # Sumamos al costo total del plato
            costo_total += precio_mercado

        # Instanciamos el Plato (POO) para que calcule el precio sugerido
        nuevo_plato = Plato(nombre_plato, costo_total)

        # ORDENAMIENTO: Ordenamos la lista de ingredientes del más caro al más barato
        # Usamos una función lambda para decirle a sort() que ordene basándose en el precio de mercado
        lista_ingredientes.sort(key=lambda x: x.get_precio_mercado(), reverse=True)

        print("\n" + "*"*50)
        print(f"🍲 RESUMEN DE RECETA: {nuevo_plato.get_nombre().upper()}")
        print("*"*50)
        
        # Imprimimos la lista ordenada y con las alertas de inflación
        for ing in lista_ingredientes:
            print(f"🥦 {ing.get_nombre().ljust(20)} | ${ing.get_precio_mercado():,.0f} | {ing.evaluar_alerta()}")
            
        print("-" * 50)
        print(f"📉 COSTO TOTAL:       ${nuevo_plato.get_costo():,.0f} COP")
        print(f"📈 PRECIO SUGERIDO:   ${nuevo_plato.get_precio_sugerido():,.0f} COP (Margen 35%)")
        print("*"*50)

        # GUARDAR EN BASE DE DATOS (Transacción)
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("PRAGMA foreign_keys = ON;")
            
            # 1. Guardamos el plato
            cursor.execute('''
                INSERT INTO platos (nombre_plato, costo_total, precio_sugerido)
                VALUES (?, ?, ?)
            ''', (nuevo_plato.get_nombre(), nuevo_plato.get_costo(), nuevo_plato.get_precio_sugerido()))
            
            # Recuperamos el ID del plato recién creado para asignarlo a los ingredientes (Llave Foránea)
            id_plato_generado = cursor.lastrowid
            
            # 2. Guardamos los ingredientes vinculados al plato y al proveedor
            for ing in lista_ingredientes:
                cursor.execute('''
                    INSERT INTO ingredientes_mercado (nombre_ingrediente, precio_base, precio_mercado, id_plato, id_proveedor)
                    VALUES (?, ?, ?, ?, ?)
                ''', (ing.get_nombre(), ing.get_precio_base(), ing.get_precio_mercado(), id_plato_generado, ing.get_id_proveedor()))
            
            conn.commit()
            print("\n✅ Receta y sus ingredientes guardados exitosamente en la Base de Datos.")

    except ValueError:
        print("\n⚠️ Error: Ingresó un texto donde se esperaba un número (Precio o ID).")
    except sqlite3.IntegrityError:
        print("\n⚠️ Error de Integridad: El ID del proveedor ingresado no existe en la base de datos.")
    except sqlite3.Error as e:
        print(f"\n⚠️ Error de base de datos: {e}")


def ver_recetas_y_costos():
    """[READ] Muestra un reporte cruzando Platos, Ingredientes y Proveedores usando INNER JOIN."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Consulta JOIN uniendo las 3 tablas
            consulta = '''
                SELECT 
                    platos.nombre_plato,
                    ingredientes_mercado.nombre_ingrediente,
                    ingredientes_mercado.precio_mercado,
                    proveedores.nombre AS nombre_proveedor
                FROM ingredientes_mercado
                INNER JOIN platos ON ingredientes_mercado.id_plato = platos.id_plato
                INNER JOIN proveedores ON ingredientes_mercado.id_proveedor = proveedores.id_proveedor
                ORDER BY platos.nombre_plato ASC, ingredientes_mercado.precio_mercado DESC
            '''
            cursor.execute(consulta)
            registros = cursor.fetchall()

            print("\n" + "="*80)
            print("📄 REPORTE DE RECETAS, COSTOS Y PROVEEDORES (INNER JOIN)")
            print("="*80)
            
            if not registros:
                print("No hay ingredientes registrados.")
            else:
                plato_actual = ""
                for r in registros:
                    # Agrupamos visualmente por plato
                    if r[0] != plato_actual:
                        plato_actual = r[0]
                        print(f"\n🍲 PLATO: {plato_actual.upper()}")
                        print("-" * 80)
                    
                    print(f"   🥦 Ingrediente: {r[1].ljust(20)} | Costo: ${r[2]:>6,.0f} | 🏢 Prov: {r[3]}")
            print("="*80)
            
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar reporte: {e}")


def actualizar_precio_ingrediente():
    """[UPDATE] Actualiza el precio de mercado de un ingrediente existente."""
    try:
        id_ingrediente = int(input("\nIngrese el ID del ingrediente a actualizar: "))
        nuevo_precio = float(input("Ingrese el nuevo precio de mercado en COP: "))

        if nuevo_precio < 0:
            print("❌ Error: El precio no puede ser negativo.")
            return

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("UPDATE ingredientes_mercado SET precio_mercado = ? WHERE id_ingrediente = ?", (nuevo_precio, id_ingrediente))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Precio del ingrediente #{id_ingrediente} actualizado a ${nuevo_precio:,.0f}.")
            else:
                print(f"⚠️ No se encontró ningún ingrediente con el ID {id_ingrediente}.")
                
    except ValueError:
        print("⚠️ Error: Debe ingresar valores numéricos válidos.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al actualizar: {e}")


def eliminar_plato():
    """[DELETE] Elimina un plato. (Los ingredientes se borran en cascada gracias a ON DELETE CASCADE)."""
    try:
        id_plato = int(input("\nIngrese el ID del plato a eliminar del menú: "))

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("PRAGMA foreign_keys = ON;") # Vital para que funcione el borrado en cascada
            
            cursor.execute("DELETE FROM platos WHERE id_plato = ?", (id_plato,))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Plato #{id_plato} y sus ingredientes asociados eliminados exitosamente.")
            else:
                print(f"⚠️ No se encontró ningún plato con el ID {id_plato}.")
                
    except ValueError:
        print("⚠️ Error: El ID debe ser un número entero.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al eliminar: {e}")


def menu_principal():
    """Controlador principal del programa (Ciclo While)."""
    # 1. Preparamos la base de datos
    inicializar_base_datos()
    
    # 2. Imprimimos el menú y los registros iniciales creados
    mostrar_datos_iniciales()
    
    # 3. Bucle infinito para mantener el programa vivo
    while True:
        print("\n" + "*"*45)
        print("👨‍🍳 SISTEMA GASTRONÓMICO 'CHEF-COSTOS' 👩‍🍳")
        print("*"*45)
        print("1. [CREATE] Registrar nueva receta (5 ingredientes)")
        print("2. [READ] Ver Reporte de Costos y Proveedores (JOIN)")
        print("3. [UPDATE] Actualizar precio de un ingrediente")
        print("4. [DELETE] Eliminar un plato del menú")
        print("5. Salir del sistema")
        
        opcion = input("\n>> Seleccione una opción (1-5): ").strip()
        
        if opcion == "1":
            registrar_receta()
        elif opcion == "2":
            ver_recetas_y_costos()
        elif opcion == "3":
            actualizar_precio_ingrediente()
        elif opcion == "4":
            eliminar_plato()
        elif opcion == "5":
            print("\nCerrando el sistema Chef-Costos... ¡Éxitos en el servicio! 🍽️")
            break
        else:
            print("\n❌ Opción no válida. Por favor, digite un número del 1 al 5.")

# ==============================================================================
# 4. PUNTO DE ENTRADA DEL SCRIPT
# ==============================================================================
if __name__ == "__main__":
    menu_principal()
```

# 📊 RETO 3: ECONOMÍA - "Dolar-Track"

Perfil: Economía y Finanzas.

Problema: Un inversionista necesita analizar el comportamiento de la TRM (Tasa Representativa del Mercado) de la última semana para decidir si comprar dólares.

### Funcionalidad:

Ingrese la TRM de los últimos 7 días.

Calcule el promedio de la semana.

Use un for para identificar cuántos días la TRM estuvo por encima del promedio (Alerta de Venta) y cuántos por debajo (Alerta de Compra).

Determine la volatilidad (Diferencia entre el valor máximo y mínimo).

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# TALLER CORTE 2 - RETO 3: ECONOMÍA "DOLAR-TRACK"
# ARQUITECTURA: POO (3 Padres, 3 Hijos, Encapsulamiento) + SQLite (CRUD, JOIN, PK/FK)
# ==============================================================================

import sqlite3

# Definimos el nombre de la base de datos como una constante global
DB_NAME = "reto_3_dolar_track.db"

# ==============================================================================
# 1. PROGRAMACIÓN ORIENTADA A OBJETOS (POO) - JERARQUÍA DE CLASES
# ==============================================================================

# --- JERARQUÍA 1: INVERSIONISTAS ---
class Persona:
    """Clase Padre 1: Representa una persona genérica en el sistema."""
    def __init__(self, nombre):
        # Atributo protegido: accesible por las clases hijas
        self._nombre = nombre

class Inversionista(Persona):
    """Clase Hijo 1: Representa a un inversionista con perfil de riesgo."""
    def __init__(self, id_inversionista, nombre, perfil_riesgo):
        # Llamamos al constructor de la clase padre
        super().__init__(nombre)
        # Encapsulamiento: Atributos privados
        self.__id_inversionista = id_inversionista
        self.__perfil_riesgo = perfil_riesgo

    # Getters para acceder a los datos privados de forma segura
    def get_id(self): return self.__id_inversionista
    def get_nombre(self): return self._nombre
    def get_perfil(self): return self.__perfil_riesgo


# --- JERARQUÍA 2: MONEDAS ---
class Divisa:
    """Clase Padre 2: Representa una divisa genérica."""
    def __init__(self, nombre_divisa):
        self._nombre_divisa = nombre_divisa

class MonedaExtranjera(Divisa):
    """Clase Hijo 2: Representa una moneda extranjera específica con su código."""
    def __init__(self, codigo, nombre_divisa):
        super().__init__(nombre_divisa)
        self.__codigo = codigo

    def get_codigo(self): return self.__codigo
    def get_nombre(self): return self._nombre_divisa


# --- JERARQUÍA 3: ANÁLISIS FINANCIERO ---
class AnalisisFinanciero:
    """Clase Padre 3: Representa un análisis matemático base sobre un conjunto de datos."""
    def __init__(self, lista_valores):
        # Protegemos la lista de valores numéricos
        self._valores = lista_valores

class AnalisisTRM(AnalisisFinanciero):
    """Clase Hijo 3: Contiene la lógica de negocio específica para analizar la TRM."""
    def __init__(self, lista_valores):
        super().__init__(lista_valores)

    # Método para calcular el promedio de la semana
    def calcular_promedio(self):
        if len(self._valores) == 0:
            return 0.0
        return sum(self._valores) / len(self._valores)

    # Método para calcular la volatilidad (Máximo - Mínimo)
    def calcular_volatilidad(self):
        if len(self._valores) == 0:
            return 0.0
        return max(self._valores) - min(self._valores)

    # Método que usa un ciclo FOR para generar alertas de compra/venta
    def generar_alertas(self):
        promedio = self.calcular_promedio()
        dias_venta = 0
        dias_compra = 0
        
        # Recorremos los valores para compararlos contra el promedio
        for trm in self._valores:
            if trm > promedio:
                dias_venta += 1  # Alerta de Venta
            elif trm < promedio:
                dias_compra += 1 # Alerta de Compra
                
        return dias_venta, dias_compra


# ==============================================================================
# 2. PERSISTENCIA DE DATOS (SQLite) - INICIALIZACIÓN Y POBLADO
# ==============================================================================

def inicializar_base_datos():
    """Crea las tablas y asegura el mínimo de 5 registros por tabla."""
    try:
        # El bloque 'with' abre y cierra la conexión automáticamente de forma segura
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Activamos las llaves foráneas (Obligatorio en SQLite)
            cursor.execute("PRAGMA foreign_keys = ON;")

            # Creación de Tabla 1: Inversionistas (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS inversionistas (
                    id_inversionista INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre TEXT NOT NULL,
                    perfil_riesgo TEXT NOT NULL
                )
            ''')

            # Creación de Tabla 2: Monedas (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS monedas (
                    codigo TEXT PRIMARY KEY,
                    nombre_moneda TEXT NOT NULL
                )
            ''')

            # Creación de Tabla 3: Registros TRM (Tiene 2 FK, relaciona Inversionistas y Monedas)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS registros_trm (
                    id_registro INTEGER PRIMARY KEY AUTOINCREMENT,
                    dia INTEGER NOT NULL,
                    valor_trm REAL NOT NULL,
                    codigo_moneda TEXT,
                    id_inversionista INTEGER,
                    FOREIGN KEY (codigo_moneda) REFERENCES monedas(codigo),
                    FOREIGN KEY (id_inversionista) REFERENCES inversionistas(id_inversionista)
                )
            ''')

            # Inserción de 5 registros mínimos (Regla de Oro)
            # Verificamos si la tabla inversionistas está vacía para no duplicar datos
            cursor.execute("SELECT COUNT(*) FROM inversionistas")
            if cursor.fetchone()[0] == 0:
                inversionistas_data =[
                    ('Diego Zuluaga', 'Conservador'),
                    ('Ana Gómez', 'Agresivo'),
                    ('Carlos Ruiz', 'Moderado'),
                    ('María López', 'Agresivo'),
                    ('Luis Pérez', 'Conservador')
                ]
                cursor.executemany("INSERT INTO inversionistas (nombre, perfil_riesgo) VALUES (?, ?)", inversionistas_data)

            # Verificamos si la tabla monedas está vacía
            cursor.execute("SELECT COUNT(*) FROM monedas")
            if cursor.fetchone()[0] == 0:
                monedas_data =[
                    ('USD', 'Dólar Estadounidense'),
                    ('EUR', 'Euro'),
                    ('GBP', 'Libra Esterlina'),
                    ('JPY', 'Yen Japonés'),
                    ('CAD', 'Dólar Canadiense')
                ]
                cursor.executemany("INSERT INTO monedas (codigo, nombre_moneda) VALUES (?, ?)", monedas_data)

            # Verificamos si la tabla registros_trm está vacía
            cursor.execute("SELECT COUNT(*) FROM registros_trm")
            if cursor.fetchone()[0] == 0:
                # Insertamos 5 registros iniciales simulando días anteriores
                trm_data =[
                    (1, 3900.50, 'USD', 1),
                    (2, 3910.00, 'USD', 1),
                    (3, 3895.20, 'USD', 1),
                    (4, 3920.10, 'USD', 1),
                    (5, 3905.00, 'USD', 1)
                ]
                cursor.executemany('''
                    INSERT INTO registros_trm (dia, valor_trm, codigo_moneda, id_inversionista) 
                    VALUES (?, ?, ?, ?)
                ''', trm_data)

            conn.commit()
    except sqlite3.Error as e:
        print(f"⚠️ Error al inicializar la base de datos: {e}")

def mostrar_datos_iniciales():
    """Imprime todos los registros creados en las tablas de prueba al iniciar el programa."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            print("\n" + "="*70)
            print("📂 DATOS INICIALES DISPONIBLES EN LA BASE DE DATOS")
            print("="*70)
            
            # Mostrar Inversionistas
            cursor.execute("SELECT * FROM inversionistas")
            print("👤 INVERSIONISTAS (ID | Nombre | Perfil):")
            for inv in cursor.fetchall(): print(f"   {inv[0]} | {inv[1]} | {inv[2]}")
            
            # Mostrar Monedas
            cursor.execute("SELECT * FROM monedas")
            print("\n💵 MONEDAS (Código | Nombre):")
            for mon in cursor.fetchall(): print(f"   {mon[0]} | {mon[1]}")
            
            # Mostrar Registros TRM
            cursor.execute("SELECT * FROM registros_trm")
            print("\n📈 REGISTROS TRM (ID | Día | Valor | Moneda | ID Inversionista):")
            for reg in cursor.fetchall(): print(f"   {reg[0]} | Día {reg[1]} | ${reg[2]:,.2f} | {reg[3]} | Inv: {reg[4]}")
            print("="*70 + "\n")
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar datos iniciales: {e}")


# ==============================================================================
# 3. OPERACIONES CRUD Y MENÚ INTERACTIVO
# ==============================================================================

def registrar_semana_trm():
    """[CREATE] Registra 7 días de TRM en la base de datos."""
    try:
        print("\n--- 📝 REGISTRAR SEMANA DE TRM ---")
        id_inversionista = int(input("ID del Inversionista (Ej: 1 al 5): "))
        codigo_moneda = input("Código de la Moneda (Ej: USD, EUR): ").strip().upper()
        
        valores_ingresados =[]

        # Pedimos exactamente 7 días usando un ciclo for
        for dia in range(1, 8):
            valor = float(input(f"   Ingrese la TRM del Día {dia} en COP: "))
            
            # Validación lógica: La TRM no puede ser negativa
            if valor <= 0:
                print("   ❌ Error: La TRM debe ser mayor a 0. Intente de nuevo toda la operación.")
                return
            
            # Guardamos temporalmente en una lista de tuplas para insertarlos luego
            valores_ingresados.append((dia, valor, codigo_moneda, id_inversionista))

        # GUARDAR EN BASE DE DATOS (Transacción)
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("PRAGMA foreign_keys = ON;")
            
            # Insertamos los 7 registros de una vez
            cursor.executemany('''
                INSERT INTO registros_trm (dia, valor_trm, codigo_moneda, id_inversionista)
                VALUES (?, ?, ?, ?)
            ''', valores_ingresados)
            
            conn.commit()
            print("\n✅ Los 7 días de TRM han sido registrados exitosamente en la Base de Datos.")

    except ValueError:
        print("\n⚠️ Error: Ingresó un texto donde se esperaba un número.")
    except sqlite3.IntegrityError:
        print("\n⚠️ Error de Integridad: El ID del inversionista o el código de moneda no existen.")
    except sqlite3.Error as e:
        print(f"\n⚠️ Error de base de datos: {e}")


def analizar_trm():
    """[READ] Muestra un reporte cruzando las 3 tablas usando INNER JOIN y aplica POO para el análisis."""
    try:
        id_inv = int(input("\nIngrese el ID del Inversionista para analizar su historial: "))
        cod_moneda = input("Ingrese el Código de la Moneda a analizar (Ej: USD): ").strip().upper()

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Consulta JOIN uniendo las 3 tablas para traer nombres reales en lugar de solo IDs
            consulta = '''
                SELECT 
                    registros_trm.dia,
                    registros_trm.valor_trm,
                    monedas.nombre_moneda,
                    inversionistas.nombre
                FROM registros_trm
                INNER JOIN monedas ON registros_trm.codigo_moneda = monedas.codigo
                INNER JOIN inversionistas ON registros_trm.id_inversionista = inversionistas.id_inversionista
                WHERE registros_trm.id_inversionista = ? AND registros_trm.codigo_moneda = ?
                ORDER BY registros_trm.id_registro ASC
            '''
            cursor.execute(consulta, (id_inv, cod_moneda))
            registros = cursor.fetchall()

            if not registros:
                print("⚠️ No hay registros para este inversionista y moneda.")
                return

            print("\n" + "="*80)
            print("📄 REPORTE DE MERCADO Y ANÁLISIS DE TRM (INNER JOIN)")
            print("="*80)
            
            # Extraemos el nombre del inversionista y la moneda del primer registro
            nombre_inversionista = registros[0][3]
            nombre_moneda = registros[0][2]
            
            print(f"👤 Inversionista: {nombre_inversionista} | 💵 Moneda: {nombre_moneda}\n")
            
            lista_valores_trm =[]
            
            # Mostramos el detalle día a día
            for r in registros:
                dia = r[0]
                valor = r[1]
                lista_valores_trm.append(valor) # Guardamos el valor para el análisis POO
                print(f"   📅 Día {dia}: ${valor:,.2f} COP")
            
            # --- APLICACIÓN DE POO PARA EL ANÁLISIS ---
            # Instanciamos nuestra clase AnalisisTRM pasándole la lista de valores extraída de la BD
            analisis = AnalisisTRM(lista_valores_trm)
            
            promedio = analisis.calcular_promedio()
            volatilidad = analisis.calcular_volatilidad()
            dias_venta, dias_compra = analisis.generar_alertas()
            
            print("-" * 80)
            print(f"📊 Promedio de la semana: ${promedio:,.2f} COP")
            print(f"⚡ Volatilidad (Max - Min): ${volatilidad:,.2f} COP")
            print(f"🔴 Alertas de Venta (Días > Promedio): {dias_venta} días")
            print(f"🟢 Alertas de Compra (Días < Promedio): {dias_compra} días")
            print("="*80)
            
    except ValueError:
        print("\n⚠️ Error: Ingresó un formato incorrecto.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar reporte: {e}")


def actualizar_registro_trm():
    """[UPDATE] Actualiza el valor de un registro de TRM específico."""
    try:
        id_registro = int(input("\nIngrese el ID del registro TRM a actualizar: "))
        nuevo_valor = float(input("Ingrese el nuevo valor de la TRM en COP: "))

        if nuevo_valor <= 0:
            print("❌ Error: La TRM no puede ser negativa o cero.")
            return

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("UPDATE registros_trm SET valor_trm = ? WHERE id_registro = ?", (nuevo_valor, id_registro))
            
            # Verificamos si realmente se actualizó alguna fila
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Registro #{id_registro} actualizado exitosamente a ${nuevo_valor:,.2f}.")
            else:
                print(f"⚠️ No se encontró ningún registro con el ID {id_registro}.")
                
    except ValueError:
        print("⚠️ Error: Debe ingresar valores numéricos válidos.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al actualizar: {e}")


def eliminar_registro_trm():
    """[DELETE] Elimina un registro de TRM específico."""
    try:
        id_registro = int(input("\nIngrese el ID del registro TRM a eliminar: "))

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("PRAGMA foreign_keys = ON;") 
            
            cursor.execute("DELETE FROM registros_trm WHERE id_registro = ?", (id_registro,))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Registro TRM #{id_registro} eliminado exitosamente.")
            else:
                print(f"⚠️ No se encontró ningún registro con el ID {id_registro}.")
                
    except ValueError:
        print("⚠️ Error: El ID debe ser un número entero.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al eliminar: {e}")


def menu_principal():
    """Controlador principal del programa (Ciclo While)."""
    # 1. Preparamos la base de datos
    inicializar_base_datos()
    
    # 2. Imprimimos el menú y los registros iniciales creados
    mostrar_datos_iniciales()
    
    # 3. Bucle infinito para mantener el programa vivo
    while True:
        print("\n" + "*"*45)
        print("📈 SISTEMA FINANCIERO 'DOLAR-TRACK' 📉")
        print("*"*45)
        print("1. [CREATE] Registrar semana de TRM (7 días)")
        print("2. [READ] Analizar TRM y Volatilidad (JOIN + POO)")
        print("3. [UPDATE] Corregir un registro de TRM")
        print("4. [DELETE] Eliminar un registro de TRM")
        print("5. Salir del sistema")
        
        opcion = input("\n>> Seleccione una opción (1-5): ").strip()
        
        if opcion == "1":
            registrar_semana_trm()
        elif opcion == "2":
            analizar_trm()
        elif opcion == "3":
            actualizar_registro_trm()
        elif opcion == "4":
            eliminar_registro_trm()
        elif opcion == "5":
            print("\nCerrando el sistema Dolar-Track... ¡Éxitos en sus inversiones! 💼")
            break
        else:
            print("\n❌ Opción no válida. Por favor, digite un número del 1 al 5.")

# ==============================================================================
# 4. PUNTO DE ENTRADA DEL SCRIPT
# ==============================================================================
if __name__ == "__main__":
    menu_principal()
```

Problema: Una tienda de ropa quiere lanzar un programa de lealtad pero no sabe quiénes son sus mejores clientes.

### Funcionalidad:

Registre el nombre y el valor total de compra de 5 clientes.

Asigne una categoría: "Bronce" (< $100k), "Plata" (< $500k), "Oro" (> $500k).

Si es "Oro", calcule un bono de regalo del 10% de su compra.

Imprima el reporte de clientes ordenado alfabéticamente para enviar correos.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# TALLER CORTE 2 - RETO 4: MERCADEO "FIDELIZA-PUNTOS"
# ARQUITECTURA: POO (3 Padres, 3 Hijos, Encapsulamiento) + SQLite (CRUD, JOIN, PK/FK)
# ==============================================================================

import sqlite3
from datetime import datetime

# Definimos el nombre de la base de datos como una constante global
DB_NAME = "reto_4_fideliza_puntos.db"

# ==============================================================================
# 1. PROGRAMACIÓN ORIENTADA A OBJETOS (POO) - JERARQUÍA DE CLASES
# ==============================================================================

# --- JERARQUÍA 1: PERSONAS Y CLIENTES ---
class Persona:
    """Clase Padre 1: Representa a una persona genérica."""
    def __init__(self, nombre):
        # Atributo protegido: accesible por las clases hijas
        self._nombre = nombre

class ClienteFidelizado(Persona):
    """Clase Hijo 1: Representa a un cliente del programa de lealtad."""
    def __init__(self, nombre, email, compras_acumuladas=0.0):
        super().__init__(nombre)
        # Encapsulamiento: Atributos privados
        self.__email = email
        self.__compras_acumuladas = compras_acumuladas
        self.__categoria = "Bronce"
        self.__bono_regalo = 0.0
        # Al instanciar, calculamos automáticamente su categoría y bono
        self._actualizar_estado_lealtad()

    # Método privado para encapsular la lógica de negocio del Reto 4
    def _actualizar_estado_lealtad(self):
        """Asigna la categoría y calcula el bono del 10% si es Oro."""
        if self.__compras_acumuladas < 100000:
            self.__categoria = "Bronce"
            self.__bono_regalo = 0.0
        elif self.__compras_acumuladas < 500000:
            self.__categoria = "Plata"
            self.__bono_regalo = 0.0
        else:
            self.__categoria = "Oro"
            # Bono del 10% sobre el total de sus compras
            self.__bono_regalo = self.__compras_acumuladas * 0.10

    # Getters para acceder a los datos de forma segura
    def get_nombre(self): return self._nombre
    def get_email(self): return self.__email
    def get_compras_acumuladas(self): return self.__compras_acumuladas
    def get_categoria(self): return self.__categoria
    def get_bono(self): return self.__bono_regalo


# --- JERARQUÍA 2: SEDES Y TIENDAS ---
class Sede:
    """Clase Padre 2: Representa una ubicación física genérica."""
    def __init__(self, ciudad):
        self._ciudad = ciudad

class TiendaFisica(Sede):
    """Clase Hijo 2: Representa una sucursal de la tienda de ropa."""
    def __init__(self, id_tienda, nombre_sucursal, ciudad):
        super().__init__(ciudad)
        self.__id_tienda = id_tienda
        self.__nombre_sucursal = nombre_sucursal

    def get_id(self): return self.__id_tienda
    def get_nombre(self): return self.__nombre_sucursal
    def get_ciudad(self): return self._ciudad


# --- JERARQUÍA 3: TRANSACCIONES Y COMPRAS ---
class Transaccion:
    """Clase Padre 3: Representa un movimiento financiero genérico."""
    def __init__(self, valor_total):
        self._valor_total = valor_total

class Compra(Transaccion):
    """Clase Hijo 3: Representa una compra realizada por un cliente en una tienda."""
    def __init__(self, id_cliente, id_tienda, valor_total):
        super().__init__(valor_total)
        self.__id_cliente = id_cliente
        self.__id_tienda = id_tienda
        # Generamos la fecha actual automáticamente
        self.__fecha = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    def get_valor(self): return self._valor_total
    def get_id_cliente(self): return self.__id_cliente
    def get_id_tienda(self): return self.__id_tienda
    def get_fecha(self): return self.__fecha


# ==============================================================================
# 2. PERSISTENCIA DE DATOS (SQLite) - INICIALIZACIÓN Y POBLADO
# ==============================================================================

def inicializar_base_datos():
    """Crea las tablas y asegura el mínimo de 5 registros por tabla."""
    try:
        # El bloque 'with' abre y cierra la conexión automáticamente de forma segura
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Activamos las llaves foráneas (Obligatorio en SQLite)
            cursor.execute("PRAGMA foreign_keys = ON;")

            # Creación de Tabla 1: Tiendas (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS tiendas (
                    id_tienda INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre_sucursal TEXT NOT NULL,
                    ciudad TEXT NOT NULL
                )
            ''')

            # Creación de Tabla 2: Clientes (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS clientes (
                    id_cliente INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre TEXT NOT NULL,
                    email TEXT UNIQUE,
                    compras_acumuladas REAL NOT NULL,
                    categoria TEXT NOT NULL,
                    bono_regalo REAL NOT NULL
                )
            ''')

            # Creación de Tabla 3: Compras (Tiene 2 FK, relaciona Clientes y Tiendas)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS compras (
                    id_compra INTEGER PRIMARY KEY AUTOINCREMENT,
                    id_cliente INTEGER,
                    id_tienda INTEGER,
                    valor_compra REAL NOT NULL,
                    fecha TEXT NOT NULL,
                    FOREIGN KEY (id_cliente) REFERENCES clientes(id_cliente) ON DELETE CASCADE,
                    FOREIGN KEY (id_tienda) REFERENCES tiendas(id_tienda)
                )
            ''')

            # Inserción de 5 registros mínimos (Regla de Oro)
            # Verificamos si la tabla tiendas está vacía para no duplicar datos
            cursor.execute("SELECT COUNT(*) FROM tiendas")
            if cursor.fetchone()[0] == 0:
                tiendas_data =[
                    ('Sucursal Norte', 'Bogotá'),
                    ('Sucursal Centro', 'Medellín'),
                    ('Sucursal Sur', 'Cali'),
                    ('Boutique VIP', 'Cartagena'),
                    ('Outlet Premium', 'Barranquilla')
                ]
                cursor.executemany("INSERT INTO tiendas (nombre_sucursal, ciudad) VALUES (?, ?)", tiendas_data)

            # Verificamos si la tabla clientes está vacía
            cursor.execute("SELECT COUNT(*) FROM clientes")
            if cursor.fetchone()[0] == 0:
                # Instanciamos objetos POO para calcular categorías y bonos automáticamente
                c1 = ClienteFidelizado('Ana Gómez', 'ana@correo.com', 85000)    # Bronce
                c2 = ClienteFidelizado('Carlos Ruiz', 'carlos@correo.com', 250000) # Plata
                c3 = ClienteFidelizado('María López', 'maria@correo.com', 600000)  # Oro
                c4 = ClienteFidelizado('Luis Pérez', 'luis@correo.com', 1200000)   # Oro
                c5 = ClienteFidelizado('Diana Rojas', 'diana@correo.com', 45000)   # Bronce
                
                clientes_data =[
                    (c.get_nombre(), c.get_email(), c.get_compras_acumuladas(), c.get_categoria(), c.get_bono())
                    for c in[c1, c2, c3, c4, c5]
                ]
                cursor.executemany('''
                    INSERT INTO clientes (nombre, email, compras_acumuladas, categoria, bono_regalo) 
                    VALUES (?, ?, ?, ?, ?)
                ''', clientes_data)

            # Verificamos si la tabla compras está vacía
            cursor.execute("SELECT COUNT(*) FROM compras")
            if cursor.fetchone()[0] == 0:
                compras_data =[
                    (1, 1, 85000, '2026-04-01 10:30:00'),
                    (2, 2, 250000, '2026-04-02 11:15:00'),
                    (3, 3, 600000, '2026-04-03 14:20:00'),
                    (4, 4, 1200000, '2026-04-04 16:45:00'),
                    (5, 5, 45000, '2026-04-05 09:10:00')
                ]
                cursor.executemany('''
                    INSERT INTO compras (id_cliente, id_tienda, valor_compra, fecha) 
                    VALUES (?, ?, ?, ?)
                ''', compras_data)

            conn.commit()
    except sqlite3.Error as e:
        print(f"⚠️ Error al inicializar la base de datos: {e}")

def mostrar_datos_iniciales():
    """Imprime todos los registros creados en las tablas de prueba al iniciar el programa."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            print("\n" + "="*80)
            print("📂 DATOS INICIALES DISPONIBLES EN LA BASE DE DATOS")
            print("="*80)
            
            # Mostrar Tiendas
            cursor.execute("SELECT * FROM tiendas")
            print("🏬 TIENDAS (ID | Nombre | Ciudad):")
            for t in cursor.fetchall(): print(f"   {t[0]} | {t[1]} | {t[2]}")
            
            # Mostrar Clientes
            cursor.execute("SELECT * FROM clientes")
            print("\n👥 CLIENTES (ID | Nombre | Email | Compras Acumuladas | Categoría | Bono):")
            for c in cursor.fetchall(): print(f"   {c[0]} | {c[1]} | {c[2]} | ${c[3]:,.0f} | {c[4]} | Bono: ${c[5]:,.0f}")
            
            # Mostrar Compras
            cursor.execute("SELECT * FROM compras")
            print("\n🛍️ COMPRAS (ID | ID Cliente | ID Tienda | Valor | Fecha):")
            for comp in cursor.fetchall(): print(f"   {comp[0]} | Cliente: {comp[1]} | Tienda: {comp[2]} | ${comp[3]:,.0f} | {comp[4]}")
            print("="*80 + "\n")
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar datos iniciales: {e}")


# ==============================================================================
# 3. OPERACIONES CRUD Y MENÚ INTERACTIVO
# ==============================================================================

def registrar_lote_clientes():
    """[CREATE] Registra 5 clientes nuevos y sus compras iniciales (Requisito del Reto 4)."""
    try:
        print("\n--- 📝 REGISTRAR LOTE DE 5 CLIENTES NUEVOS ---")
        id_tienda = int(input("Ingrese el ID de la Tienda donde se realizan las compras (Ej: 1 al 5): "))
        
        nuevos_clientes = []
        nuevas_compras =[]

        # Pedimos exactamente 5 clientes usando un ciclo for
        for i in range(1, 6):
            print(f"\n🔸 Cliente #{i}")
            nombre = input("   Nombre completo: ").strip().title()
            email = input("   Correo electrónico: ").strip().lower()
            valor_compra = float(input("   Valor total de la compra en COP: "))
            
            if valor_compra < 0:
                print("   ❌ Error: El valor de compra no puede ser negativo.")
                return

            # Instanciamos el objeto POO para que calcule categoría y bono automáticamente
            cliente_obj = ClienteFidelizado(nombre, email, valor_compra)
            
            # Guardamos los datos extraídos del objeto para la tabla clientes
            nuevos_clientes.append((
                cliente_obj.get_nombre(), 
                cliente_obj.get_email(), 
                cliente_obj.get_compras_acumuladas(), 
                cliente_obj.get_categoria(), 
                cliente_obj.get_bono()
            ))
            
            # Guardamos el valor de la compra para la tabla compras (el ID del cliente se asignará en BD)
            nuevas_compras.append(valor_compra)

        # GUARDAR EN BASE DE DATOS (Transacción)
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("PRAGMA foreign_keys = ON;")
            
            for i in range(5):
                # 1. Insertamos al cliente
                cursor.execute('''
                    INSERT INTO clientes (nombre, email, compras_acumuladas, categoria, bono_regalo)
                    VALUES (?, ?, ?, ?, ?)
                ''', nuevos_clientes[i])
                
                # 2. Recuperamos el ID autogenerado del cliente
                id_cliente_generado = cursor.lastrowid
                
                # 3. Instanciamos el objeto Compra (POO)
                compra_obj = Compra(id_cliente_generado, id_tienda, nuevas_compras[i])
                
                # 4. Insertamos la compra vinculada al cliente y a la tienda
                cursor.execute('''
                    INSERT INTO compras (id_cliente, id_tienda, valor_compra, fecha)
                    VALUES (?, ?, ?, ?)
                ''', (compra_obj.get_id_cliente(), compra_obj.get_id_tienda(), compra_obj.get_valor(), compra_obj.get_fecha()))
            
            conn.commit()
            print("\n✅ Los 5 clientes y sus compras han sido registrados y clasificados exitosamente.")

    except ValueError:
        print("\n⚠️ Error: Ingresó un texto donde se esperaba un número.")
    except sqlite3.IntegrityError:
        print("\n⚠️ Error de Integridad: El correo ya existe o el ID de la tienda es inválido.")
    except sqlite3.Error as e:
        print(f"\n⚠️ Error de base de datos: {e}")


def ver_reporte_lealtad():
    """[READ] Muestra el reporte de clientes ordenado alfabéticamente usando INNER JOIN."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Consulta JOIN uniendo clientes y compras, agrupando por cliente
            # Ordenamos alfabéticamente por el nombre del cliente (Requisito del Reto 4)
            consulta = '''
                SELECT 
                    clientes.id_cliente,
                    clientes.nombre,
                    clientes.email,
                    clientes.categoria,
                    clientes.compras_acumuladas,
                    clientes.bono_regalo,
                    COUNT(compras.id_compra) AS total_transacciones
                FROM clientes
                LEFT JOIN compras ON clientes.id_cliente = compras.id_cliente
                GROUP BY clientes.id_cliente
                ORDER BY clientes.nombre ASC
            '''
            cursor.execute(consulta)
            registros = cursor.fetchall()

            print("\n" + "="*90)
            print("💌 REPORTE DE LEALTAD PARA EMAIL MARKETING (ORDEN ALFABÉTICO)")
            print("="*90)
            
            if not registros:
                print("No hay clientes registrados.")
            else:
                for r in registros:
                    id_cliente = r[0]
                    nombre = r[1]
                    email = r[2]
                    categoria = r[3]
                    compras = r[4]
                    bono = r[5]
                    transacciones = r[6]
                    
                    # Formateamos el mensaje dependiendo de si el cliente tiene bono o no
                    if categoria == "Oro":
                        mensaje_bono = f"🎁 ¡Bono de Regalo ganado!: ${bono:,.2f}"
                    else:
                        mensaje_bono = "Sigue comprando para llegar a Oro."
                        
                    print(f"ID: {id_cliente} | 👤 {nombre.ljust(20)} | 📧 {email.ljust(25)} | 🏆 {categoria.upper().ljust(6)}")
                    print(f"   ↳ Compras Totales: ${compras:,.2f} ({transacciones} transacciones) | {mensaje_bono}\n")
            print("="*90)
            
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar reporte: {e}")


def actualizar_email_cliente():
    """[UPDATE] Actualiza el correo electrónico de un cliente existente."""
    try:
        id_cliente = int(input("\nIngrese el ID del cliente a actualizar: "))
        nuevo_email = input("Ingrese el nuevo correo electrónico: ").strip().lower()

        if "@" not in nuevo_email or "." not in nuevo_email:
            print("❌ Error: Ingrese un formato de correo válido.")
            return

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("UPDATE clientes SET email = ? WHERE id_cliente = ?", (nuevo_email, id_cliente))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Correo del cliente #{id_cliente} actualizado a '{nuevo_email}'.")
            else:
                print(f"⚠️ No se encontró ningún cliente con el ID {id_cliente}.")
                
    except ValueError:
        print("⚠️ Error: Debe ingresar un ID numérico válido.")
    except sqlite3.IntegrityError:
        print("⚠️ Error: Ese correo electrónico ya está registrado por otro cliente.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al actualizar: {e}")


def eliminar_cliente():
    """[DELETE] Elimina un cliente y sus compras asociadas (ON DELETE CASCADE)."""
    try:
        id_cliente = int(input("\nIngrese el ID del cliente a eliminar del programa de lealtad: "))

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            # Vital para que funcione el borrado en cascada de las compras asociadas
            cursor.execute("PRAGMA foreign_keys = ON;") 
            
            cursor.execute("DELETE FROM clientes WHERE id_cliente = ?", (id_cliente,))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Cliente #{id_cliente} y su historial de compras eliminados exitosamente.")
            else:
                print(f"⚠️ No se encontró ningún cliente con el ID {id_cliente}.")
                
    except ValueError:
        print("⚠️ Error: El ID debe ser un número entero.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al eliminar: {e}")


def menu_principal():
    """Controlador principal del programa (Ciclo While)."""
    # 1. Preparamos la base de datos
    inicializar_base_datos()
    
    # 2. Imprimimos los registros iniciales creados
    mostrar_datos_iniciales()
    
    # 3. Bucle infinito para mantener el programa vivo
    while True:
        print("\n" + "*"*45)
        print("🛍️ SISTEMA DE LEALTAD 'FIDELIZA-PUNTOS' 🛍️")
        print("*"*45)
        print("1. [CREATE] Registrar lote de 5 clientes nuevos")
        print("2. [READ] Ver Reporte de Lealtad (Alfabético)")
        print("3. [UPDATE] Actualizar correo de un cliente")
        print("4. [DELETE] Eliminar un cliente del programa")
        print("5. Salir del sistema")
        
        opcion = input("\n>> Seleccione una opción (1-5): ").strip()
        
        if opcion == "1":
            registrar_lote_clientes()
        elif opcion == "2":
            ver_reporte_lealtad()
        elif opcion == "3":
            actualizar_email_cliente()
        elif opcion == "4":
            eliminar_cliente()
        elif opcion == "5":
            print("\nCerrando el sistema Fideliza-Puntos... ¡Aumentar las ventas es el objetivo! 🚀")
            break
        else:
            print("\n❌ Opción no válida. Por favor, digite un número del 1 al 5.")

# ==============================================================================
# 4. PUNTO DE ENTRADA DEL SCRIPT
# ==============================================================================
if __name__ == "__main__":
    menu_principal()
```

# 🏢 RETO 5: ADMINISTRACIÓN - "Nomina-Express"

Perfil: Administración de Empresas / Servicio.

Problema: Una PyME necesita liquidar la nómina semanal de sus empleados operativos teniendo en cuenta horas extra.

### Funcionalidad:

Ingrese nombre y horas trabajadas de 5 empleados.

Si horas > 48, las adicionales se pagan como extras (+25% del valor hora).

Si el sueldo total es < 2 SMMLV, agregue Auxilio de Transporte (Simulado).

Muestre el desprendible de pago total de la empresa.

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# TALLER CORTE 2 - RETO 5: ADMINISTRACIÓN "NOMINA-EXPRESS"
# ARQUITECTURA: POO (3 Padres, 3 Hijos, Encapsulamiento) + SQLite (CRUD, JOIN, PK/FK)
# ==============================================================================

import sqlite3
from datetime import datetime

# Definimos el nombre de la base de datos como una constante global
DB_NAME = "reto_5_nomina_express.db"

# Constantes económicas para la liquidación (Simuladas para 2026)
SMMLV_MENSUAL = 2000000
AUX_TRANS_MENSUAL = 249095
# Como la nómina es semanal, dividimos los topes mensuales entre 4
TOPE_2_SMMLV_SEMANAL = (SMMLV_MENSUAL * 2) / 4
AUX_TRANS_SEMANAL = AUX_TRANS_MENSUAL / 4

# ==============================================================================
# 1. PROGRAMACIÓN ORIENTADA A OBJETOS (POO) - JERARQUÍA DE CLASES
# ==============================================================================

# --- JERARQUÍA 1: PERSONAS Y EMPLEADOS ---
class Persona:
    """Clase Padre 1: Representa a un individuo genérico."""
    def __init__(self, nombre):
        # Atributo protegido: accesible por las clases hijas
        self._nombre = nombre

class EmpleadoOperativo(Persona):
    """Clase Hijo 1: Representa a un trabajador de la PyME."""
    def __init__(self, nombre, valor_hora, id_departamento):
        super().__init__(nombre)
        # Encapsulamiento: Atributos privados
        self.__valor_hora = valor_hora
        self.__id_departamento = id_departamento

    # Getters para acceder a los datos de forma segura
    def get_nombre(self): return self._nombre
    def get_valor_hora(self): return self.__valor_hora
    def get_id_departamento(self): return self.__id_departamento


# --- JERARQUÍA 2: ESTRUCTURA ORGANIZACIONAL ---
class AreaEmpresa:
    """Clase Padre 2: Representa una división genérica de la empresa."""
    def __init__(self, nombre_area):
        self._nombre_area = nombre_area

class Departamento(AreaEmpresa):
    """Clase Hijo 2: Representa un departamento específico."""
    def __init__(self, id_depto, nombre_area):
        super().__init__(nombre_area)
        self.__id_depto = id_depto

    def get_id(self): return self.__id_depto
    def get_nombre(self): return self._nombre_area


# --- JERARQUÍA 3: FINANZAS Y LIQUIDACIÓN ---
class TransaccionFinanciera:
    """Clase Padre 3: Representa un movimiento de dinero en la empresa."""
    def __init__(self, total_pagar):
        self._total_pagar = total_pagar

class LiquidacionSemanal(TransaccionFinanciera):
    """Clase Hijo 3: Contiene la lógica de negocio para liquidar la nómina."""
    def __init__(self, empleado, horas_trabajadas):
        # Recibimos un OBJETO de tipo EmpleadoOperativo (Interacción entre objetos)
        self.__empleado = empleado
        self.__horas_trabajadas = horas_trabajadas
        
        # Atributos que se calcularán automáticamente
        self.__horas_normales = 0
        self.__horas_extras = 0
        self.__pago_base = 0.0
        self.__pago_extras = 0.0
        self.__auxilio_transporte = 0.0
        
        # Ejecutamos la lógica de liquidación al instanciar
        self._calcular_nomina()
        
        # Llamamos al constructor padre con el total final calculado
        total_final = self.__pago_base + self.__pago_extras + self.__auxilio_transporte
        super().__init__(total_final)

    def _calcular_nomina(self):
        """Lógica de negocio encapsulada: Horas extra y Auxilio de Transporte."""
        valor_hora = self.__empleado.get_valor_hora()
        
        # 1. Cálculo de horas (Tope legal: 48h)
        if self.__horas_trabajadas > 48:
            self.__horas_normales = 48
            self.__horas_extras = self.__horas_trabajadas - 48
        else:
            self.__horas_normales = self.__horas_trabajadas
            self.__horas_extras = 0
            
        # 2. Cálculo de pagos (+25% para extras)
        self.__pago_base = self.__horas_normales * valor_hora
        self.__pago_extras = self.__horas_extras * (valor_hora * 1.25)
        
        # 3. Cálculo de Auxilio de Transporte (Si gana menos de 2 SMMLV)
        subtotal_salarial = self.__pago_base + self.__pago_extras
        if subtotal_salarial < TOPE_2_SMMLV_SEMANAL:
            self.__auxilio_transporte = AUX_TRANS_SEMANAL

    # Getters para extraer los resultados y guardarlos en la BD
    def get_horas_normales(self): return self.__horas_normales
    def get_horas_extras(self): return self.__horas_extras
    def get_auxilio_transporte(self): return self.__auxilio_transporte
    def get_total_pagar(self): return self._total_pagar


# ==============================================================================
# 2. PERSISTENCIA DE DATOS (SQLite) - INICIALIZACIÓN Y POBLADO
# ==============================================================================

def inicializar_base_datos():
    """Crea las tablas y asegura el mínimo de 5 registros por tabla."""
    try:
        # El bloque 'with' asegura que la conexión se cierre correctamente
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Activamos las llaves foráneas (Obligatorio en SQLite)
            cursor.execute("PRAGMA foreign_keys = ON;")

            # Creación de Tabla 1: Departamentos (No tiene FK)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS departamentos (
                    id_depto INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre_area TEXT NOT NULL UNIQUE
                )
            ''')

            # Creación de Tabla 2: Empleados (Tiene 1 FK -> Departamentos)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS empleados (
                    id_empleado INTEGER PRIMARY KEY AUTOINCREMENT,
                    nombre TEXT NOT NULL,
                    valor_hora REAL NOT NULL,
                    id_depto INTEGER,
                    FOREIGN KEY (id_depto) REFERENCES departamentos(id_depto)
                )
            ''')

            # Creación de Tabla 3: Nominas (Tiene 1 FK -> Empleados)
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS nominas (
                    id_nomina INTEGER PRIMARY KEY AUTOINCREMENT,
                    id_empleado INTEGER,
                    horas_normales REAL NOT NULL,
                    horas_extras REAL NOT NULL,
                    auxilio_transporte REAL NOT NULL,
                    total_pagado REAL NOT NULL,
                    fecha TEXT NOT NULL,
                    FOREIGN KEY (id_empleado) REFERENCES empleados(id_empleado) ON DELETE CASCADE
                )
            ''')

            # Inserción de 5 registros mínimos (Regla de Oro)
            cursor.execute("SELECT COUNT(*) FROM departamentos")
            if cursor.fetchone()[0] == 0:
                deptos_data = [('Producción',), ('Logística',), ('Mantenimiento',), ('Empaque',), ('Calidad',)]
                cursor.executemany("INSERT INTO departamentos (nombre_area) VALUES (?)", deptos_data)

            cursor.execute("SELECT COUNT(*) FROM empleados")
            if cursor.fetchone()[0] == 0:
                empleados_data =[
                    ('Carlos Ruiz', 9500, 1),
                    ('Ana Gómez', 10000, 2),
                    ('Luis Pérez', 8500, 3),
                    ('María López', 9000, 4),
                    ('Jorge Díaz', 11000, 5)
                ]
                cursor.executemany("INSERT INTO empleados (nombre, valor_hora, id_depto) VALUES (?, ?, ?)", empleados_data)

            cursor.execute("SELECT COUNT(*) FROM nominas")
            if cursor.fetchone()[0] == 0:
                nominas_data =[
                    (1, 48, 2, 40500, 480000, '2026-04-01'),
                    (2, 48, 0, 40500, 480000, '2026-04-01'),
                    (3, 40, 0, 40500, 340000, '2026-04-01'),
                    (4, 48, 5, 40500, 488250, '2026-04-01'),
                    (5, 48, 10, 0, 665500, '2026-04-01') # Gana más de 2 SMMLV, no tiene auxilio
                ]
                cursor.executemany('''
                    INSERT INTO nominas (id_empleado, horas_normales, horas_extras, auxilio_transporte, total_pagado, fecha) 
                    VALUES (?, ?, ?, ?, ?, ?)
                ''', nominas_data)

            conn.commit()
    except sqlite3.Error as e:
        print(f"⚠️ Error al inicializar la base de datos: {e}")

def mostrar_datos_iniciales():
    """Imprime todos los registros creados en las tablas de prueba al iniciar el programa."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            print("\n" + "="*85)
            print("📂 DATOS INICIALES DISPONIBLES EN LA BASE DE DATOS")
            print("="*85)
            
            cursor.execute("SELECT * FROM departamentos")
            print("🏢 DEPARTAMENTOS (ID | Nombre):")
            for d in cursor.fetchall(): print(f"   {d[0]} | {d[1]}")
            
            cursor.execute("SELECT * FROM empleados")
            print("\n👷 EMPLEADOS (ID | Nombre | Valor Hora | ID Depto):")
            for e in cursor.fetchall(): print(f"   {e[0]} | {e[1].ljust(15)} | ${e[2]:,.0f}/h | Depto: {e[3]}")
            
            cursor.execute("SELECT * FROM nominas")
            print("\n💰 NÓMINAS HISTÓRICAS (ID | ID Emp | H.Norm | H.Ext | Aux.Trans | Total | Fecha):")
            for n in cursor.fetchall(): 
                print(f"   {n[0]} | Emp: {n[1]} | {n[2]}h | {n[3]}h | ${n[4]:,.0f} | ${n[5]:,.0f} | {n[6]}")
            print("="*85 + "\n")
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar datos iniciales: {e}")


# ==============================================================================
# 3. OPERACIONES CRUD Y MENÚ INTERACTIVO
# ==============================================================================

def registrar_lote_nomina():
    """[CREATE] Registra 5 empleados nuevos y liquida su nómina semanal."""
    try:
        print("\n--- 📝 REGISTRAR LOTE DE 5 EMPLEADOS Y NÓMINA ---")
        id_depto = int(input("Ingrese el ID del Departamento para estos empleados (Ej: 1 al 5): "))
        
        nuevos_empleados = []
        nuevas_nominas =[]

        # Pedimos exactamente 5 empleados usando un ciclo for
        for i in range(1, 6):
            print(f"\n🔸 Empleado #{i}")
            nombre = input("   Nombre completo: ").strip().title()
            valor_hora = float(input("   Valor de la hora (COP): "))
            horas_trabajadas = float(input("   Horas totales trabajadas en la semana: "))
            
            if valor_hora <= 0 or horas_trabajadas < 0:
                print("   ❌ Error: Valores inválidos. Operación cancelada.")
                return

            # 1. Instanciamos el Empleado (POO)
            empleado_obj = EmpleadoOperativo(nombre, valor_hora, id_depto)
            
            # 2. Instanciamos la Liquidación pasándole el objeto Empleado (POO)
            liquidacion_obj = LiquidacionSemanal(empleado_obj, horas_trabajadas)
            
            # Guardamos los datos extraídos para la tabla empleados
            nuevos_empleados.append((empleado_obj.get_nombre(), empleado_obj.get_valor_hora(), empleado_obj.get_id_departamento()))
            
            # Guardamos los datos de la nómina (el ID del empleado se asignará en BD)
            nuevas_nominas.append(liquidacion_obj)

        # GUARDAR EN BASE DE DATOS (Transacción)
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("PRAGMA foreign_keys = ON;")
            fecha_actual = datetime.now().strftime("%Y-%m-%d")
            
            for i in range(5):
                # Insertamos al empleado
                cursor.execute('''
                    INSERT INTO empleados (nombre, valor_hora, id_depto)
                    VALUES (?, ?, ?)
                ''', nuevos_empleados[i])
                
                # Recuperamos el ID autogenerado del empleado
                id_empleado_generado = cursor.lastrowid
                
                # Extraemos el objeto liquidación correspondiente
                liq = nuevas_nominas[i]
                
                # Insertamos la nómina vinculada al empleado
                cursor.execute('''
                    INSERT INTO nominas (id_empleado, horas_normales, horas_extras, auxilio_transporte, total_pagado, fecha)
                    VALUES (?, ?, ?, ?, ?, ?)
                ''', (id_empleado_generado, liq.get_horas_normales(), liq.get_horas_extras(), 
                      liq.get_auxilio_transporte(), liq.get_total_pagar(), fecha_actual))
            
            conn.commit()
            print("\n✅ Los 5 empleados y sus nóminas han sido liquidados y guardados exitosamente.")

    except ValueError:
        print("\n⚠️ Error: Ingresó un texto donde se esperaba un número.")
    except sqlite3.IntegrityError:
        print("\n⚠️ Error de Integridad: El ID del departamento no existe.")
    except sqlite3.Error as e:
        print(f"\n⚠️ Error de base de datos: {e}")


def ver_desprendibles_pago():
    """[READ] Muestra el reporte de nómina cruzando las 3 tablas usando INNER JOIN."""
    try:
        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            
            # Consulta JOIN uniendo nominas, empleados y departamentos
            consulta = '''
                SELECT 
                    empleados.id_empleado,
                    empleados.nombre,
                    departamentos.nombre_area,
                    empleados.valor_hora,
                    nominas.horas_normales,
                    nominas.horas_extras,
                    nominas.auxilio_transporte,
                    nominas.total_pagado,
                    nominas.fecha
                FROM nominas
                INNER JOIN empleados ON nominas.id_empleado = empleados.id_empleado
                INNER JOIN departamentos ON empleados.id_depto = departamentos.id_depto
                ORDER BY nominas.total_pagado DESC
            '''
            cursor.execute(consulta)
            registros = cursor.fetchall()

            print("\n" + "="*95)
            print("📄 DESPRENDIBLES DE PAGO SEMANAL (ORDENADOS POR TOTAL A PAGAR)")
            print("="*95)
            
            if not registros:
                print("No hay nóminas registradas.")
            else:
                total_empresa = 0.0
                for r in registros:
                    id_empleado = r[0]
                    nombre = r[1]
                    area = r[2]
                    v_hora = r[3]
                    h_norm = r[4]
                    h_ext = r[5]
                    aux = r[6]
                    total = r[7]
                    fecha = r[8]
                    
                    total_empresa += total
                    
                    print(f"ID: {id_empleado} | 👤 Empleado: {nombre.ljust(15)} | 🏢 Área: {area.ljust(15)} | 📅 Fecha: {fecha}")
                    print(f"   ↳ Tarifa: ${v_hora:,.0f}/h | H. Normales: {h_norm}h | H. Extras: {h_ext}h")
                    print(f"   ↳ Aux. Transporte: ${aux:,.0f} | 💰 TOTAL PAGADO: ${total:,.0f} COP\n")
                    
                print("-" * 95)
                print(f"🏦 COSTO TOTAL NÓMINA EMPRESA: ${total_empresa:,.0f} COP")
            print("="*95)
            
    except sqlite3.Error as e:
        print(f"⚠️ Error al consultar reporte: {e}")


def actualizar_valor_hora():
    """[UPDATE] Actualiza el valor de la hora de un empleado existente."""
    try:
        id_empleado = int(input("\nIngrese el ID del empleado a actualizar: "))
        nuevo_valor = float(input("Ingrese el nuevo valor de la hora en COP: "))

        if nuevo_valor <= 0:
            print("❌ Error: El valor de la hora debe ser mayor a 0.")
            return

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("UPDATE empleados SET valor_hora = ? WHERE id_empleado = ?", (nuevo_valor, id_empleado))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Valor de hora del empleado #{id_empleado} actualizado a ${nuevo_valor:,.0f}.")
            else:
                print(f"⚠️ No se encontró ningún empleado con el ID {id_empleado}.")
                
    except ValueError:
        print("⚠️ Error: Debe ingresar un valor numérico válido.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al actualizar: {e}")


def eliminar_nomina():
    """[DELETE] Elimina un registro de nómina específico."""
    try:
        id_nomina = int(input("\nIngrese el ID de la nómina a eliminar: "))

        with sqlite3.connect(DB_NAME) as conn:
            cursor = conn.cursor()
            cursor.execute("DELETE FROM nominas WHERE id_nomina = ?", (id_nomina,))
            
            if cursor.rowcount > 0:
                conn.commit()
                print(f"✅ Registro de nómina #{id_nomina} eliminado exitosamente.")
            else:
                print(f"⚠️ No se encontró ninguna nómina con el ID {id_nomina}.")
                
    except ValueError:
        print("⚠️ Error: El ID debe ser un número entero.")
    except sqlite3.Error as e:
        print(f"⚠️ Error al eliminar: {e}")


def menu_principal():
    """Controlador principal del programa (Ciclo While)."""
    # 1. Preparamos la base de datos
    inicializar_base_datos()
    
    # 2. Imprimimos los registros iniciales creados
    mostrar_datos_iniciales()
    
    # 3. Bucle infinito para mantener el programa vivo
    while True:
        print("\n" + "*"*45)
        print("🏢 SISTEMA DE RECURSOS HUMANOS 'NOMINA-EXPRESS' 🏢")
        print("*"*45)
        print("1.[CREATE] Registrar lote de 5 empleados y liquidar")
        print("2. [READ] Ver Desprendibles de Pago (JOIN)")
        print("3. [UPDATE] Actualizar tarifa por hora de un empleado")
        print("4. [DELETE] Eliminar un registro de nómina")
        print("5. Salir del sistema")
        
        opcion = input("\n>> Seleccione una opción (1-5): ").strip()
        
        if opcion == "1":
            registrar_lote_nomina()
        elif opcion == "2":
            ver_desprendibles_pago()
        elif opcion == "3":
            actualizar_valor_hora()
        elif opcion == "4":
            eliminar_nomina()
        elif opcion == "5":
            print("\nCerrando el sistema Nomina-Express... ¡Nómina liquidada con éxito! 🏦")
            break
        else:
            print("\n❌ Opción no válida. Por favor, digite un número del 1 al 5.")

# ==============================================================================
# 4. PUNTO DE ENTRADA DEL SCRIPT
# ==============================================================================
if __name__ == "__main__":
    menu_principal()
```

```python
# ==============================================================================
# IMPORTACIÓN DE LIBRERÍAS
# ==============================================================================
import pandas as pd
import sqlite3
```

```python
# ==============================================================================
# FASE 1: EXTRACCIÓN Y CREACIÓN DE ARCHIVOS (CSV)
# ==============================================================================
print("🔄 FASE 1: Creando y cargando archivos CSV sucios...")

# (Nota: Aquí asumimos que los strings 'cafeteria_productos_sucios', etc., 
# ya fueron ejecutados en la celda anterior del enunciado).

# 1. Guardamos los strings en archivos físicos .csv

with open('parcial_2_productos_sucios.csv', 'w', encoding='utf-8') as f:
    f.write(cafeteria_productos_sucios)

with open('parcial_2_clientes_sucios.csv', 'w', encoding='utf-8') as f:
    f.write(cafeteria_clientes_sucios)

with open('parcial_2_proveedores_sucios.csv', 'w', encoding='utf-8') as f:
    f.write(cafeteria_proveedores_sucios)

# 2. Cargamos los archivos en DataFrames de Pandas

df_productos = pd.read_csv('parcial_2_productos_sucios.csv')

df_clientes = pd.read_csv('parcial_2_clientes_sucios.csv')

df_proveedores = pd.read_csv('parcial_2_proveedores_sucios.csv')


print("✅ Archivos cargados exitosamente en memoria (DataFrames).")
```

```python
# FASE 2: Limpieza y Normalización (Pandas)
# Aplicaremos 1NF, 2NF y 3NF, estandarizaremos textos y manejaremos valores nulos (NaN).

print("\n🧹 FASE 2: Iniciando limpieza de datos con Pandas...")

# ------------------------------------------------------------------------------
# A. LIMPIEZA DE PRODUCTOS
# ------------------------------------------------------------------------------

# 1. Normalización 1NF: Separar 'producto_categoria'

df_productos[['Nombre_Producto', 'Categoria']] = df_productos['producto_categoria'].str.split(' - ', expand=True)
df_productos = df_productos.drop(columns=['producto_categoria'])

print(df_productos)
```

```python
# 2. Estandarización de texto

df_productos['Nombre_Producto'] = df_productos['Nombre_Producto'].str.title()
df_productos['Categoria'] = df_productos['Categoria'].str.title()

print(df_productos)
```

```python
# 3. Limpieza de precios (Quitar '$' y convertir a numérico)

df_productos['precio'] = df_productos['precio'].astype(str).str.replace('$', '', regex=False)
df_productos['precio'] = pd.to_numeric(df_productos['precio'])

print(df_productos)
```

```python
# 4. Manejo de Nulos (NaN) en Stock

# Asumimos que si el stock es NaN, es porque hay 0 unidades disponibles.
df_productos['stock'] = df_productos['stock'].fillna(0).astype(int)

print(df_productos)
```

```python
# ------------------------------------------------------------------------------
# B. LIMPIEZA DE CLIENTES
# ------------------------------------------------------------------------------

# 1. Normalización 1NF: Separar 'cliente_tipo'
df_clientes[['Nombre_Cliente', 'Tipo_Cliente']] = df_clientes['cliente_tipo'].str.split(' - ', expand=True)
df_clientes = df_clientes.drop(columns=['cliente_tipo'])
print(df_clientes)
```

```python
# 2. Estandarización de texto
df_clientes['Nombre_Cliente'] = df_clientes['Nombre_Cliente'].str.title()
df_clientes['Tipo_Cliente'] = df_clientes['Tipo_Cliente'].str.title()

print(df_clientes)
```

```python
# 3. Manejo de Nulos
df_clientes['email'] = df_clientes['email'].fillna('sin_correo@cafeteria.com')
df_clientes['telefono'] = df_clientes['telefono'].fillna('No Registra')

print(df_clientes)
```

```python
# 4. Normalización 3NF: Eliminar columna redundante 'edad'
# La edad se puede calcular a partir de la 'fecha_nacimiento', por lo que es redundante.
df_clientes = df_clientes.drop(columns=['edad'])

print(df_clientes)
```

```python
# ------------------------------------------------------------------------------
# C. LIMPIEZA DE PROVEEDORES
# ------------------------------------------------------------------------------

# 1. Normalización 1NF: Separar 'empresa_ciudad'

df_proveedores[['Nombre_Empresa', 'Ciudad']] = df_proveedores['empresa_ciudad'].str.split(' - ', expand=True)
df_proveedores = df_proveedores.drop(columns=['empresa_ciudad'])

print(df_proveedores)
```

```python
# 2. Manejo de Nulos

df_proveedores['contacto'] = df_proveedores['contacto'].fillna('Departamento de Ventas')
df_proveedores['telefono'] = df_proveedores['telefono'].fillna('No Registra')
df_proveedores['email'] = df_proveedores['email'].fillna('contacto@empresa.com')

print(df_proveedores)
```

```python
# ------------------------------------------------------------------------------
# D. EXPORTACIÓN DE DATOS LIMPIOS
# ------------------------------------------------------------------------------

df_productos.to_csv('parcial_2_productos_limpios.csv', index=False)
df_clientes.to_csv('parcial_2_clientes_limpios.csv', index=False)
df_proveedores.to_csv('parcial_2_proveedores_limpios.csv', index=False)

print("✅ Limpieza finalizada. Archivos limpios exportados al disco duro.")
```

```python
# # FASE 3: Migración a Base de Datos Relacional (SQLite)
# Pasaremos los DataFrames de la memoria RAM a un archivo `.db` permanente.

print("\n🗄️ FASE 3: Migrando DataFrames a SQLite...")

DB_NAME = 'parcial_2_cafeteria.db'

# Conectamos a la base de datos (se crea si no existe)
with sqlite3.connect(DB_NAME) as conn:

    # Usamos la función to_sql() de Pandas. 
    # 'if_exists="replace"' sobrescribe la tabla si ejecutamos la celda varias veces.
    # 'index=False' evita que se guarde la columna de índices (0, 1, 2...) de Pandas.

    df_productos.to_sql('productos', conn, if_exists='replace', index=False)
    df_clientes.to_sql('clientes', conn, if_exists='replace', index=False)
    df_proveedores.to_sql('proveedores', conn, if_exists='replace', index=False)

print("✅ Tablas 'productos', 'clientes' y 'proveedores' creadas exitosamente en la BD 'parcial_2_cafeteria.db'.")
```

```python
# # FASE 4: Arquitectura SQL y Operaciones CRUD
# Crearemos la tabla transaccional `ventas` con Llaves Foráneas y ejecutaremos el CRUD.

print("\n⚙️ FASE 4: Ejecutando Arquitectura SQL y CRUD...")

with sqlite3.connect(DB_NAME) as conn:
    cursor = conn.cursor()
    
    # 1. Habilitar llaves foráneas (Obligatorio en SQLite)
    cursor.execute("PRAGMA foreign_keys = ON;")
    
    # 2. CREACIÓN DE TABLA VENTAS (Con PK y FK)
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS ventas (
            id_venta INTEGER PRIMARY KEY AUTOINCREMENT,
            id_cliente INTEGER,
            id_producto INTEGER,
            nit_proveedor INTEGER,
            cantidad INTEGER NOT NULL,
            total_venta REAL NOT NULL,
            fecha_venta TEXT DEFAULT CURRENT_TIMESTAMP,
            FOREIGN KEY (id_cliente) REFERENCES clientes(id_cliente),
            FOREIGN KEY (id_producto) REFERENCES productos(id_producto),
            FOREIGN KEY (nit_proveedor) REFERENCES proveedores(nit_proveedor)
        )
    ''')

print ("✅ Tabla 'ventas' creada exitosamente con Llaves Foráneas.")    
```

```python
    # 3. CRUD: CREATE (Insertar 5 ventas)

with sqlite3.connect(DB_NAME) as conn:
    cursor = conn.cursor()

    ventas_data =[
        (101, 1, 900111, 2, 10000.0),  # Diego compra 2 Cafés
        (102, 2, 800222, 5, 6000.0),   # Ana compra 5 Chocolatinas
        (103, 3, 900111, 3, 9000.0),   # Carlos compra 3 Empanadas
        (104, 4, 700333, 1, 2500.0),   # Maria compra 1 Jugo Hit
        (105, 1, 900111, 4, 20000.0)   # Luis compra 4 Cafés
    ]

    cursor.executemany('''
        INSERT INTO ventas (id_cliente, id_producto, nit_proveedor, cantidad, total_venta)
        VALUES (?, ?, ?, ?, ?)
    ''', ventas_data)

    conn.commit()
    
    print("➕ [CREATE] 5 ventas registradas exitosamente.")
```

```python
    # 4. CRUD: UPDATE (Actualizar venta ID 1)

with sqlite3.connect(DB_NAME) as conn:
    cursor = conn.cursor()

    cursor.execute('''
        UPDATE ventas 
        SET cantidad = ?, total_venta = ? 
        WHERE id_venta = ?
    ''', (3, 15000.0, 1))
    conn.commit()
    print("🔄 [UPDATE] Venta ID 1 actualizada (Cantidad: 3, Total: $15,000).")
```

```python
    # 5. CRUD: DELETE (Eliminar venta ID 3)

with sqlite3.connect(DB_NAME) as conn:
    cursor = conn.cursor()

    cursor.execute("DELETE FROM ventas WHERE id_venta = ?", (3,))
    conn.commit()
    print("❌ [DELETE] Venta ID 3 eliminada del sistema.")
```

```python
    # 6. CRUD: READ (SELECT con INNER JOIN)

with sqlite3.connect(DB_NAME) as conn:
    cursor = conn.cursor()

    print("\n📊 REPORTE FINAL DE VENTAS (INNER JOIN):")
    print("="*70)
    
    consulta_join = '''
        SELECT 
            ventas.id_venta,
            clientes.Nombre_Cliente,
            productos.Nombre_Producto,
            ventas.cantidad,
            ventas.total_venta
        FROM ventas
        INNER JOIN clientes ON ventas.id_cliente = clientes.id_cliente
        INNER JOIN productos ON ventas.id_producto = productos.id_producto
        ORDER BY ventas.id_venta ASC
    '''
    cursor.execute(consulta_join)
    resultados = cursor.fetchall()
    
    for fila in resultados:
        print(f"Venta #{fila[0]} | Cliente: {fila[1].ljust(15)} | Producto: {fila[2].ljust(15)} | Cant: {fila[3]} | Total: ${fila[4]:,.2f}")
    print("="*70)

print("\n🎉 PARCIAL FINALIZADO CON ÉXITO.")
```

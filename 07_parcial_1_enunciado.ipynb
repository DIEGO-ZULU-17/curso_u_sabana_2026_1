**Su Misión:**
Así como en el Taller 1 construyeron el MVP (Producto Mínimo Viable) de una empresa en grupos, hoy demostrarán su capacidad técnica individual creando el motor lógico de **"FinVes"**, una App que permite a los colombianos registrar sus inversiones en bolsa, calcular sus ingresos pasivos y medir qué tan cerca están de su libertad financiera.

## 🛠️ REQUERIMIENTOS TÉCNICOS Y PASO A PASO

Su programa en Python debe cumplir exactamente con el siguiente flujo:

### PASO 1: Configuración de Estructuras de Datos
1. Cree un **Diccionario** llamado `perfil_usuario` que almacene: El nombre del usuario y su "Meta de Libertad Financiera" (cuánto dinero necesita al mes para vivir sin trabajar). *Estos datos deben ser solicitados al usuario con `input()` al iniciar el programa.*
2. Cree una **Tupla** llamada `activos_disponibles` que contenga exactamente estos 4 mercados autorizados: `("Bancolombia", "Ecopetrol", "S&P500", "Bonos del Estado")`.
3. Cree una **Lista** vacía llamada `portafolio` donde se guardarán las inversiones que el usuario registre.

### PASO 2: Menú Interactivo (`while`)
Implemente un ciclo `while` que mantenga el programa en ejecución mostrando 3 opciones:
*   **Opción 1:** Registrar nueva inversión.
*   **Opción 2:** Ver Estado del Portafolio y Libertad Financiera.
*   **Opción 3:** Salir de la App.

### PASO 3: Lógica de la Opción 1 (Registrar Inversión)
*   Muestre al usuario la tupla de `activos_disponibles` e invítelo a escribir en cuál desea invertir.
*   Pida el **Monto a Invertir** (en COP) y la **Tasa de Rendimiento Anual** esperada (Ej: 10 para 10%).
*   Guarde esta información estructurada (puede ser una sub-lista o un diccionario) dentro de la lista `portafolio`.

### PASO 4: Lógica de la Opción 2 (Análisis con `for` e `if/else`)
*   Si el `portafolio` está vacío, muestre un mensaje de advertencia y regrese al menú.
*   Si hay datos, use un bucle **`for`** para recorrer las inversiones.
*   Calcule el **Ingreso Pasivo Mensual Total**: *(Monto Invertido * (Tasa / 100)) / 12*. Sume el de todas las inversiones.
*   **La Decisión (`if/else`):** Compare el *Ingreso Pasivo Mensual Total* contra la *Meta de Libertad Financiera* guardada en el diccionario `perfil_usuario`.
    *   Si el ingreso pasivo supera la meta: Imprima *"¡Felicidades [Nombre]! Has alcanzado la Libertad Financiera."*
    *   Si no la supera: Calcule el porcentaje de avance `(Ingreso Pasivo / Meta) * 100` e imprima: *"Te falta un camino por recorrer. Tu avance es del X.XX%."*

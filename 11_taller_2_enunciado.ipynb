## 3. REQUERIMIENTOS TÉCNICOS (LA NUEVA ARQUITECTURA)

Su nuevo sistema debe cumplir estrictamente con los siguientes pilares:

1. **Programación Orientada a Objetos (POO):**

   * Deben crear al menos **tres clases padre** con **tres clases hijo** que representen las entidades de su negocio (Ej: `Cliente` y `Transaccion`, o `Ingrediente` y `Plato`).
   * Aplicar **Encapsulamiento**: Los atributos sensibles deben ser privados (`__atributo`) y accederse mediante métodos `get` y `set`.

2. **Persistencia de Datos (SQLite):**
   * El programa debe conectarse a un archivo `.db` (Ej: `base_datos_startup_grupo_#.db`).
   * Al iniciar, el código debe crear las tablas automáticamente si no existen (`CREATE TABLE IF NOT EXISTS`).

3. **Normalización y Relaciones (Llaves SQL):**
   * Deben existir al menos **tres tablas relacionadas**.
   * Uso obligatorio de Llaves Primarias (**PK**) y Llaves Foráneas (**FK**).
   * *Ejemplo:* Una tabla de `Clientes` y una tabla de `Ventas`, donde la venta guarde el `id_cliente`.

4. **Operaciones CRUD y Datos Mínimos:**
   * El Menú Interactivo (`while`) debe permitir: **C**rear, **L**eer, **A**ctualizar y **E**liminar registros de la base de datos.
   * **Regla de Oro:** Al entregar el proyecto, las tablas deben contener **al menos 5 registros** insertados previamente mediante código para poder auditar su funcionamiento.
   * Deben implementar al menos una consulta que una las dos tablas utilizando la cláusula **`JOIN`**.

5. **Manejo de Errores (Robustez):**
   * Implementar bloques `try-except` para evitar que el programa colapse si el usuario ingresa texto en lugar de números o si hay un error de conexión a la base de datos.

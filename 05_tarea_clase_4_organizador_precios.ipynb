# 🚀 Reto Opcional: El "Organizador de Precios" (Bono +0.5)

**Objetivo del Ejercicio:**
Consolidar el dominio sobre el manejo de colecciones (Listas) y la automatización mediante ciclos (`for`). Este ejercicio es fundamental para comprender la diferencia entre modificar una lista existente (mutabilidad) y generar una nueva a partir de cálculos iterativos.

**Instrucciones:**
Cree un script que realice las siguientes operaciones paso a paso:

1. Tenga una lista desordenada de precios: `[100, 20, 50, 80, 10]`.
2. Use el método `.sort()` para ordenarla.
3. Use un bucle **for** para aplicar el **IVA (19%)** a cada precio y guárdelos en una nueva lista.
4. Imprima la lista original y la lista con IVA.

---
*Nota de Autogestión: Recuerden documentar cada línea de su código utilizando `#` para hacer visible su proceso de razonamiento. Revisen la celda de código inferior para analizar la solución técnica propuesta.*

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# TEMA: Manipulación de Listas y Ciclos For
# ACTIVIDAD: El "Organizador de Precios" (Reto de Refuerzo)
# ==============================================================================

# 1. DEFINICIÓN DE DATOS
# Creamos la lista desordenada de precios base según el requerimiento.
precios_originales = [100, 20, 50, 80, 10] 
# Se usan llaves [] para definir una lista en Python. Los elementos están separados por comas.
# Se usan llaves {} para definir un diccionario, que es una estructura de datos diferente (clave-valor).
# Se usan llaves () para definir una tupla, que es una estructura de datos inmutable (no se puede modificar después de creada).

# 2. ORDENAMIENTO
# Utilizamos el método .sort() para organizar la lista de menor a mayor.
# Nota: Este método modifica la lista original directamente (in-place).
precios_originales.sort()

# 3. PROCESAMIENTO DE DATOS (IVA)
# Definimos la tasa de IVA como una constante (19%).
IVA = 0.19

# Creamos una lista vacía para almacenar los nuevos precios con el impuesto aplicado.
precios_con_iva = []

# Iniciamos un bucle 'for' para recorrer cada elemento de la lista ya ordenada.
for precio in precios_originales:
    # Calculamos el precio final (Precio Base + 19% de IVA).
    precio_final = precio * (1 + IVA)
    
    # Agregamos el resultado a nuestra nueva lista usando el método .append().
    precios_con_iva.append(precio_final)

# 4. SALIDA DE RESULTADOS (OUTPUT)
# Imprimimos ambas listas para comparar la transformación de los datos.
print("--- REPORTE DE PRECIOS ---")
print(f"Lista Original (Ordenada): {precios_originales}")
print(f"Lista con IVA (19%):       {[f'{p:.2f}' for p in precios_con_iva]}") # Formateamos los precios con dos decimales.
print("--------------------------")

# Adicionalmente, mostramos una lista detallada que relaciona cada precio original con su correspondiente precio con IVA.
print("Lista Detallada con IVA (19%):")
for precio_orig, precio_iva in zip(precios_originales, precios_con_iva): 
    # La función zip() empareja elementos de ambas listas posición por posición, 
    # permitiendo acceder al precio original y al precio con IVA en cada iteración del bucle.
    print(f"  {precio_orig} * {IVA} = ${precio_iva:.2f}")
```

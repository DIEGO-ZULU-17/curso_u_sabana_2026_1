# Universidad de La Sabana - EICEA
## Programación y Decisiones 2026-1
### Solución Ideal - Parcial 1: Motor Lógico "FinVes"

---

### 1. PSEUDOCÓDIGO (Diseño Lógico)

1. INICIO

2. Solicitar y limpiar el nombre del usuario.

3. MIENTRAS la meta financiera no sea un número válido:

   3.1 INTENTAR capturar la meta financiera.

   3.2 SI hay error (letras), mostrar advertencia y repetir.

4. CREAR DICCIONARIO `perfil_usuario` con el nombre y la meta.

5. CREAR TUPLA `activos_disponibles` con los 4 mercados.

6. CREAR LISTA vacía `portafolio`.

7. MIENTRAS el sistema esté activo (Menú Principal):

   7.1 MOSTRAR opciones (1. Registrar, 2. Ver Estado, 3. Salir).

   7.2 LEER opción.

   7.3 SI opción es 1:

       7.3.1 MOSTRAR `activos_disponibles`.

       7.3.2 LEER activo elegido.

       7.3.3 MIENTRAS el monto y la tasa no sean válidos:

             7.3.3.1 INTENTAR capturar monto y tasa.

             7.3.3.2 SI hay error, mostrar advertencia y repetir.

       7.3.4 GUARDAR inversión como diccionario en la lista `portafolio`.

   7.4 SI opción es 2:

       7.4.1 SI `portafolio` está vacío, MOSTRAR advertencia.

       7.4.2 SI NO:

             7.4.2.1 INICIALIZAR `ingreso_pasivo_total` en 0.

             7.4.2.2 PARA CADA inversión en `portafolio`:

                     7.4.2.2.1 CALCULAR pasivo mensual.

                     7.4.2.2.2 SUMAR al `ingreso_pasivo_total`.

             7.4.2.3 SI `ingreso_pasivo_total` >= meta del usuario:

                     7.4.2.3.1 IMPRIMIR mensaje de Libertad Financiera alcanzada.

             7.4.2.4 SI NO:

                     7.4.2.4.1 CALCULAR porcentaje de avance.

                     7.4.2.4.2 IMPRIMIR mensaje con el porcentaje.

   7.5 SI opción es 3:

       7.5.1 TERMINAR el programa (Romper ciclo).

8. FIN

```python
# ==============================================================================
# MOTOR LÓGICO FINVES - SOLUCIÓN IDEAL Y ROBUSTA
# ==============================================================================

print("🚀 BIENVENIDO A FINVES - TU CAMINO A LA LIBERTAD FINANCIERA 🚀\n")

# ==============================================================================
# PASO 1: CONFIGURACIÓN DE ESTRUCTURAS DE DATOS Y PERFIL
# ==============================================================================

# Usamos .strip() para quitar espacios accidentales y .title() para estandarizar mayúsculas.
# Captura y valida que el nombre tenga solo letras
while True:
    try:
        nombre_usuario = input("Ingresa tu nombre: ").strip().title()
        # Quitamos espacios para la comprobación y verificamos que todo sean letras
        if not nombre_usuario.replace(" ", "").isalpha():
            raise ValueError("El nombre solo debe contener letras.")
        print(f"\n Bienvenido a FINVES {nombre_usuario} \n")
        break
    except ValueError as e:
        print(f"❌ {e} Ingresa nuevamente el nombre, por favor.")

# BUCLE DE VALIDACIÓN: Garantiza que el programa no avance hasta tener un número válido.
while True:
    try:
        # Intentamos convertir la entrada a float (Casting)
        meta_input = float(input("Ingresa tu Meta de Libertad Financiera mensual (COP): "))
        # Mostrar la meta en el formato correcto.
        print(f"{nombre_usuario}. Tu meta de libertad financiera es $ {meta_input:,.0f} COP".replace(",", "."))
        
        # Sanity Check: Una meta no puede ser negativa ni cero
        if meta_input <= 0:
            print("⚠️ La meta debe ser mayor a $0. Intenta de nuevo.")
            continue # Vuelve al inicio del while
            
        break # Si todo es correcto, rompemos el bucle de validación y avanzamos.
        
    except ValueError:
        # Si el usuario escribe "cinco millones" en letras, Python salta aquí en vez de colapsar.
        print("❌ Error: Ingresa solo números (sin letras, puntos ni signos de pesos).")

# 1.1 DICCIONARIO: Almacena las propiedades del usuario.
perfil_usuario = {
    "nombre": nombre_usuario,
    "meta": meta_input
}

# 1.2 TUPLA: Colección inmutable. Los mercados autorizados no deben cambiar durante la ejecución.
activos_disponibles = ("Bancolombia", "Ecopetrol", "S&P500", "Bonos")

# 1.3 LISTA: Colección mutable y vacía para ir agregando inversiones dinámicamente.
portafolio =[]


# ==============================================================================
# PASO 2: MENÚ INTERACTIVO (CICLO WHILE PRINCIPAL)
# ==============================================================================
while True:
    print("\n" + "="*50)
    print("📊 MENÚ PRINCIPAL FINVES 📊")
    print("="*50)
    print("1. Registrar nueva inversión")
    print("2. Ver Estado del Portafolio y Libertad Financiera")
    print("3. Salir de la App")
    print("\n" + "="*50)
    
    # .strip() evita que un espacio accidental (" 1") dañe la validación del if.
    opcion = input("\n>> Selecciona una opción (1, 2 o 3): ").strip()
    
    # ==========================================================================
    # PASO 3: LÓGICA DE LA OPCIÓN 1 (REGISTRAR INVERSIÓN)
    # ==========================================================================
    if opcion == "1":
        print("\n--- MERCADOS AUTORIZADOS ---")
        
        # Recorremos la tupla para mostrar los activos disponibles
        for activo in activos_disponibles:
            print(f"- {activo}")
            
        activo_elegido = input("\nEscribe el nombre del activo en el que deseas invertir: ").strip().title()
        
        # BUCLE DE VALIDACIÓN PARA NOMBRE DEL ACTIVO: Blindamos la entrada de dinero.
        # validamos que el nombre del activo exista exactamente en la tupla
        while activo_elegido not in activos_disponibles:
            print("Ingresa el nombre de un activo disponible, por favor.")
            activo_elegido = input("Escribe el nombre del activo en el que deseas invertir: ").strip().title()

        # BUCLE DE VALIDACIÓN PARA MONTOS: Blindamos la entrada de dinero.
        while True:
            try:
                monto = float(input("Ingresa el monto a invertir en COP: "))
                tasa = float(input("Ingresa la Tasa de Rendimiento Anual esperada (Ej: 10 para 10%): "))
                
                if monto <= 0 or tasa <= 0:
                    print("⚠️ El monto y la tasa deben ser mayores a 0.")
                    continue
                    
                break  # datos numéricos correctos
                
            except ValueError:
                print("❌ Error de formato: Ingresa solo valores numéricos.")
        
        # Guardamos la inversión como un diccionario dentro de la lista.
        nueva_inversion = {
            "activo": activo_elegido,
            "monto": monto,
            "tasa": tasa
        }
        
        portafolio.append(nueva_inversion)
        print(f"\n ✅ {nombre_usuario}, se ha registrado tu inversión en el activo {activo_elegido} "
              f"por un valor de ${monto:,.0f} COP y una tasa de rentabilidad anual de {tasa}%.")

    # ==========================================================================
    # PASO 4: LÓGICA DE LA OPCIÓN 2 (ANÁLISIS Y MATEMÁTICAS)
    # ==========================================================================
    elif opcion == "2":
        print("\n--- ESTADO DEL PORTAFOLIO ---")
        
        # Validación: Si la lista está vacía (longitud 0), no hay nada que calcular.
        if len(portafolio) == 0:
            print("⚠️ Aún no tienes inversiones registradas. Ve a la opción 1.")
            continue # Regresa al menú principal
            
        # Acumulador para sumar los ingresos de todas las inversiones
        ingreso_pasivo_total = 0.0
        
        # Recorremos la lista de diccionarios
        for inversion in portafolio:
            # Extraemos los valores usando las claves del diccionario
            monto_inv = inversion["monto"]
            tasa_inv = inversion["tasa"]
            nombre_activo = inversion["activo"]
            
            # Fórmula financiera: (Monto * (Tasa / 100)) / 12 meses
            pasivo_mensual = (monto_inv * (tasa_inv / 100)) / 12
            
            # Sumamos al acumulador global
            ingreso_pasivo_total += pasivo_mensual
            
            # Imprimimos el detalle de cada activo con formato de moneda
            print(f"🔸 {nombre_activo}: Generando ${pasivo_mensual:,.2f} COP mensuales.")
        
        print("-" * 50)
        print(f"💰 Ingreso Pasivo Mensual Total: ${ingreso_pasivo_total:,.2f} COP")
        print(f"🎯 Tu Meta Mensual:              ${perfil_usuario['meta']:,.2f} COP")
        print("-" * 50)
        
        # TOMA DE DECISIÓN: ¿Alcanzó la meta?
        if ingreso_pasivo_total >= perfil_usuario["meta"]:
            print(f"🎉 ¡Felicidades {perfil_usuario['nombre']}! Has alcanzado la Libertad Financiera.")
            print("Tus inversiones pagan tu estilo de vida.")
        else:
            # Regla de tres simple para sacar el porcentaje
            porcentaje_avance = (ingreso_pasivo_total / perfil_usuario["meta"]) * 100
            print(f"🚀 Te falta un camino por recorrer. Tu avance es del {porcentaje_avance:.2f}%.")

    # ==========================================================================
    # SALIDA DEL SISTEMA
    # ==========================================================================
    elif opcion == "3":
        print(f"\nSaliendo de FinVes... ¡El interés compuesto es tu mejor amigo, {perfil_usuario['nombre']}! 👋")
        break # Rompe el ciclo principal y apaga el programa
        
    # ==========================================================================
    # MANEJO DE ERRORES DEL MENÚ
    # ==========================================================================
    else:
        print("❌ Opción no válida. Por favor digita 1, 2 o 3.")
```

Este es un ejercicio de refuerzo diseñado para consolidar su dominio sobre los ciclos while y la lógica de condicionales anidados. 

El objetivo es simular el protocolo de seguridad de una aplicación bancaria real.

## Requisitos Técnicos del Script:

Seguridad: Definir una variable contrasena_correcta.

Control de Intentos: Implementar un bucle while que limite el acceso a un máximo de 3 intentos.

Interacción: Solicitar la clave mediante input().

Lógica de Éxito: Si la clave es correcta, mostrar "Bienvenido" y terminar el ciclo inmediatamente con break.

Lógica de Fallo: Si es incorrecta, informar al usuario cuántos intentos le quedan.

Bloqueo: Al agotar las oportunidades, mostrar el mensaje "Cuenta Bloqueada".

## Incentivo de Autogestión:

```python
# ==============================================================================
# UNIVERSIDAD DE LA SABANA - EICEA
# ASIGNATURA: Programación y Decisiones
# TEMA: Ciclo While y Control de Intentos
# RETO: Sistema de Login Seguro
# ==============================================================================

# 1. DEFINICIÓN DE VARIABLES DE CONFIGURACIÓN
# Establecemos la contraseña maestra que el sistema debe validar.
contrasena_correcta = "Sabana2026"

# Definimos el límite de oportunidades permitidas.
max_intentos = 3

# Inicializamos un contador para llevar el registro de los intentos fallidos.
intentos_realizados = 0

# 2. ESTRUCTURA DE CONTROL (BUCLE WHILE)
# El ciclo se repetirá mientras el contador sea menor a 3 (0, 1, 2).
while intentos_realizados < max_intentos:
    
    # 3. INTERACCIÓN CON EL USUARIO
    # Solicitamos la clave. El valor ingresado siempre entra como String.
    entrada_usuario = input("🔐 Ingrese su contraseña de acceso: ")

    # 4. LÓGICA DE VALIDACIÓN (IF / ELSE)
    if entrada_usuario == contrasena_correcta:
        # CASO DE ÉXITO:
        print("✅ Acceso concedido. Bienvenido al sistema.")
        
        # El comando 'break' es vital: detiene el bucle inmediatamente al acertar.
        break
    
    else:
        # CASO DE FALLO:
        # Incrementamos el contador en 1 por cada error cometido.
        intentos_realizados += 1
        
        # Calculamos los intentos que le quedan al usuario para informar.
        intentos_restantes = max_intentos - intentos_realizados
        
        # Verificamos si aún tiene intentos disponibles.
        if intentos_restantes > 0:
            print(f"❌ Contraseña incorrecta. Le quedan {intentos_restantes} intentos.")
        else:
            # 5. LÓGICA DE BLOQUEO
            # Si el contador llegó al máximo, mostramos el mensaje de bloqueo.
            print("🚫 Cuenta Bloqueada. Por seguridad, su acceso ha sido restringido.")

# 6. CIERRE DEL PROGRAMA
print("--- Sesión finalizada ---")
```

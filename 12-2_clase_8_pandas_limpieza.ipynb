```python
# Usamos import pandas as pd para trabajar con pandas.
# pd significa que cada vez que queramos usar una función de pandas, 
# lo haremos a través de pd, por ejemplo: pd.read_csv() o pd.DataFrame().

# os lo usamos para manejar archivos y directorios, como crear carpetas o escribir archivos.

import pandas as pd
import os

# ==============================================================================
# SETUP: CREACIÓN DE ARCHIVOS SUCIOS PARA LA CLASE (Simulación)
# ==============================================================================
# Creamos un CSV "afuera" de la carpeta actual y un Excel "adentro"

# os.makedirs lo usamos para crear la carpeta 'datos_externos' si no existe, donde guardaremos el CSV sucio.
# La carpeta 'datos_externos' se ubica un nivel arriba de la carpeta actual, es decir, en el directorio padre.

os.makedirs('../datos_externos', exist_ok=True)

csv_data = """id,nombre_completo,email,edad,fecha_nacimiento
1,diego zuluaga,diego@correo.com,,1985-05-10
2,ANA GOMEZ,ana@correo.com,22,2004-01-15
3,carlos ruiz,,35,1989-08-20
4,maria lopez,maria@correo.com,NaN,1995-11-30"""

# Escribimos el CSV sucio en la carpeta 'datos_externos' con codificación utf-8.
with open('../datos_externos/clientes_sucios.csv', 'w', encoding='utf-8') as f:
    f.write(csv_data)

# Usamos pd.DataFrame para crear un DataFrame con datos sucios 
# y luego lo guardamos como un archivo Excel dentro de la carpeta actual.

excel_data = pd.DataFrame({
    'ID_Prod': [1, 2, 3],
    'Nombre_Categoria':['Café Tostao - Bebida', 'Chocolatina Jet - Snack', 'Empanada - Snack'],
    'Precio':['$5000', '1200', 'NaN']
})

# Usamos .to_excel para guardar el DataFrame como un archivo Excel llamado 'productos_sucios.xlsx' sin incluir el índice.
excel_data.to_excel('productos_sucios.xlsx', index=False)

print("✅ Archivos sucios generados para la práctica.")
```

## 1. Carga de Archivos y Data Wrangler

# Vamos a cargar los archivos usando rutas relativas y absolutas.

*Tip: Abran la extensión "Data Wrangler" en VS Code para ver el DataFrame.*


```python
# FORMAS DE CARGAR LOS ARCHIVOS CON PANDAS

# Ruta Externa (Sube un nivel de carpeta con '..')
df_clientes = pd.read_csv('../datos_externos/clientes_sucios.csv')

# Ruta Absoluta (Ruta completa desde la raíz del sistema)
# Se usa r para indicar que es una cadena raw y evitar problemas con las barras invertidas en Windows.
df_clientes = pd.read_csv(r"C:\Users\diego\OneDrive\Documentos\Clase_Python_U_Sabana\datos_externos\clientes_sucios.csv")

# Ruta Interna (Misma carpeta)
df_productos = pd.read_excel('productos_sucios.xlsx')

print("--- CLIENTES SUCIOS ---")
print(df_clientes)
print("\n--- PRODUCTOS SUCIOS ---")
print(df_productos)
```

# 2. Limpieza de Datos (Data Cleaning)

# Corrigiendo inconsistencias de texto y valores nulos.

```python
# Limpiando Clientes

# 1. Estandarizar nombres (Title Case)
df_clientes['nombre_completo'] = df_clientes['nombre_completo'].str.title()
print(df_clientes)

```

```python
# 2. Rellenar emails nulos con un texto por defecto
df_clientes['email'] = df_clientes['email'].fillna('sin_correo@empresa.com')

print("--- CLIENTES LIMPIOS ---")
print(df_clientes)
```

```python
# Limpiando Productos

print("--- PRODUCTOS LIMPIOS ---")
print("\n")

# 1. Quitar el signo '$' del precio y convertir a número (float)

# .astype(str).str se usa para convertir la columna a string 
# y luego aplicar el método str.replace() para eliminar el símbolo '$'.

df_productos['Precio'] = df_productos['Precio'].astype(str).str.replace('$', '')
print(df_productos)
print("\n")

# errors='coerce' se usa para convertir a numérico 
# y convertir cualquier valor no convertible (como 'NaN') en un valor nulo real (NaN de pandas).
df_productos['Precio'] = pd.to_numeric(df_productos['Precio'], errors='coerce') # Convierte 'NaN' a nulo real
print(df_productos)
print("\n")

# .fillna(0) se usa para reemplazar los valores nulos (NaN) con 0, 
# lo que es útil para evitar problemas en cálculos posteriores.
df_productos['Precio'] = df_productos['Precio'].fillna(0) # Rellenar nulos con 0
print(df_productos)
print("\n")
```

# 3. Normalización con Pandas (1NF, 2NF, 3NF)

```python
# 1NF (Atomicidad): En Productos, la columna 'Nombre_Categoria' tiene 2 valores. Hay que separarlos.

# Se usa .str.split() para dividir la columna 'Nombre_Categoria' en dos nuevas columnas 'Nombre' y 'Categoria',
# usando ' - ' como separador. expand=True se usa para expandir el resultado en columnas separadas.

df_productos[['Nombre', 'Categoria']] = df_productos['Nombre_Categoria'].str.split(' - ', expand=True)
print("--- 1NF: PRODUCTOS NORMALIZADOS ---")
print(df_productos)
print("\n")

# .drop se usa para eliminar la columna original 'Nombre_Categoria' que ya no es necesaria después de la separación.
df_productos = df_productos.drop(columns=['Nombre_Categoria']) # Borramos la columna original sucia

print("--- 1NF: PRODUCTOS NORMALIZADOS ---")
print(df_productos)
```

```python
# 3NF (Dependencia Transitiva): En Clientes, tenemos 'edad' y 'fecha_nacimiento'. 

# La edad depende de la fecha, es redundante. La eliminamos con .drop().
df_clientes = df_clientes.drop(columns=['edad'])

print("\n--- 3NF: CLIENTES NORMALIZADOS ---")
print(df_clientes)
```

# 4. CRUD: Pandas vs SQL

# Cómo hacer operaciones de base de datos directamente en la memoria RAM.

```python
# CREATE (Insertar fila)

# Usamos loc para insertar en el último índice

nuevo_cliente =[5, 'Luis Perez', 'luis@correo.com', '2000-05-05']

df_clientes.loc[len(df_clientes)] = nuevo_cliente

print("\n--- CLIENTES CON NUEVO REGISTRO ---")
print(df_clientes)
```

```python
# READ (Seleccionar con filtros - SELECT * WHERE)
clientes_bogota = df_clientes[df_clientes['nombre_completo'] == 'Ana Gomez']

print("\n--- CLIENTES EN BOGOTÁ ---")
print(clientes_bogota)
```

```python
# UPDATE (Actualizar un valor específico)

# Cambiamos el email del cliente con ID 1 (índice 0)

df_clientes.loc[df_clientes['id'] == 1, 'email'] = 'diego.zuluaga@unisabana.edu.co'

print("\n--- CLIENTES CON EMAIL ACTUALIZADO ---")
print(df_clientes)
```

```python
# DELETE (Eliminar fila)

# Eliminamos el índice 2 (Carlos Ruiz)
df_clientes = df_clientes.drop(2)

print("--- DATAFRAME DESPUÉS DEL CRUD ---")
print(df_clientes)
```

# Importar los csv con SQL

```python
import sqlite3

# Conectamos a una base de datos SQLite en archivo
conn = sqlite3.connect('datos_limpios.db')

# Leemos los CSV ya generados
df_productos_sql = pd.read_csv('productos_limpios.csv')
df_clientes_sql = pd.read_csv('clientes_limpios.csv')

# Cargamos los DataFrames a tablas SQLite
df_productos_sql.to_sql('productos', conn, if_exists='replace', index=False)
df_clientes_sql.to_sql('clientes', conn, if_exists='replace', index=False)

# Verificamos que se cargaron correctamente con consultas SQL
print("--- PRODUCTOS DESDE SQLITE ---")
print(pd.read_sql_query('SELECT * FROM productos', conn))

print("\n--- CLIENTES DESDE SQLITE ---")
print(pd.read_sql_query('SELECT * FROM clientes', conn))

conn.close()
```

# Insertar datos nuevos usando SQL

```python
conn = sqlite3.connect('datos_limpios.db')
cursor = conn.cursor()

# Insertar un nuevo producto
cursor.execute(
    "INSERT INTO productos (ID_Prod, Precio, Nombre, Categoria) VALUES (?, ?, ?, ?)",
    (4, 2500.0, 'Jugo Hit', 'Bebida')
)

# Insertar un nuevo cliente
cursor.execute(
    "INSERT INTO clientes (id, nombre_completo, email, fecha_nacimiento) VALUES (?, ?, ?, ?)",
    (6, 'Sofía Martínez', 'sofia@correo.com', '1998-09-20')
)

conn.commit()

print("--- NUEVO PRODUCTO INSERTADO ---")
print(pd.read_sql_query("SELECT * FROM productos WHERE ID_Prod = 4", conn))

print("\n--- NUEVO CLIENTE INSERTADO ---")
print(pd.read_sql_query("SELECT * FROM clientes WHERE id = 6", conn))

conn.close()
```

# Guardar la tabla actualizada en nuevos archivos csv

# 5. POO + Pandas (Arquitectura)

# Encapsulando Pandas dentro de una clase de negocio.

```python
class GestorDatosCafeteria:
    def __init__(self, ruta_csv):
        self.ruta = ruta_csv
        self.df = pd.read_csv(self.ruta)
        
    def limpiar_nombres(self):
        self.df['nombre_completo'] = self.df['nombre_completo'].str.title()
        
    def obtener_cliente(self, id_cliente):
        return self.df[self.df['id'] == id_cliente]

# Uso de la clase

gestor = GestorDatosCafeteria('../datos_externos/clientes_sucios.csv')

gestor.limpiar_nombres()

print("\n--- BÚSQUEDA DESDE POO ---")
print(gestor.obtener_cliente(1))
```
